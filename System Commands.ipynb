## 1. Why a Linux Virtual Machine?

### The Need for a Virtual Machine (VM)

Modern laptops often come with Windows pre‑installed. Removing it to install Linux directly (a “bare‑metal” installation) means losing the license you paid for. Dual‑booting—where both operating systems coexist and you choose one at boot time—can be blocked by UEFI/BIOS restrictions tied to the Original Equipment Manufacturer (OEM). A virtual machine solves these problems:

- You keep the original Windows installation untouched.
- You run a full Linux operating system inside a window, almost as if it were natively installed.
- You can experiment, learn, and even break the Linux environment without risk to the host system.
- You can take snapshots of the VM state and revert if something goes wrong.

### Requirements to Run a Linux VM

- **ISO image of a Linux distribution**: The lectures recommend **Ubuntu 20.04 LTS** (Long‑Term Support). LTS versions are stable, receive security updates for five years, and are ideal for learning. The image is about 3 GB and is downloaded from [ubuntu.com](https://ubuntu.com).
- **A hypervisor**: This is the software that creates and runs virtual machines. Two free options are demonstrated:
  - Oracle VirtualBox
  - VMware Workstation Player
- **Sufficient disk space**: A VM typically needs a virtual hard disk of 20 GB or more.
- **Sufficient RAM**: The host machine should have at least 8 GB of RAM so that the VM can be allocated 2–4 GB without crippling the host. On a machine with only 4 GB, running a VM is not recommended.

### The Concept of a Hypervisor

A hypervisor is a layer of software that sits between the physical hardware and the virtual machines. It emulates a complete computer (CPU, memory, storage, network) in software. The “host” is the real machine (Windows), and the “guest” is the Linux VM. Modern CPUs have hardware extensions (Intel VT‑x, AMD‑V) that the hypervisor uses to make the guest run nearly as fast as the host.

### Installation Walkthrough

#### VirtualBox
1. Download the VirtualBox installer for Windows from [virtualbox.org](https://www.virtualbox.org).
2. Run the installer as Administrator. The installation creates a desktop icon.
3. Launch VirtualBox and click **New** to create a VM. Give it a name (e.g., “Ubuntu 20.04”).
4. Allocate RAM (at least 2048 MB).
5. Create a virtual hard disk. The default is 10 GB, but 20 GB is recommended. The disk can be dynamically allocated, so it only takes up space on the host as data is written.
6. Before starting the VM, go to **Settings → Storage**, click the empty optical drive, and choose the Ubuntu ISO file you downloaded.
7. Start the VM. It boots from the ISO and begins the Ubuntu installer.
8. Choose “Install Ubuntu”, select normal installation, and when asked, “Erase disk and install Ubuntu”. **Important**: This erases the *virtual* disk, not the host’s real disk.
9. Select time zone (India), create the first user account with a password. This account will have administrator privileges via `sudo`.
10. After installation, reboot the VM and log in. Open the terminal (Ctrl+Alt+T). You now have a full Ubuntu system.

#### VMware Workstation Player
1. Download the free VMware Workstation Player from VMware’s website.
2. Install as Administrator. Launch the player, choose “Use for free for non‑commercial use”.
3. Choose “Create a New Virtual Machine” and point to the Ubuntu ISO.
4. Enter the user credentials, specify disk size (20 GB), and choose to store the virtual disk as multiple files (easier to move).
5. The VM boots, checks the installation medium, and installs Ubuntu automatically. During installation, a slideshow of features is shown. After a reboot, you log in and can add the terminal to favorites.

#### Windows Subsystem for Linux (WSL)
WSL is a Microsoft‑developed compatibility layer that allows a genuine Linux kernel and user‑space to run directly on Windows, without a full virtual machine. It is lightweight and integrated.

- Open PowerShell as Administrator and run:  
  `wsl --install`
- This installs the WSL feature, the Linux kernel, and downloads Ubuntu by default.
- After reboot, the Ubuntu setup asks for a new UNIX username and password.
- You can then start WSL by typing `wsl` or launching “Ubuntu” from the Start menu. This gives a bash prompt that behaves like a real Linux environment, with access to the Windows filesystem via `/mnt/c/`.

#### Cloud‑based Alternatives
If you cannot install anything locally, cloud platforms offer instant Linux terminals in a browser:

- **Replit.com**: Log in with Google/GitHub, create a “Bash” project, and a shell is provided running a Linux kernel (5.11). Files persist between sessions.
- **CoCalc.com**: Similar; sign in, create a project, start a terminal. You can install additional packages. The free tier has limited resources.

#### Mobile Solution: Termux
Termux is an Android app (by Fredrik Fornwall) that provides a minimal Linux user‑space in a terminal. It does not require root. It uses a kernel provided by Android (4.14 in the demo) but gives you a bash shell, package manager (`pkg`), and the ability to practice command‑line exercises on a phone.

## 2. The Command Line Environment

### Terminal Emulators
A terminal emulator is a GUI application that simulates a physical terminal (like the old VT100). It provides a window where you type commands and see output. Different emulators offer various features:

- **Gnome Terminal (default “Terminal” app in Ubuntu)**: Supports multiple tabs, custom fonts/colours, transparency, and keyboard shortcuts. It is the one used in most demos.
- **xterm**: The classic, minimalist terminal. Few customisation options but very lightweight.
- **Guake**: A drop‑down terminal, toggled by pressing F12. It stays in the background and slides down when needed.
- **Konsole**: The terminal for the KDE Plasma desktop, similar features to Gnome Terminal.

The terminal emulator runs a **shell** inside it. A shell is a program that reads commands, interprets them, and communicates with the operating system. In the lectures, the shell is **bash** (Bourne Again SHell), the default on Ubuntu.

### The Command Prompt
When you open a terminal, you see something like:  
`username@hostname:~$ `

This string is called the **command prompt**, defined by the environment variable `PS1`. It can be customised. The components are:
- `username` – the logged‑in user.
- `@` – separator.
- `hostname` – the name of the machine.
- `:` – separator.
- `~` – short for the user’s home directory (e.g., `/home/username`). If you change directory, it shows the current path, e.g., `/usr/bin`.
- `$` – indicates you are a normal user. The root user’s prompt typically ends with `#`.

The shell waits for you to type a command and press Enter. It then executes the command, displays output, and shows the prompt again.

### The First Commands

- **`pwd`** (print working directory): Tells you where you are in the filesystem. After login, this is `/home/yourname`. The filesystem is a tree starting from `/` (root). Your home directory is your personal space where you have full read/write permissions.
- **`ls`** (list): Lists the contents of the current directory. Without options, it shows just the names. Common files/folders in a fresh Ubuntu home include `Desktop`, `Documents`, `Downloads`, `Music`, `Pictures`, `Videos`.
- **`ps`** (process status): Shows the processes running in the current shell. Typically you’ll see `bash` and the `ps` command itself. More on `ps` later.
- **`uname`** (Unix name): Prints system information. `uname` alone gives “Linux”. `uname -a` prints all details: kernel version, architecture, etc. In the demo, the kernel is 5.11.
- **`clear`**: Clears the terminal screen. Also achieved by Ctrl+L.
- **`exit`**: Closes the shell. Alternatively, Ctrl+D sends an end‑of‑file (EOF) marker, which also causes the shell to exit.

### Commands, Options, and Arguments
A Linux command typically has the structure:

```
command [options] [arguments]
```

- **Command**: The executable or built‑in you want to run (e.g., `ls`).
- **Options**: Modify the behaviour of the command. Short options start with a hyphen (e.g., `-a`, `-l`). Long options start with two hyphens (e.g., `--all`, `--long`). Options can often be combined: `ls -la` is the same as `ls -l -a`.
- **Arguments**: The targets of the command, often file or directory names (e.g., `ls /etc`).

Example: `ls -l /home`  
Lists the contents of `/home` in the long format.

### Getting Help

- **`man`** (manual): The built‑in manual system. `man ls` displays the manual page for `ls`. Manual pages are divided into sections:
  - Section 1: User commands (shell commands)
  - Section 2: System calls
  - Section 3: Library functions
  - Section 4: Special files (devices)
  - Section 5: File formats and conventions
  - Section 6: Games
  - Section 7: Miscellaneous
  - Section 8: System administration commands

  To open a specific section, use `man 1 ls`. If a command exists in multiple sections, specifying the section avoids ambiguity. For example, `passwd` is both a command (section 1) and a file format (section 5). Inside man, press `q` to quit, `/` to search, `n` for next match.

- **`whatis`**: Displays a one‑line description of a command taken from the manual page’s NAME section.  
  Example: `whatis ls` → `ls (1) - list directory contents`

- **`apropos`** (or `man -k`): Searches the manual page descriptions for a keyword. Useful when you don’t remember the exact command.  
  Example: `apropos calendar` might return `cal`, `ncal`, etc.

- **`info`**: An alternative hypertext documentation system. Some commands (especially GNU tools) have more comprehensive info pages than man pages. Navigating info uses keys like `n` (next node), `p` (previous), `u` (up), `q` (quit).

- **`type`**: Tells you what kind of command something is – a shell built‑in, an alias, or an external executable.  
  Example: `type ls` → `ls is aliased to 'ls --color=auto'`  
  Example: `type cd` → `cd is a shell builtin`

- **`which`**: Shows the full path of an executable that the shell would run.  
  Example: `which python3` → `/usr/bin/python3`

- **`help`**: In bash, `help` displays a list of shell built‑ins. For help on a specific built‑in, use `help cd`.

### The Linux Filesystem Hierarchy Standard (FHS)

The FHS version 3.0 defines the directory structure and contents of Linux distributions. Understanding it is crucial because:

- You can locate configuration files, logs, libraries, and executables.
- You can navigate efficiently.
- You can troubleshoot compilation errors (missing headers, libraries).

All paths start at **`/`** (root). The forward slash is also used as a separator between directories.

#### Key Top‑Level Directories

| Directory | Purpose and Details |
|-----------|---------------------|
| **`/bin`** | Essential command binaries needed in single‑user mode (e.g., `ls`, `cat`, `cp`). On modern systems, often a symbolic link to `/usr/bin`. |
| **`/boot`** | Files for booting: the Linux kernel (`vmlinuz-...`), initial ramdisk (`initrd.img-...`), and bootloader configuration (GRUB). |
| **`/dev`** | Device files. In Linux, almost everything is represented as a file. Hard disks appear as `/dev/sda`, terminals as `/dev/tty`, etc. File types include character devices (e.g., keyboard, serial ports, denoted by ‘c’ in `ls -l`) and block devices (e.g., hard disks, denoted by ‘b’). |
| **`/etc`** | Host‑specific system configuration files. Examples: `/etc/passwd` (user accounts), `/etc/fstab` (filesystem mount table), `/etc/hosts` (static hostname‑to‑IP mapping), `/etc/profile` (global shell setup). |
| **`/lib`** | Essential shared libraries and kernel modules. Libraries are needed by programs in `/bin` and `/sbin`. Files usually have `.so` extensions with version numbers, e.g., `libc.so.6`. |
| **`/media`** | Mount points for removable media like USB drives, CD‑ROMs. The system automatically creates subdirectories here when you plug in a device. |
| **`/mnt`** | Temporary mount point for administrators to manually mount filesystems. |
| **`/opt`** | Add‑on application software packages. Often used by third‑party software (e.g., Google Chrome, proprietary tools). |
| **`/run`** | Runtime variable data. A tmpfs (memory‑based filesystem) that stores information about the running system since last boot, such as process IDs (`.pid` files) and sockets. |
| **`/sbin`** | Essential system administration binaries (e.g., `fdisk`, `mkfs`, `mount`). Like `/bin`, often merged into `/usr/sbin` via a symlink. |
| **`/srv`** | Data for services provided by the system (e.g., web server data might go in `/srv/www`). |
| **`/tmp`** | Temporary files. All users can write here, but files may be deleted on reboot (often done by a cleanup script). |
| **`/usr`** | Secondary hierarchy for read‑only, shareable user data. It contains the bulk of the operating system. It has its own sub‑hierarchy (see below). |
| **`/var`** | Variable data that changes during normal operation, such as logs, spool files, databases. Persists across reboots. |

#### `/usr` Hierarchy (deep dive)
- **`/usr/bin`**: The majority of user commands not needed for booting (e.g., `python`, `gcc`, `firefox`).
- **`/usr/lib`**: Libraries for programs in `/usr/bin` and `/usr/sbin`.
- **`/usr/local`**: A tertiary hierarchy for locally compiled software. When you install software from source (not via the package manager), it typically goes under `/usr/local/bin`, `/usr/local/lib`, etc., to avoid conflicts with package‑managed files.
- **`/usr/sbin`**: Non‑essential system administration binaries (e.g., web server daemons).
- **`/usr/share`**: Architecture‑independent data (documentation, icons, themes, man pages).
- **`/usr/include`**: Header files for C/C++ development (e.g., `stdio.h`, `stdlib.h`). When you compile a program, the compiler looks here for included headers.
- **`/usr/src`**: Source code (kernel source if installed).

#### `/var` Hierarchy
- **`/var/log`**: System and service log files (`syslog`, `auth.log`, `kern.log`, `apt/history.log`). Essential for debugging.
- **`/var/tmp`**: Temporary files that should survive a reboot (unlike `/tmp`).
- **`/var/lock`**: Lock files to prevent multiple processes from using a resource simultaneously (now often moved to `/run/lock`).
- **`/var/lib`**: Persistent state information for applications (e.g., package manager databases, MySQL data).

#### Filesystem Classification: Shareable vs. Unshareable, Static vs. Variable
The FHS organizes directories based on whether they can be shared between machines (over network) and whether their content changes during operation.

|                | **Shareable**                       | **Unshareable**                  |
|----------------|--------------------------------------|----------------------------------|
| **Static**     | `/usr` (software), `/opt` (packages) | `/etc` (configs), `/boot` (kernel) |
| **Variable**   | `/var/mail` (mail spool)             | `/var/run` (runtime state), `/var/lock` |

This design makes network‑based systems (like diskless workstations) possible: static, shareable directories can be mounted read‑only from a central server, while variable, unshareable directories are local to each machine.

### Navigating the Filesystem

- **`cd`** (change directory): Move around the tree.
  - `cd /` — go to root.
  - `cd /usr/bin` — go to that directory.
  - `cd ..` — go up one level (parent directory).
  - `cd .` — stay in the current directory (`.` means “here”).
  - `cd ~` or just `cd` — go to your home directory.
  - `cd -` — go to the previous directory you were in. Very useful for toggling between two locations.

- **Special directory entries**: In every directory, there are two hidden files:
  - `.` – a hard link to itself.
  - `..` – a hard link to its parent directory. In the root directory, `..` points to itself (root is its own parent).

### File Types in `ls -l`
The first character of the long listing tells the file type:

| Character | Type | Description |
|-----------|------|-------------|
| `-` | Regular file | Text, data, executable, etc. |
| `d` | Directory | A folder containing other files. |
| `l` | Symbolic link | A pointer to another file. |
| `c` | Character device | Device that transfers data character by character (terminal, serial port). |
| `b` | Block device | Device that transfers data in blocks (hard disk, USB drive). |
| `s` | Socket | Used for inter‑process communication (network or local). |
| `p` | Named pipe (FIFO) | Similar to a pipe, but with a filesystem name; allows one‑way communication between processes. |

Regular files and directories are the most common. Device files and sockets should be handled with care.

## 3. Simple Commands in Linux – Part 1

### Date and Calendar

- **`date`**: Prints the current date and time.  
  `date` → `Thu Nov 25 16:50:00 IST 2021`  
  Options allow formatting. For example, `date -R` gives RFC 5322 format (used in email headers): `Thu, 25 Nov 2021 16:50:00 +0530`.  
  You can use format specifiers (see `man date`): `date +"%A, %d %B %Y"` might give `Thursday, 25 November 2021`.  
  The system clock may not be accurate unless synchronised with NTP; `date` reflects the current system time setting.

- **`cal`**: Prints a text calendar.  
  `cal` — current month.  
  `cal 8 1947` — August 1947, showing India’s independence day was on a Friday.  
  `ncal` — alternative layout (transposed month), same arguments.

### Memory Statistics

- **`free`**: Displays total, used, free physical memory (RAM) and swap space.  
  `free` gives numbers in kibibytes, not easily readable.  
  `free -h` (“human‑readable”) shows values with appropriate units (GiB, MiB).  
  Example output:
  ```
            total   used   free   shared   buff/cache   available
  Mem:       31Gi   1.2Gi   26Gi     ...        ...          ...
  Swap:      2.0Gi   0B     2.0Gi
  ```
  - **Swap** is disk space used as an overflow when RAM is full. If swap is actively used, the system will feel slow because disk I/O is much slower than RAM.  
  The kernel uses free memory for disk caching (buff/cache), which can be released if applications need more memory. “Available” is an estimate of how much memory can be allocated to new processes without swapping.

### User and Group Information

- **`groups`**: Lists the groups the current user belongs to.  
  On a default Ubuntu installation, a user is in groups like `username` (a private group with only that user), `sudo` (ability to run commands as root), and others like `adm`, `dialout`, etc.  
  The group `sudo` is special because members can use `sudo` to temporarily gain root privileges.

### Understanding `ls -l` Output in Detail

A line from `ls -l` looks like:  
`drwxr-xr-x 2 user group 4096 Jul 28 2020 Desktop`

Let’s break it down field by field:

1. **File type and permissions**: `drwxr-xr-x`
   - First character: `d` (directory).
   - Next nine characters: three sets of three (`rwx`, `r-x`, `r-x`).
     - Owner (`user`): `rwx` → read, write, execute.
     - Group (`group`): `r-x` → read, execute (no write).
     - Others (everyone else): `r-x` → read, execute.
   - Each position can be `r`, `w`, `x`, or `-` (absent).  
     For a directory, “execute” means permission to traverse (enter) the directory. “Read” means listing its contents. “Write” means creating or deleting files inside.

2. **Number of hard links**: `2`
   - For a file, this is the count of directory entries pointing to the same mode.  
   - For a directory, it starts at 2 (the directory itself and its `.` entry) and increases by one for each subdirectory (because of `..` inside the subdirectory).

3. **Owner**: `user` – the username of the file’s owner.

4. **Group**: `group` – the group associated with the file.

5. **Size**: `4096` – size in bytes. For directories, it’s the size of the directory listing, not the sum of contents.

6. **Last modification timestamp**: `Jul 28 2020`. If the modification is within the current year, the time is shown instead of the year.

7. **Filename**: `Desktop`

### Inodes and Hard Links

An **inode** is a data structure that stores all information about a file except its name and its directory location. Each file has exactly one inode, containing:
- File type and permissions.
- Owner and group IDs.
- Size.
- Timestamps (modification, access, change).
- Pointers to the data blocks on disk.

The filename is stored in a directory entry, which maps the name to an inode number. Multiple directory entries can point to the **same inode**. Those are called **hard links**.

- `ls -i` shows the inode number as the first column.  
  Example: `ls -li file1` → `123456 -rw-r--r-- 1 user group ... file1`

- When you create a hard link with `ln file1 file3`, `file1` and `file3` have the same inode number. The link count increases to 2. If you delete `file1`, the data is not freed until the last hard link is removed. This is how backup systems can conserve space: multiple directory entries for unchanged files, all pointing to the same data blocks.

- Directories work similarly: a directory’s link count is 2 + number of subdirectories. The `.` entry in the directory and the directory entry in its parent both link to the same inode. The `..` entry in a subdirectory is a hard link to the parent.

### Permissions and `chmod`

Permissions can be changed with `chmod` (change mode). There are two syntaxes:

#### Symbolic Mode
- User categories: `u` (owner), `g` (group), `o` (others), `a` (all).
- Operators: `+` (add), `-` (remove), `=` (set exactly).
- Permissions: `r`, `w`, `x`.

Examples:
- `chmod g-w level1` – remove write permission from group.
- `chmod o+x file` – add execute for others.
- `chmod a=r file` – set all to read‑only.

#### Octal (Numeric) Mode
Each permission set (owner, group, others) is represented by a number from 0 to 7, which is the sum of:
- `4` – read
- `2` – write
- `1` – execute

| Binary | Octal | Permissions |
|--------|-------|-------------|
| 000    | 0     | --- (none) |
| 001    | 1     | --x (execute) |
| 010    | 2     | -w- (write) |
| 011    | 3     | -wx (write+execute) |
| 100    | 4     | r-- (read) |
| 101    | 5     | r-x (read+execute) |
| 110    | 6     | rw- (read+write) |
| 111    | 7     | rwx (all) |

The full permissions string is three octal digits. E.g., `755` means:
- Owner: `rwx` (7)
- Group: `r-x` (5)
- Others: `r-x` (5)

`700` gives owner full access and no access to group or others.

### File Manipulation

- **`mkdir`** (make directory): Creates one or more directories.  
  `mkdir level1` creates a directory named `level1`.

- **`touch`**: Two purposes:
  1. If the file does not exist, it creates an empty file (zero bytes). Very useful for placeholder files.
  2. If the file exists, it updates its access and modification timestamps to the current time.
  
  `touch file1 file2` creates both files.

- **`cp`** (copy): Copies files or directories.
  - `cp source destination`
  - If destination is a directory, the source file is copied inside it.
  - If destination is a filename, the copy gets that name (and overwrites if file exists, unless `-i` is used for interactive prompt).
  - To copy directories, use `-r` (recursive): `cp -r dir1 dir2` creates `dir2` as a copy of `dir1`.

- **`mv`** (move): Moves or renames files/directories.
  - `mv oldname newname` renames.
  - `mv file directory/` moves file into that directory.
  - Works on directories without `-r` because it’s simply changing the directory entry, not copying data.

- **`rm`** (remove): Deletes files.
  - `rm file` removes the file. By default, it does not ask for confirmation.
  - `rm -i file` asks before removing (safe for beginners).
  - `rm -r directory` removes a directory and its contents recursively. Very powerful and potentially dangerous.

- **`rmdir`**: Removes an empty directory. Safer than `rm -r`.

### Aliases

An alias is a nickname for a command. In the lecture, `rm` was aliased to `rm -i` to add safety.  
- `alias` lists all current aliases.
- `alias ll='ls -l'` creates a new command `ll`.
- `unalias ll` removes the alias.

Aliases are defined in shell configuration files like `~/.bashrc` or globally in `/etc/profile` to make them persistent.

### Permission Denial and Security

The demonstrations show that normal users cannot damage the system because:
- Files in `/etc`, `/var/log`, etc., are owned by `root` or system groups.
- Permissions typically give others no write access, and often no read access to sensitive logs.
- Even if a user tries to change permissions with `chmod` on a root‑owned file, it fails: only the owner (or root) can change permissions.
- Directories without execute (`x`) permission for others cannot be entered with `cd`.

This permission model is a core security feature inherited from Unix. It ensures that ordinary users can only write in their home directory and temporary areas like `/tmp`.

### File Type Identification

- **`file`** command: Determines file type by examining magic numbers (a few bytes at the start) rather than relying on extensions.
  - `file /usr/bin/znew` → “ASCII text, executable script” (a shell script).
  - `file /usr/bin/zoom` (a compiled program) → “ELF 64‑bit LSB shared object, x86‑64…”  
  - `file .` → “directory”.
  - `file file1` → “empty” if touched.

This helps you avoid opening binary files with a text viewer (which would produce garbled output and potentially freeze the terminal).

## 4. Simple Commands in Linux – Part 2

### More on `ls` Options and Syntax Flexibility

- **The `-d` option**: `ls -l directory` lists the *contents* of that directory. `ls -ld directory` lists the directory entry itself. This is useful to check the permissions of a directory without seeing inside.
- **Combining options**: `ls -ldi level1` shows long format, directory itself, and mode number. Order of short options doesn’t matter: `-ldi`, `-idl`, `-dil` all work the same. However, if an option takes its own argument (e.g., `--sort=time`), placement matters.
- **Long options**: Many commands have long GNU‑style options, e.g., `--all` for `-a`, `--directory` for `-d`. They are more self‑documenting but longer to type.  
  `ls --directory --inode level1` is equivalent to `ls -di level1`.

### Viewing File Content

- **`less`**: A pager that lets you scroll forward and backward through a file.  
  `less /etc/profile` opens the file. Navigation keys: `Space` (page down), `b` (page up), `q` (quit), `/pattern` (search), `n` (next match).  
  `less` does not need to read the entire file into memory, making it efficient for large files.

- **`cat`** (concatenate): Dumps the entire file to the terminal at once. If the file is long, the output scrolls off the screen. Use `cat` for short files or to combine files (`cat file1 file2 > combined`).

- **`more`**: An older pager, similar to `less` but with fewer features (e.g., cannot scroll backward). The joke “less is more” refers to the fact that `less` has more features than `more`.

- **`head`**: Displays the first 10 lines by default.  
  `head -n 5 /etc/profile` shows the first 5 lines. The option can also be written as `-5` (short form).

- **`tail`**: Displays the last 10 lines.  
  `tail -n 5 file` shows last 5 lines. A very common use is `tail -f /var/log/syslog`, which continuously monitors the file and prints new lines as they are added (follow mode).

- **`wc`** (word count): Prints line count, word count, and byte count.  
  `wc /etc/profile` → `27 97 581` means 27 lines, 97 words, 581 bytes.  
  `wc -l` outputs only line count, `-w` for words, `-c` for bytes.

### Discovering Commands

- **`apropos`**: Searches manual page names and descriptions.  
  `apropos who` returns commands like `who`, `whoami`, `w`. It is equivalent to `man -k who`.

- **`which`**: Locates the executable.  
  `which less` → `/usr/bin/less`.  
  This only works for external commands, not shell built‑ins or aliases.

- **`type`**: More comprehensive than `which`.  
  `type ls` might say “ls is aliased to `ls --color=auto`” or “ls is /usr/bin/ls”.  
  `type cd` says “cd is a shell builtin”.  
  This is the best way to know how the shell will interpret a command.

- **`file`** on executables: Use `file $(which less)` to see if a command is a script or a binary.

### Shell Built‑ins vs. External Commands

Some commands are part of the shell itself (built‑ins) for efficiency. Examples: `cd`, `echo`, `pwd` (sometimes an external also exists), `exit`, `type`. They run directly in the shell process without spawning a child process. External commands (e.g., `ls`, `grep`, `find`) are separate executable files in `/bin`, `/usr/bin`, etc. `help` in bash only applies to built‑ins; for external commands, use `man`.

### Hard Links vs. Symbolic (Soft) Links

| Feature | Hard Link | Symbolic Link |
|---------|-----------|---------------|
| Creation command | `ln target link` | `ln -s target link` |
| Inode number | Same as target | Different (new inode) |
| Cross‑filesystem | No (must be on same filesystem) | Yes (stores path as text) |
| Pointing to directories | Not allowed (usually) | Allowed |
| Behaviour if target deleted | Data still accessible via hard link | Link becomes broken (dangling symlink) |
| Shows as `l` in `ls -l` | No, shows as regular file with link count >1 | Yes, `l` with `-> target` |
| Disk usage | No extra data (just another directory entry) | Small file containing the path |

A symbolic link is a separate file whose content is a path to the target. If you move or delete the target, the symlink breaks. Hard links are indistinguishable from the original file; they share the same data blocks.

#### Practical Use
- Hard links are used by backup utilities (e.g., `rsync` with `--link-dest`) to create space‑efficient snapshots.
- Symbolic links are widely used to create shortcuts, maintain multiple versions of libraries (`libfoo.so -> libfoo.so.2`), and redirect paths.

### File Size, Blocks, and Disk Usage

- **`ls -l`** shows the logical size (bytes of data). For sparse files, this can be larger than disk usage.
- **`stat`** gives detailed information: size, blocks allocated, access/modify/change timestamps, inode, permissions.  
  `stat /usr/bin/znew` shows Size: 4553, Blocks: 16 (each block is 512 bytes, so 16×512 = 8192 bytes allocated). The actual disk usage is 8 KiB, even though the file is only 4.5 KiB, because the filesystem allocates space in blocks. The block size of the filesystem is often 4096 bytes (4 KiB). A file that is 4553 bytes will occupy two 4K blocks, i.e., 8 KiB. `stat` reports in 512‑byte blocks by tradition, so 16 blocks = 8 KiB.
- **`du`** (disk usage): Estimates file space usage.  
  `du -h /usr/bin/znew` → `8.0K` (same as stat’s allocation).  
  `du -sh directory` gives a summary of total usage.  
  `du` traverses the directory tree and sums sizes.

### The `/proc` and `/sys` Virtual Filesystems

These are pseudo‑filesystems that exist only in RAM. They provide a window into the kernel and hardware.

#### `/proc`
- Contains numbered subdirectories for each running process (PID). Inside, files like `cmdline`, `status`, `environ`, `cwd` (symlink to current working directory), `fd/` (file descriptors) provide deep introspection of processes.
- Non‑process files:
  - `cpuinfo`: CPU details (model, cores, flags).
  - `meminfo`: Memory statistics (more detailed than `free`).
  - `version`: Kernel version string (equivalent to `uname -a`).
  - `partitions`: List of disk partitions by major/minor number.
  - `kcore`: An ELF core file representing the entire physical address space of the kernel. Its size is reported as the maximum virtual address space (e.g., 128 TB on 64‑bit systems). This is a virtual file and does not consume that much RAM.
- Many files report size 0 when listed with `ls`, but reading them (`cat`, `less`) returns dynamic content generated on the fly.

#### `/sys`
- A more structured and modern (kernel 2.6+) interface to kernel objects (kobjects).
- Organised by bus, class, devices, etc. For example:
  - `/sys/bus/usb/devices/` – USB device tree.
  - Subdirectories named like `1-1` represent USB ports and hubs.
  - Inside each device directory, files like `manufacturer`, `product`, `idVendor`, `idProduct`, `speed` allow you to query device details.
  - In the lecture, reading `manufacturer` gave “JBL”, and `product` gave “JBL Quantum 300” – a USB headset.
  - Another device gave “PixArt” and “Dell M116 USB Optical Mouse”.
- Writing to certain files in `/sys` can change kernel parameters at runtime (e.g., `/sys/class/leds/` for keyboard backlight).

### Disk and Filesystem Information

- **`df`** (disk free): Shows free space on mounted filesystems.  
  `df -h` gives human‑readable output. Example:
  ```
  Filesystem      Size  Used Avail Use% Mounted on
  /dev/sda2       183G   30G  144G  18% /
  /dev/sda3       481G  6.9G  450G   2% /home
  ```
  This tells you where each partition is mounted and how much space is left.
- In the `/proc` filesystem, you can also read `/proc/mounts` or `/proc/partitions`, but `df` is more user‑friendly.

### Putting It All Together: A Safe Sandbox for Exploration

The lectures emphasise that you cannot break the system by exploring. Permissions lock you out of critical areas. The demonstrations of trying to create a file in `/`, or reading a protected log file, show permission denied errors. This safety net allows you to run commands like `cd /proc` and `cat cpuinfo` without fear. The worst that can happen in your home directory is accidental deletion of your own files—hence the use of `-i` aliases.

## Part I – Command Line Editors

### 1. Why Learn a Terminal‑Based Editor?

When you work on remote supercomputers, cloud servers, or low‑bandwidth connections, a graphical editor may be impossible or painfully slow. A terminal‑based editor:

- Runs inside a terminal emulator (no X11 forwarding needed).
- Consumes negligible bandwidth – only the characters you type and the screen updates are sent.
- Is universally available on all Unix‑like systems (even minimal installations).
- Allows you to edit configuration files and scripts without leaving the command line, and is often the only way to fix a broken system (e.g., when a GUI fails to start).

Moreover, proficiency with a powerful terminal editor like Vim or Emacs can dramatically speed up coding and text manipulation through keyboard‑driven commands, regular expressions, and macros.

### 2. The Line Editor: `ed`

`ed` is the original Unix text editor, written by Ken Thompson in 1971. It operates on a single line at a time – you never see the entire file in one view. While almost nobody uses `ed` for everyday editing today, understanding it illuminates the origins of many `vi` commands and the Unix philosophy of small, composable tools.

**Basic Operation**
- Start: `ed filename`. It reads the whole file into memory (a *buffer*), prints the number of bytes read, and waits for commands. There is no prompt by default; press `P` (capital P) to enable the `*` prompt.
- Commands are *address‑based*: you specify which line(s) to act on, followed by a single‑letter command.
- **Addresses**
  - `1` – line 1.
  - `$` – last line.
  - `.` – current line (the “dot”).
  - `%` – all lines (equivalent to `1,$`).
  - `+` and `-` – relative lines (`+1` means next line, `-2` two lines back).
  - `,` – from the first address to the second (e.g., `2,5` means lines 2 through 5).
  - `;` – from the current line to the end (e.g., `;,$` from current to last).
  - `/pattern/` – the first line matching the regular expression.
- **Common commands**
  - `p` – print the addressed lines.
  - `n` – print lines with line numbers.
  - `d` – delete lines.
  - `a` – append text after the addressed line (enter text, finish with a line containing only `.`).
  - `i` – insert before the addressed line.
  - `c` – change the addressed lines (replace with new text).
  - `s/regex/replacement/` – substitute on the addressed line(s). Add `g` for global (all occurrences on the line), and a number for a specific occurrence.
  - `w` – write buffer to file.
  - `q` – quit.
  - `wq` or just `w` then `q` – save and quit.
  - `u` – undo the last command.
  - `!command` – execute a shell command (output displayed, does not modify buffer).
  - `r !command` – read output of a shell command into the buffer after the current line.
  - `r filename` – read a file into the buffer at the current position.
  - `f` – show current filename.

**Example session**
```bash
$ echo "hello world" > test.txt
$ echo "welcome to line editor" >> test.txt
$ echo "ed is oldest" >> test.txt
$ echo "end of file" >> test.txt
$ ed test.txt
117    # bytes read
* ,p   # print all lines
hello world
welcome to line editor
ed is oldest
end of file
* $    # go to last line
end of file
* 2,3p  # print lines 2 to 3
welcome to line editor
ed is oldest
* /ed/  # search for 'ed'
ed is oldest
* s/oldest/oldest editor/p
ed is oldest editor
* w
126
* q
```
Key takeaway: `ed` introduced the concept of line addressing and regular expression substitution that became central to `ex` and later `vi`’s “ex mode”.

### 3. The Visual Editors: `vi` and `vim`

`vi` (pronounced “vee‑eye”) was created by Bill Joy in 1976 as an extension of the line editor `ex`. It added a *visual mode* where you can see a full screen of text and move the cursor freely. `vim` (Vi IMproved) by Bram Moolenaar (1991) is the modern, feature‑rich successor, present on virtually every Linux distribution. When you type `vi`, you often get `vim`.

#### 3.1. Modes – The Heart of `vi`

`vi` is a modal editor. The same key performs different actions depending on the mode. This is its greatest strength (once learned) and the greatest hurdle for beginners.

1. **Command Mode** (default) – keystrokes are interpreted as commands (move, delete, copy, etc.). Press `Esc` to ensure you are here.
2. **Insert Mode** – keystrokes are inserted as text. Entered by pressing `i`, `a`, `o`, `I`, `A`, `O`, etc. The bottom of the screen usually shows `-- INSERT --`. Leave it with `Esc`.
3. **Ex Mode** – press `:` from Command Mode. The cursor jumps to the bottom line, where you can enter `ex`‑style commands (`:w`, `:q`, `:%s/foo/bar/g`). After executing, you return to Command Mode.

Understanding and mastering these modes is essential. Beginners often get stuck by accidentally typing in Command Mode and wondering why text appears in odd places.

#### 3.2. Essential Commands in Command Mode

**Movement (the cursor keys also work, but these are universal)**
- `h`, `j`, `k`, `l` – left, down, up, right. `j` has a tactile notch on most keyboards.
- `w` – next word (beginning).
- `b` – previous word (beginning).
- `e` – end of current/next word.
- `0` – beginning of line.
- `^` – first non‑blank character of line.
- `$` – end of line.
- `Ctrl+f` – page down (Forward).
- `Ctrl+b` – page up (Back).
- `Ctrl+d` – half‑page down.
- `Ctrl+u` – half‑page up.
- `gg` or `1G` – go to first line.
- `G` – go to last line.
- `NG` or `:N` – go to line N (e.g., `42G` or `:42`).
- `%` – jump to matching parenthesis, bracket, or brace.

**Inserting Text**
- `i` – insert before cursor.
- `a` – append after cursor.
- `I` – insert at beginning of line.
- `A` – append at end of line.
- `o` – open a new line below and insert.
- `O` – open a new line above and insert.

**Deleting, Changing, Yanking (Copy)**
- `x` – delete character under cursor (like `Del`).
- `X` – delete character before cursor (like `Backspace`).
- `dw` – delete from cursor to end of word (including trailing space).

**Undo and Redo**

**Visual Mode (character/line/block selection)**

#### 3.3. Ex Mode Commands (preceded by `:`)

These are the direct descendants of `ed`/`ex` commands, immensely powerful.

- `:q` – quit.

- Perform your edits.

**Regular Expressions in Searches**

**Configuration and Plugins**

## Part II – Networking Commands and SSH

### 6. IP Addressing Fundamentals

Every device on a TCP/IP network needs an IP address. We focus on **IPv4** (32‑bit), although IPv6 (128‑bit) is increasingly important.

- An IPv4 address is usually written in dotted‑decimal notation: `192.168.1.10`.
- The address is divided into a *network* portion and a *host* portion, determined by the **subnet mask** or **CIDR** notation. E.g., `192.168.1.10/24` means the first 24 bits are the network, the remaining 8 bits are the host. The subnet mask would be `255.255.255.0`.

#### 6.1. Public vs. Private Addresses

The internet routes only **public** IP addresses. To conserve public address space, certain ranges are reserved for **private** networks. Packets with these source/destination addresses are not routed on the public internet; they stay inside local networks.

**Private IPv4 ranges:**
- **Class A**: `10.0.0.0` – `10.255.255.255` (mask `255.0.0.0`, or `/8`). 16,777,216 addresses.
- **Class B**: `172.16.0.0` – `172.31.255.255` (mask `255.240.0.0`, or `/12`). 1,048,576 addresses.
- **Class C**: `192.168.0.0` – `192.168.255.255` (mask `255.255.0.0`, but commonly used as `/16`; technically it’s 256 contiguous `/24` networks). 65,536 addresses.

Home routers typically use `192.168.x.x` (often `192.168.0.0/24` or `192.168.1.0/24`). Corporate/campus networks may use `10.x.x.x`.

The **loopback address** `127.0.0.0/8` (commonly `127.0.0.1`) is a virtual interface that routes packets back to the same machine. It’s always present.

#### 6.2. Network Address Translation (NAT)

Since private addresses are not globally routable, a home or office network uses NAT. The router has a public IP on its WAN interface and assigns private IPs to devices inside. When an internal device sends a packet to the internet, the router rewrites the source IP to its own public IP, remembers the mapping (port number), and forwards the packet. Reply packets are translated back. This is how many devices share one public IP.

#### 6.3. Subnetting and Hierarchical Networks

Large organizations can create complex internal networks using multiple private ranges, separating departments or security zones. Routers and firewalls control traffic between subnets. For instance, a campus might have:
- Backbone: `10.0.0.0/16`
- Academic subnet: `10.1.0.0/16`
- Data center: `10.2.0.0/16`
- Each can have further subdivisions.

This is accomplished by routing tables and Access Control Lists (ACLs) on network switches.

---

### 7. Ports and Services

An IP address identifies a host, but a **port** identifies a specific service/process on that host. Ports are 16‑bit numbers (0–65535). The first 1024 are well‑known ports, assigned by IANA to standard services.

Common well‑known ports:
- `20/21` – FTP (data/control)
- `22` – SSH (Secure Shell)
- `25` – SMTP (email sending)
- `53` – DNS
- `80` – HTTP (unencrypted web)
- `443` – HTTPS (encrypted web)
- `631` – IPP (printing)
- `3306` – MySQL/MariaDB database
- `5432` – PostgreSQL

A socket is an IP:port pair; a connection is a pair of sockets. For example, your browser might connect from `192.168.1.5:52341` to `203.0.113.10:443`.

#### 7.1. Security by Obscurity: Changing Default Ports

Administrators sometimes change SSH from port `22` to a high port (e.g., `2222`) to reduce automated brute‑force attacks. This is not real security but reduces log noise. Firewalls still protect.

---

### 8. Firewalls

A firewall filters network traffic based on rules (source/destination IP, port, protocol, state). In Linux, the standard firewall is `iptables` (or its newer front‑ends `ufw`, `firewalld`).

- **Host‑based firewall**: Runs on the machine itself (e.g., `ufw allow 22/tcp`).
- **Network firewall**: A dedicated appliance or router ACL.
- **Web Application Firewall (WAF)**: Filters HTTP/HTTPS traffic to block SQL injection, XSS, etc. (e.g., Apache mod_security, Cloudflare WAF).

The lectures mention that the IIT Madras website uses a WAF. A DNS lookup from outside the campus returns a public IP that likely sits behind a reverse proxy/WAF, while internal lookups give a private IP.

Firewalls can also limit the rate of connections, block entire countries via GeoIP, and log suspicious activity.

---

### 9. SELinux (Security-Enhanced Linux)

Developed by the NSA, SELinux adds **Mandatory Access Control (MAC)** on top of the traditional Unix Discretionary Access Control (DAC, i.e., `rwx` permissions).

- DAC allows the owner of a file to set its permissions. A compromised process running as the owner can do anything.
- MAC enforces system‑wide policies that even the owner cannot override. Every process and file is given a **security context**: `user:role:type:level`.
- **Type Enforcement**: The core of SELinux. Rules define which *types* (or domains for processes) can access which file types and with what operations. For example, a web server process (`httpd_t`) can only read files labeled `httpd_sys_content_t`, even if those files are world‑readable.
- **Modes**:
  - **Enforcing**: Policies are applied, violations are denied and logged.
  - **Permissive**: Violations are only logged but not denied – useful for debugging.
  - **Disabled**: SELinux is off.
- Checking contexts: `ls -Z`, `ps -Z`. Adjusting contexts: `chcon`, `restorecon`, `semanage`.
- SELinux can also provide **sandboxing**: even if one service is breached, the attacker cannot access other parts of the system because the process is confined to its domain.
- Historically used on Red Hat, CentOS, Fedora; now also available on Ubuntu.

The lectures showed that the HTTP server process had a context `system_u:system_r:httpd_t:s0`. This restricts it to only read web content and prevents it from writing to system files.

### 10. Network Diagnostic Tools

#### 10.1. `ping`
Sends ICMP Echo Request packets to a target. If the target replies, you know the network path works. However, many hosts block ICMP, so no reply doesn’t guarantee the host is down. Only positive replies are conclusive.

#### 10.2. `traceroute` / `tracepath`
Maps the route packets take to a destination. Sends packets with increasing TTL (Time To Live). Each router decrements TTL; when it hits 0, the router sends back an ICMP Time Exceeded message. This reveals intermediate hops. *Note: some routers don’t send these messages, showing `*` in output.*

#### 10.3. `nslookup` / `dig`
Query DNS servers to translate hostnames to IP addresses (forward lookup) or IP to hostname (reverse lookup, `dig -x`).
- `nslookup www.iitm.ac.in` might return a private IP from inside campus, and a different public IP from outside.
- `dig` gives more detailed output (authority, TTL, etc.) and is preferred for scripting.

#### 10.4. `netstat` / `ss`
Display network connections, routing tables, interface statistics. `netstat -tulpn` shows listening TCP/UDP ports with owning process. The newer `ss` is faster. In the demo, `netstat` showed active SSH and web connections.

#### 10.5. `ifconfig` (deprecated) / `ip addr`
Show network interface configuration (IP addresses, MAC addresses, MTU). The machine in the lecture had multiple private IPs on different interfaces (virtual machines, containers, separate networks).

#### 10.6. `nmap` (Network Mapper)
Scans ports on remote hosts to discover open services. Should only be used with permission on networks you own/manage; unauthorized scanning can be considered an attack.

#### 10.7. `wireshark` / `tcpdump`
Packet capture and analysis. `tcpdump` is a command‑line tool; `wireshark` has a GUI. It can decode protocols and show raw data. Essential for debugging network applications, but captures all traffic on the interface, which is a security and privacy risk if misused.

#### 10.8. `who` / `w`
Show who is logged into the system, what they are doing, and from where. Useful on multi‑user servers.

### 11. SSH – Secure Shell

`ssh` allows encrypted remote login and command execution. It replaces insecure protocols like `telnet` and `rlogin`.

- **Basic syntax**: `ssh username@hostname` (or `ssh -p port user@host`).
- **Authentication**:
  - Password: simplest, but vulnerable to brute‑force.
  - Public key: generate a key pair (`ssh-keygen`), copy the public key to the server’s `~/.ssh/authorized_keys`. The client proves possession of the private key without transmitting it. This is much more secure and can be used without passwords.
- **SSH Tunneling / Port Forwarding**:
  - Local: `ssh -L local_port:remote_host:remote_port user@ssh_server`. Traffic to `localhost:local_port` is forwarded over SSH to `remote_host:remote_port` from the SSH server.
  - Remote: `ssh -R remote_port:local_host:local_port user@ssh_server`. The SSH server listens on `remote_port` and forwards traffic back to the local client.
  - Dynamic: `ssh -D local_port user@ssh_server` acts as a SOCKS proxy.
- **SSH in a multi‑hop scenario**: You can chain SSH connections or use `ProxyJump`/`ProxyCommand` to reach a machine that is not directly accessible. E.g., `ssh -J user@gw.example.com user@internal.host`.
- **Secure Copy (`scp`)**: `scp file user@host:/path` copies files over SSH. The lecture used `scp` to retrieve a tar archive from another machine.
- **SFTP**: FTP‑like interface over SSH.

**Using SSH to Access HPC Facilities**
As the lecture highlights, all top‑500 supercomputers run Linux. Users typically `ssh` into a login node and submit batch jobs to a scheduler (SLURM, PBS, etc.). They rarely run intensive tasks interactively. Knowledge of the command line, text editors, and SSH is essential to manage files, edit scripts, and submit jobs.

## Part I – Combining Commands and Files

The philosophy of Unix is built on small, sharp tools that do one thing well. The real power comes when you chain them together. This is achieved through **command separators**, **exit codes**, and **I/O redirection**.

### 1. Command Separators and Conditional Execution

You can place multiple commands on a single line. How the shell executes them depends on the separators used.

| Separator | Meaning |
|-----------|---------|
| `;` (semicolon) | Execute commands sequentially, irrespective of success or failure. |
| `&&` (double ampersand) | Logical AND – execute the next command **only if** the previous one succeeded (exit code 0). |
| `||` (double pipe) | Logical OR – execute the next command **only if** the previous one failed (exit code non‑zero). |

These mirror the C‑language logical operators. The exit code (`$?`) of a command is an integer 0–255: 0 means success (true), any other value means an error (false).

#### Examples
```bash
# Run three commands regardless of errors
ls / ; date ; wc -l /etc/profile

# Update only if the repository check succeeds
sudo apt-get update && sudo apt-get upgrade

# If compilation fails, open the error log
make || less error.log
```
These can be combined:
```bash
command1 && command2 || command3
```
If `command1` succeeds, `command2` runs; if `command1` fails, `command3` runs (as a fallback). But careful: if `command1` succeeds and `command2` fails, then `command3` will also run! A better pattern is:
```bash
command1 && { command2; true; } || command3
```

#### Exit Codes in Detail
Every command returns an exit status. Common reserved codes:
- `0` – success.
- `1` – general error (often “permission denied”).
- `2` – misuse of shell builtins (e.g., invalid option).
- `126` – command found but not executable.
- `127` – command not found.
- `128+n` – terminated by signal `n` (e.g., `130` = 128+2, interrupted by Ctrl+C, SIGINT).
- `255` – exit status out of range.

You can use `echo $?` right after any command to see its exit code. Scripts often use these to make decisions.

### 2. Subshells and Command Grouping

Placing commands inside parentheses `( ... )` runs them in a **subshell** – a child process that inherits variables, but any changes (directory, variables) are lost when it exits.

```bash
echo $BASH_SUBSHELL   # shows 0
(echo $BASH_SUBSHELL) # shows 1 – inside a subshell
( (echo $BASH_SUBSHELL) ) # shows 2 – nested subshell
```
Subshells are useful for temporary environment changes without affecting the parent shell:
```bash
(cd /tmp && ls)   # list /tmp, but stay in current directory afterward
```
You can also group commands without a subshell using `{ ...; }`. Commands run in the current shell (no new process), and a semicolon and spaces are required:
```bash
{ cd /tmp; ls; }
```

### 3. File Descriptors and Redirection

Every process inherits three standard file descriptors (FDs):
- **0 – stdin** (standard input, usually the keyboard)
- **1 – stdout** (standard output, usually the screen)
- **2 – stderr** (standard error, also screen, unbuffered)

Redirection operators change where these streams go.

| Operator | Action |
|----------|--------|
| `command > file` | Redirect stdout to `file` (overwrite) |
| `command >> file` | Redirect stdout to `file` (append) |
| `command 2> file` | Redirect stderr to `file` |
| `command &> file` | Redirect both stdout and stderr to `file` (bash 4+) |
| `command < file` | Take stdin from `file` instead of keyboard |
| `command <<< "string"` | Here‑string – feed a string as stdin |
| `command <<EOF` ... `EOF` | Here‑document – feed multiple lines of inline text |

#### Overwriting vs. Appending
- `>` **always** creates a new file (or truncates an existing one to zero length) and then writes the output from the beginning.
- `>>` creates the file if it doesn’t exist; if it does, it appends the output to the end. It never destroys previous content.

In the lecture, running `ls /blah > file1` emptied `file1` because `ls` produced no stdout (only an error on stderr). The redirection only captured stdout, so the file became empty. If you use `>>`, a single `>` in a subsequent command will overwrite everything previously appended. This is dangerous if you accidentally combine `>` and `>>` in a script.

#### Capturing Both Output and Errors
If you want both stdout and stderr in the same file, use `&>` or `2>&1`:
```bash
ls /blah > out.txt 2>&1   # merge stderr into stdout stream going to file
ls /blah &> out.txt        # shorter equivalent
```
#### Standard Input Redirection
Commands like `cat` read from files listed as arguments; if none, they read from stdin.  
`cat > file` creates a new file and takes input from the keyboard until you press Ctrl+D (EOF). This is a quick way to create text files without an editor.

`cat >> file` does the same but appends.  
`cat` without arguments simply echoes each line you type to stdout (screen). Press Ctrl+D to end.

### 4. Pipelines

The pipe `|` connects the stdout of one command to the stdin of the next. It’s the quintessence of Unix composability.

```bash
ls -l /usr/bin | sort | less
```
This lists files, sorts them, and pages the result. Pipelines run all commands concurrently; the shell sets up the pipe before starting any of them.

#### Named vs. Unnamed Pipes
- Unnamed pipe: `|`
- Named pipe (FIFO): `mkfifo mypipe` – a persistent special file that allows unrelated processes to communicate.

### 5. Practical File Manipulation with Redirection

**Creating a multi‑line file quickly:**
```bash
cat > poem.txt <<EOF
Roses are red,
Violets are blue.
EOF
```
**Capturing system info for later analysis:**
```bash
hwinfo > hwinfo.txt
# ... later
less hwinfo.txt
```
**Combining multiple commands’ output into one file (safely):**
```bash
{ date; wc -l /etc/profile; stat /bin/ls; } >> combined.log
```
Notice the use of `{ ... }` to group commands and redirect their combined stdout.

## Part II – Software Management

In the early days, installing software on Linux meant downloading source tarballs, running `./configure`, `make`, and `make install`, and manually tracking dependencies. Today, **package managers** automate all of this, maintaining a database of installed files, versions, dependencies, and upgrade paths.

### 1. Package Management Concepts

A package is a compressed archive containing compiled binaries, libraries, configuration files, documentation, and metadata (version, dependencies, architecture, checksums). The package manager:
- **Installs** new software, resolving dependencies.
- **Updates** existing packages to newer versions while preserving configurations.
- **Removes** packages cleanly, optionally deleting configuration files (purge).
- **Verifies** integrity using cryptographic signatures and checksums.
- **Searches** a remote repository for available packages.

Two dominant package formats exist:
- **RPM** (Red Hat Package Manager) – used by Red Hat, Fedora, CentOS, SUSE, openSUSE.
- **DEB** (Debian) – used by Debian, Ubuntu, Mint, Knoppix.

The lecture focuses on **DEB** and the Advanced Package Tool (`apt`) because Ubuntu is the most popular desktop Linux.

### 2. Ubuntu Repositories and Package Naming

Ubuntu stores its packages in online repositories (archives). The list of repositories is defined in `/etc/apt/sources.list` and files under `/etc/apt/sources.list.d/`.

```bash
cat /etc/apt/sources.list
```
The lines that are not commented start with `deb` or `deb-src`. Example:
```
deb http://in.archive.ubuntu.com/ubuntu/ focal main restricted universe multiverse
```
- **focal** – the codename for Ubuntu 20.04 LTS.
- **main, restricted, universe, multiverse** – the *components* (repositories). `main` is officially supported, `restricted` includes proprietary drivers, `universe` is community‑maintained, `multiverse` is non‑free.

Third‑party software (Google Chrome, Microsoft Teams, Zoom) adds their own `.list` files in `/etc/apt/sources.list.d/`, ensuring seamless updates via `apt`.

#### Package Filename Convention
A DEB package file follows a strict naming scheme:
```
<name>_<version>-<revision>_<architecture>.deb
```
Example from the lecture:
```
nmap_7.80+dfsg1-2build1_amd64.deb
```
- `nmap` – package name.
- `7.80+dfsg1` – upstream version.
- `2build1` – Debian/Ubuntu revision (packaging changes).
- `amd64` – architecture (64‑bit). Could be `i386`, `armhf`, `arm64`, `all` (architecture‑independent, e.g., documentation or scripts).

### 3. Checking Your System’s Version and Architecture

```bash
lsb_release -a   # shows distributor ID, release number, codename (focal)
uname -a         # shows kernel version and architecture (x86_64)
```

### 4. Package Management Tools Hierarchy

**Low‑level**: `dpkg` – installs/removes/querys individual `.deb` files without dependency resolution.
**High‑level**: `apt` / `apt-get` / `apt-cache` – front‑ends that handle repository downloads and dependencies. `aptitude` is an interactive text‑based front‑end, and `synaptic` is a graphical one.

`apt-get` and `apt-cache` are the traditional tools; `apt` is the newer, simpler command (introduced in Ubuntu 16.04) merging the most used functions.

#### Using `apt-cache` (read‑only, no `sudo` needed)

| Command | Purpose |
|---------|---------|
| `apt-cache search keyword` | Find packages matching a keyword in name/description |
| `apt-cache show package` | Display detailed metadata (version, size, dependencies, description, priority, section) |
| `apt-cache pkgnames [prefix]` | List all installed package names, optionally filtered by prefix |
| `apt-cache policy package` | Show installed and available versions |

The lecture searched for `fortune`, `nmap`, `wget` to illustrate finding and inspecting packages.

#### Package Priorities
Each package has a priority, which guides the system in decisions about upgrading and removing:
- **required** – essential for system boot and basic operation (e.g., `libc6`, `bash`). Never remove.
- **important** – provides functionality expected on any Unix‑like system.
- **standard** – included in a default installation (e.g., `wget`).
- **optional** – can be installed/removed without breaking the system (e.g., `fortune`, many GUI apps).
- **extra** – may conflict with higher‑priority packages; install only when needed.

In `apt-cache show`, look for the `Priority:` field.

#### Package Sections
Ubuntu categorizes packages into sections (e.g., `admin`, `devel`, `doc`, `editors`, `games`, `gnome`, `java`, `libs`, `math`, `net`, `python`, `shells`, `sound`, `utils`, `web`, `x11`). A package’s section is visible in its metadata. This helps you discover software by browsing sections on [packages.ubuntu.com](https://packages.ubuntu.com).

### 5. Integrity Verification with Checksums

Package managers ensure the downloaded binary is exactly what the maintainer intended by comparing cryptographic hash digests:

- **MD5** – 128‑bit, now considered weak but still used for accidental corruption detection.
- **SHA‑1** – 160‑bit, also deprecated against intentional attacks.
- **SHA‑256** – 256‑bit, currently secure.

Even a single‑bit change in the file produces a radically different hash. The lecture demonstrated this by creating two nearly identical text files (`file1.txt` and `file2.txt`) differing by one character (`h` vs `n`) and comparing their `md5sum`, `sha1sum`, and `sha256sum` outputs.

The package manager automatically verifies the checksums of downloaded packages against the repository’s signed `Release` file. Additionally, the repository index is signed with a GPG key. When you first add a repository, you import its public key. This chain of trust prevents tampering.

### 6. Superuser Privileges and `sudo`

Installing, updating, or removing system‑wide packages requires **superuser** (root) privileges. In Ubuntu, the root account is locked by default; users in the `sudo` group can run commands with root privileges by prefixing `sudo`.

The list of authorized sudoers is in `/etc/sudoers`. This file should never be edited directly; instead use `visudo` (which checks syntax). The lecture showed:
```bash
sudo cat /etc/sudoers   # only works if you're in sudoers
```
Failed attempts are logged to `/var/log/auth.log`:
```
Authentication failure ... user NOT in sudoers
```
Successful uses are also logged. This provides an audit trail.

**Important security principle:** Never log in as root for daily work. Use `sudo` only for the specific command that needs it. This limits the damage from mistakes or malware.

### 7. Updating and Upgrading

The system must first fetch the latest package index from repositories (`update`), then apply the new versions (`upgrade`).

```bash
sudo apt update           # download package lists
sudo apt upgrade          # upgrade all installed packages (may install new deps)
sudo apt full-upgrade     # also removes packages if needed to upgrade
sudo apt autoremove       # remove packages that were automatically installed and are no longer needed
```
`autoremove` frees disk space by deleting obsolete dependencies.

#### Installing and Removing Packages
```bash
sudo apt install nmap          # install a package
sudo apt remove nmap           # remove the package but keep configuration files
sudo apt purge nmap            # remove everything, including configs
sudo apt reinstall nmap        # reinstall if something got corrupted
```
The package manager resolves dependencies automatically. If a dependency is needed, it prompts for confirmation.

#### Cleaning Up
- `sudo apt clean` – clears the local cache of retrieved package files (`.deb`) from `/var/cache/apt/archives`. Good for freeing space.
- `sudo apt autoclean` – only removes packages that are no longer available in the repositories (older versions).

### 8. Querying with `dpkg` (Low‑level)

`dpkg` works directly on the package database, without touching repositories.

| Command | Purpose |
|---------|---------|
| `dpkg -l [pattern]` | List packages matching pattern (like `dpkg -l nmap*`) |
| `dpkg -L package` | List all files installed by the package |
| `dpkg -s package` | Show status and metadata (installed version, priority, section) |
| `dpkg -S /path/to/file` | Find which package owns a file (e.g., `dpkg -S /usr/bin/perl`) |
| `dpkg -i file.deb` | Install a local `.deb` file (without dependency resolution) |

**Warning:** `dpkg -i` does not handle dependencies. If a required library is missing, the package will be left unconfigured. Always prefer `apt install ./file.deb` which resolves dependencies.

#### The Database Directory `/var/lib/dpkg/`
- `status` – list of all installed/removed packages and their states.
- `available` – list of packages available from repositories (large file).
- `info/` – per‑package directory containing `.list` (files installed), `.conffiles`, `.md5sums` (checksums of installed files), and scripts.

Using the provided checksums, you can verify a package’s integrity:
```bash
cd /var/lib/dpkg/info
md5sum -c wget.md5sums   # checks all files; reports any mismatch
```
If a mismatch appears, the file may have been corrupted or tampered with. This is a powerful forensics tool.

#### Advanced Query with `dpkg-query`
You can extract arbitrary fields in a custom format:
```bash
dpkg-query -W -f='${Section} ${binary:Package}\n' | sort | less
```
This lists all installed packages with their section, sorted alphabetically. Filter with `grep` to see only packages in a certain section (e.g., `grep shells`). This is how you discover new tools by browsing categories.

### 9. Repository Management
- Third‑party repositories in `/etc/apt/sources.list.d/` are added either manually or by the installer of the software (e.g., `google-chrome.list`).
- To add a repository manually, you would copy its GPG key and add a `.list` file. The package manager uses these keys to verify that the repository index hasn’t been forged.

### 10. Practical Advice
- **Stay within official repositories** unless you have a strong reason. The Ubuntu repositories are curated, tested, and patched for security.
- **Avoid downloading random `.deb` files from the internet.** They may contain malware, or conflict with system libraries, breaking your system.
- When a program isn’t available via `apt`, look for a **Snap** or **Flatpak** package, or use a **Docker** container. These containerized formats isolate the application and its dependencies from the rest of the system, mitigating conflicts.

## Part III – Linux Process Management and Job Control

Understanding processes and jobs is crucial for multitasking in a terminal, killing misbehaving programs, and writing scripts.

### 1. Processes vs. Jobs

- A **process** is a running instance of a program. It has a unique Process ID (PID).
- A **job** is a shell’s representation of a pipeline or command. The shell tracks jobs in its own table, independent of PIDs. One job can contain multiple processes (e.g., `ls | sort`).

### 2. Foreground, Background, and Suspended

When you type a command and press Enter, the shell starts the process and waits for it to finish, occupying the terminal. That’s a **foreground** job.

If you append `&`, the command runs in the **background**, and the shell immediately returns the prompt. You see a job number and PID:
```bash
sleep 30 &
[1] 1234
```

Pressing **Ctrl+Z** sends the `SIGTSTP` signal, which suspends (pauses) the foreground job. The job is still alive but not executing.

### 3. The `sleep` Command as a Safe Test Target
`sleep N` does nothing for N seconds. It’s perfect for practicing job control without risk.

### 4. Job Control Commands

- `jobs` – lists all jobs of the current shell with their states (Running, Stopped).
- `fg %job` – bring a background or stopped job to the foreground. If no argument, the current job (marked `+`).
- `bg %job` – restart a stopped job in the background.
- `kill %job` – send a signal to a job (default `SIGTERM`).

Job numbers are shell‑specific; use `%1`, `%2`, etc., or `%%` for the current job.

Example:
```bash
sleep 300 &
[1] 5678
jobs
[1]+  Running                 sleep 300 &
fg %1
# (sleep 300 now occupies terminal)
# Press Ctrl+Z to suspend
[1]+  Stopped                 sleep 300
bg %1
# (resumes in background)
```

### 5. Signals and the `kill` Command

Signals are software interrupts sent to processes. The `kill` command sends them.

Common signals:
| Signal | Number | Default Action | Description |
|--------|--------|----------------|-------------|
| SIGTERM | 15 | Terminate gracefully | Politely asks process to exit (can be caught/ignored). |
| SIGKILL | 9 | Terminate immediately | Cannot be ignored; forceful kill. |
| SIGINT | 2 | Interrupt | Sent by Ctrl+C. |
| SIGTSTP | 20 | Stop (suspend) | Sent by Ctrl+Z. |
| SIGSTOP | 19 | Stop | Cannot be ignored; for debugging. |
| SIGCONT | 18 | Continue | Resume a stopped process. |

Syntax:
- By PID: `kill -9 1234` (send SIGKILL to PID 1234)
- By job: `kill -2 %1` (send SIGINT to job 1)

When you press Ctrl+C, the terminal sends SIGINT to the foreground process group. The process can trap it and clean up, or it dies with exit code `130` (128+2).

In the lecture, running `sleep 30` and then killing it with `kill -9` from another terminal demonstrated forceful termination. In the same shell, you can kill a background job: `kill %1`.

### 6. The `top` Command – Real‑time Process Monitor

`top` displays system summary and a dynamic list of processes sorted by CPU usage. While running:
- `q` – quit.
- `k` – kill a process (prompts for PID and signal).
- `r` – renice (change priority).
- `1` – toggle per‑CPU statistics.
- `P` – sort by CPU, `M` – by memory.

Pressing Ctrl+Z suspends `top`; you can then bring it back with `fg`. Pressing Ctrl+C kills it.

In the lecture, OBS Studio was seen consuming 63% CPU, illustrating how `top` helps identify resource hogs.

### 7. Process Hierarchy and `ps`

Every process has a parent (PPID). The init process (PID 1) is the ancestor. `ps --forest` or `ps axf` draws a tree.

```bash
ps --forest
```
Shows bash → sleep (child). This is invaluable for understanding process relationships.

### 8. Special Shell Variables for Process Management

- `$$` – PID of the current shell.
- `$!` – PID of the most recently started background job.
- `$?` – exit code of the last foreground command.
- `$-` – current shell option flags (explained below).

Example:
```bash
sleep 60 &
echo $!    # prints PID of the sleep command
kill $!    # kill it
```

### 9. Shell Options and `$-`

The special variable `$-` shows the active shell options. Common ones:
- `h` – remember command locations (hash).
- `i` – interactive shell.
- `m` – monitor mode (job control enabled).
- `B` – brace expansion enabled.
- `H` – history expansion enabled (the `!` operator).

You can change options with `set`. For example, `set +B` disables brace expansion; `set -B` enables it.

#### History Expansion (`!`)
`!number` re‑executes the command from the history with that number.
`history` lists commands with numbers. `!!` repeats the last command.

#### Brace Expansion
`{a,b,c}` expands to separate words. `{1..5}` expands to `1 2 3 4 5`. `{A..Z}` expands to all uppercase letters.
```bash
echo file{1,2}.txt   # file1.txt file2.txt
echo {A..C}{1..2}    # A1 A2 B1 B2 C1 C2
```
Brace expansion is performed before filename wildcards. It’s a pure shell feature.

### 10. Coprocesses (`coproc`)

`coproc` runs a command asynchronously in the background but with two‑way communication pipes attached to it. It’s a shell builtin (check with `type coproc`). The lecture used `coproc sleep 10` to run a command in the background and immediately regain the prompt. More complex uses involve reading/writing to the coprocess’s file descriptors. It’s an advanced feature for shell scripting when you need persistent interaction with a subprocess.

### 11. Foreground, Background, Suspended – The Job Lifecycle

A job can be:
- Running in foreground – terminal occupied.
- Running in background – prompt available.
- Stopped (suspended) – not consuming CPU, waiting for `fg`/`bg`.

`jobs` shows states. You can have many jobs; the shell only has one foreground job at a time. The `+` in `jobs` output marks the current job; `-` marks the previous job.

To resume a stopped job in the background, use `bg`. To kill a stopped job, bring it to the foreground and Ctrl+C, or `kill %job`.

### 12. Practical Example of Job Control Workflow

```bash
# Start a long computation
./heavy_simulation &
# Check that it's running
jobs
# Suspend it if you need to do something quick
fg %1
# Ctrl+Z
# Do something else
# Resume in background
bg %1
# Later, kill if needed
kill %1
```
Mastering job control keeps you in the terminal, seamlessly switching between tasks without opening multiple terminal windows.

# In‑Depth Guide to Pattern Matching, Regular Expressions, and Text Processing with `grep` and `cut`

The two lectures on pattern matching equip you with one of the most powerful skills in the Linux command‑line toolkit: **describing text patterns precisely** and using those descriptions to search, filter, and extract information from files and command output. This guide explains every concept from the ground up, including the **POSIX regular expression standard**, the `grep` family, the `cut` command, and how to combine these tools into efficient pipelines.

---

## 1. What is a Regular Expression?

A **regular expression** (regex or regexp) is a sequence of characters that defines a **search pattern**. Rather than searching for a fixed, literal string, a regex allows you to describe a *class* of strings that share a common structure: for example, “a line that starts with `M`, followed by any two characters, then `a`”, or “a 12‑digit number bounded by non‑word characters”.

Regular expressions are used extensively in:
- Text editors (`vi`, `emacs`, `sed`, `awk`)
- Command‑line tools (`grep`, `egrep`, `find`)
- Programming languages (Perl, Python, Ruby, Java, JavaScript)
- Database query languages (MySQL’s `REGEXP`, PostgreSQL’s `~`)
- Web servers (URL rewriting, input validation)

Because regex syntax is a small language of its own, the POSIX standard (IEEE 1003.1‑2001) defines two portable flavours:
- **BRE** – Basic Regular Expression
- **ERE** – Extended Regular Expression

Most tools default to BRE; adding the `-E` option (or using `egrep`) selects ERE.

---

## 2. The `grep` Command

`grep` (**G**lobal **R**egular **E**xpression **P**rint) is the primary tool for searching text using regular expressions. Its core operation is simple: it reads input line by line, applies the regex to each line, and **prints** lines that match.

### 2.1 Invocation
```bash
grep [options] pattern [file...]
```
If no file is given, `grep` reads from standard input (useful in pipelines).

### 2.2 Variants
- `grep` – uses BRE by default.
- `egrep` (or `grep -E`) – uses ERE.
- `fgrep` (or `grep -F`) – treats the pattern as a fixed string, not a regex (faster, no special characters).

### 2.3 Useful Options
- `-i` – ignore case.
- `-v` – invert match: print lines that **do not** match the pattern.
- `-c` – print only a count of matching lines.
- `-n` – prefix each line with its line number.
- `-o` – print only the matched part(s) of a line, not the whole line.
- `-w` – match whole words only.
- `-E` – use extended regular expressions.

### 2.4 Quoting Patterns
Regex patterns often contain characters that are also special to the shell (`*`, `?`, `$`, `(`, `)`, `{`, `}`, `|`, etc.). To prevent the shell from interpreting them, always **single‑quote** the pattern:
```bash
grep 'M[ME]' names.txt
```
Single quotes preserve the literal meaning of every character inside. This is a crucial habit.

## 3. Anatomy of a Regular Expression

A regex is built from **literal characters** (which match themselves) and **metacharacters** (which have special meaning).

### 3.1 The Dot `.`
Matches **any single character** except the newline character (and, in POSIX, the NUL character). It is the wildcard for exactly one character.

Example: `S.n` matches `Sin`, `San`, `Sun`, `S n`, etc., but **not** `Sn` (needs exactly one character between).

### 3.2 The Star `*`
In regex, `*` is a **quantifier** that applies to the **preceding element** (character or group). It means **zero or more** occurrences of that element.

- `ab*c` matches `ac`, `abc`, `abbc`, `abbbc`, etc.
- `.*` matches **any sequence of characters** (including the empty string) – the classic “match anything” pattern.

⚠️ This is a major difference from shell globbing (`*.txt`), where `*` matches any characters. In regex, `*` does nothing by itself; it modifies the preceding atom.

### 3.3 Square Brackets `[...]` – Character Classes
A pair of square brackets defines a **bracket expression** that matches **exactly one character** from the set inside.

- `[abc]` matches `a`, `b`, or `c`.
- `[a-z]` matches any lowercase letter (range, based on the locale’s collating sequence).
- `[0-9]` matches any digit.
- `[A-Za-z]` matches any alphabetic character.
- `[^abc]` – a leading `^` **negates** the set: matches any character **except** `a`, `b`, `c`.

Ranges can be combined: `[a-zA-Z0-9_]` matches any alphanumeric or underscore.

### 3.4 Anchors: `^` and `$`
Anchors do not consume characters; they assert a position.
- `^` – beginning of a line.
- `$` – end of a line.

Thus `^abc` matches lines starting with `abc`, and `xyz$` matches lines ending with `xyz`. Together, `^abc$` matches a line that contains **exactly** `abc`.

### 3.5 Escaping with Backslash `\`
To match a metacharacter literally, precede it with `\`. For example, `\.` matches a literal dot, `\*` a literal star, `\\` a literal backslash.

### 3.6 Word Boundaries
In `grep`, `\b` (or sometimes `\<` and `\>`) denotes a word boundary – the position between a word character (alphanumeric or underscore) and a non‑word character. This is a GNU extension (not strictly POSIX, but widely supported).

Example: `\bam\b` matches the word `am` but not `am` inside `exam` or `ampersand`. This is invaluable for extracting fields like pin codes or Aadhar numbers that are surrounded by whitespace or punctuation.

## 4. BRE – Basic Regular Expressions

In BRE, the following metacharacters are special: `. * [ ] ^ $ \`. All others are ordinary unless escaped.

### 4.1 Grouping and Backreferences
BRE uses **escaped parentheses** `\(` and `\)` to group parts of a pattern.
- The group can be followed by a quantifier (e.g., `\{2\}`).
- The matched text is captured and can be referenced later using **backreferences**: `\1`, `\2`, ..., `\9`.

Example: `\(ma\).*\1` matches a line that contains `ma` twice (any characters in between). `\1` refers to the actual characters captured by the first group.

Backreferences match the **literal string** that was captured, not the pattern again. That’s a powerful way to detect repeated words or structures.

### 4.2 Repetition with Escaped Braces `\{m,n\}`
In BRE, the interval expression requires backslashes.
- `X\{2\}` – exactly 2 occurrences of `X`.
- `X\{1,3\}` – between 1 and 3 occurrences.
- `X\{3,\}` – at least 3 occurrences.
- `X\{,5\}` – at most 5 occurrences (not always supported).

Example: `[0-9]\{12\}` matches exactly 12 consecutive digits.

### 4.3 BRE Special Characters Summary
| Metacharacter | Meaning |
|---------------|---------|
| `.` | Any single character |
| `*` | Zero or more of preceding element |
| `[...]` | Character class |
| `^` | Start of line anchor |
| `$` | End of line anchor |
| `\` | Escape next character |
| `\(` `\)` | Grouping / capture |
| `\n` | Backreference (n = 1–9) |
| `\{m,n\}` | Interval expression |

---

## 5. ERE – Extended Regular Expressions

ERE (used by `egrep` or `grep -E`) removes the need for many backslashes and adds new operators. The syntax is closer to that used in programming languages.

### 5.1 Grouping and Alternation
- Parentheses `(` `)` are grouping characters without escaping.
- The pipe `|` means **alternation** (logical OR). `(pattern1|pattern2)` matches either pattern.

Example: `^(ME|ED)` matches lines starting with `ME` or `ED`.

### 5.2 Quantifiers Without Backslashes
- `+` – one or more of the preceding element.
- `?` – zero or one of the preceding element (optional).
- `{m,n}` – interval expression (exactly as written, no backslashes).

Thus `M+` matches one or more `M` characters; `colou?r` matches both `color` and `colour`.

### 5.3 ERE Metacharacters Summary
| Metacharacter | Meaning |
|---------------|---------|
| `. * ^ $ [ ] \` | Same as BRE |
| `( )` | Grouping (capturing) |
| `\|` | Alternation |
| `+` | One or more |
| `?` | Zero or one |
| `{m,n}` | Interval |

**Note:** ERE does **not** support backreferences (though many implementations like GNU `grep` do as an extension). Strictly POSIX ERE has no `\1`.

### 5.4 When to Use Which
- Use **BRE** when you need backreferences and want to stay strictly POSIX‑compliant.
- Use **ERE** for cleaner syntax when you need `+`, `?`, `|`, and unescaped intervals.

## 6. POSIX Character Classes

Inside a bracket expression, you can use named character classes of the form `[:class:]`. These must be placed inside another pair of brackets: `[[:class:]]`.

Common classes:
| Class | Meaning |
|-------|---------|
| `[:alnum:]` | Alphanumeric (letters and digits) |
| `[:alpha:]` | Alphabetic characters |
| `[:blank:]` | Space and tab |
| `[:cntrl:]` | Control characters |
| `[:digit:]` | Digits 0‑9 |
| `[:graph:]` | Visible characters (excluding space) |
| `[:lower:]` | Lowercase letters |
| `[:print:]` | Printable characters (includes space) |
| `[:punct:]` | Punctuation characters |
| `[:space:]` | Whitespace (space, tab, newline, etc.) |
| `[:upper:]` | Uppercase letters |
| `[:xdigit:]` | Hexadecimal digits |

Example: `[[:digit:]]{6}` matches exactly 6 digits.  
To match a line starting with an uppercase letter: `^[[:upper:]]`.

The lecture demonstrated these by filtering a file containing lines with various character types (control characters, punctuation, digits, etc.) and using anchors to require matches at the start or end of the line. The `-v` option inverted the match, e.g., to show lines **without** control characters.

## 7. Operator Precedence

Understanding precedence helps to craft correct regex patterns without excessive parentheses. From highest to lowest (BRE/ERE):

1. **Collation** and escaped characters
2. **Bracket expansion** `[...]`
3. **Grouping** `\(\)` / `()`
4. **Backreferences** `\1` (BRE only)
5. **Quantifiers** `*`, `+`, `?`, `\{m,n\}` / `{m,n}`
6. **Concatenation** (adjacent elements)
7. **Anchors** `^`, `$`
8. **Alternation** `|` (ERE only, lowest)

So `^abc|def$` means “a line starting with `abc` OR ending with `def`”. If you want “a line starting with `abc` or `def`”, you need parentheses: `^(abc|def)`.

## 8. Practical Pattern‑Matching Demos

### 8.1 Simple String Matching
```bash
grep 'Raman' names.txt
```
Prints lines containing `Raman`. Case‑sensitive by default.

### 8.2 The Dot as a Wildcard
- `S.n` matches `Sin`, `San` etc.
- `\.` matches a literal dot (e.g., initials `M.S.`). To match an initial (one letter followed by a dot), use `[A-Z]\.`.

### 8.3 Anchoring to Line Start/End
- `^M` – lines starting with capital M (e.g., roll numbers MM, ME).
- `am$` – lines ending with `am` (names like `Manickam`).

### 8.4 Word Boundaries
- `\bam\b` matches `am` as a whole word.
- `\b[0-9]{6}\b` matches exactly a 6‑digit number surrounded by non‑word characters. This is how you extract a PIN code without catching longer numbers.

### 8.5 Character Classes and Ranges
- `[ME]` – matches either `M` or `E`. Useful for `MM` or `ME` roll number prefixes.
- `[a-z]` – any lowercase letter.
- `[^5-7]` – any character **not** in the range 5–7 (negation).

The demo found students whose roll numbers contained `B90[1-4]` vs `B90[5-7]`.

### 8.6 Repetition (Intervals)
**BRE**:
- `M\{2\}` – exactly two `M`s.
- `M\{1,2\}` – one or two `M`s.

**ERE**:
- `M{2}` – exactly two `M`s.
- `[0-9]{12}` – exactly twelve digits (Aadhar).

### 8.7 Grouping and Backreferences
In BRE, to find lines containing a pattern repeated:
```bash
grep '\(ma\).*\1' names.txt
```
This matches `ma` ... `ma` (e.g., `Umair Ahmad`). The backreference `\1` reproduces whatever the first group captured.

If you use `\(.\).*\1`, it captures any character and then finds a later occurrence of the same character. The demo used `a\(.\)` to capture a character following `a` and then look for the same pair later, e.g., `Sankaran` (the `a`+consonant pair repeats).

### 8.8 ERE Enhancements: `+`, `?`, `|`
- `M+` – one or more `M`s.
- `(ma)+` – the sequence `ma` occurring one or more times.
- `^M.*a` vs `^M.*a` with `*` operating on `M`: `M*` means zero or more `M`s, so `^M*a` would also match lines where `a` is at the start without a preceding `M`. Use `^M.*a` to ensure at least one `M` before the `a`.
- Alternation: `^(ME|ED)` matches lines starting with `ME` or `ED`. The pipe has the lowest precedence, so parentheses are essential when anchoring.

### 8.9 Advanced Filtering in Package Lists
The lecture combined `dpkg-query` with `grep` to explore installed packages.

- List all packages with their section:
  ```bash
  dpkg-query -W -f='${Section} ${binary:Package}\n' | sort | less
  ```
- Show packages whose name is exactly 4 characters long (using ERE):
  ```bash
  ... | grep -E ' [[:alnum:]]{4}$'
  ```
  The leading space ensures the match starts at the package name, not the section field. `{4}` enforces exactly four characters, and `$` anchors the end of the line.

- Show packages starting with `g` and of length 4:
  ```bash
  ... | grep -E ' g[[:alnum:]]{3}$'
  ```

- Show packages in the `math` section only:
  ```bash
  ... | grep '^math '
  ```

- Show packages related to `kde`:
  ```bash
  ... | grep -i '^kde'
  ```
These examples illustrate how field‑aware regex can precisely extract information from semi‑structured text.

### 8.10 Filtering a File of Mixed Character Types
The file `chartypes.txt` contained lines with different characters (letters, digits, control characters, punctuation). The lesson filtered for:
- `[[:alpha:]]` – any line containing a letter.
- `[[:alnum:]]` – any line with alphanumeric.
- Anchored versions to test start/end of line.
- Control characters: `[[:cntrl:]]`.
- Punctuation: `[[:punct:]]`.
- Using `-v` to show lines **without** the pattern, e.g., removing empty lines (`^$`).

Empty lines were removed with `grep -v '^$'` – matching lines that consist of nothing between start and end, then inverting.

### 8.11 Extracting Patterns from Unstructured Text
A file contained random text with embedded:
- Aadhar (12 digits): `[0-9]{12}`
- PIN code (6 digits): `\b[0-9]{6}\b` (word boundaries prevent matching a substring of Aadhar)
- Phone number (10 digits): `\b[0-9]{10}\b`
- Roll number (alphanumeric, e.g., `MM21B001`): `\b[[:alpha:]]{2}[0-9]{2}[[:alpha:]][0-9]{3}\b`
- URLs (no protocol, just `word.word`): `[[:alnum:]_]+\.[[:alnum:]_]+`

For roll numbers, a custom regex matching the structure (two letters, two digits, one letter, three digits) was required. This shows that real‑world data often needs a mix of character classes, quantifiers, and anchors to avoid false positives.

For URLs, `[[:alnum:]_]+\.[[:alnum:]_]+` uses `+` (ERE) to ensure at least one character before and after the dot, and `\.` to match a literal dot. This captured `example.com` and `sub.domain.co`.

## 9. The `cut` Command – Horizontal Trimming

While `head` and `tail` trim lines vertically (top/bottom), `cut` trims horizontally by **characters** or **fields**.

```bash
cut OPTION... [FILE]...
```

### 9.1 Cut by Character Position (`-c`)
```bash
cut -c 1-4 file.txt    # characters 1 through 4 of each line
cut -c 5-8 file.txt    # characters 5–8
cut -c 5-             # from character 5 to end of line
cut -c -10            # first 10 characters
```
Multiple ranges can be separated by commas, e.g., `-c 1-3,5-7`.

This is independent of field boundaries; it simply counts bytes/characters.

### 9.2 Cut by Fields (`-f`) with a Delimiter (`-d`)
- `-d` specifies the field delimiter (default is TAB).
- `-f` selects fields by number (1‑based). Ranges and commas are allowed.

Example:
```bash
cut -d ':' -f 1 /etc/passwd   # extract usernames
```

In the lecture, a file `fields.txt` contained lines like:
```
1;hello world,line 1
2;welcome command line,line 2
3;linux tools,line 3
```

- Delimiter `;`:
  - Field 1: the number
  - Field 2: `hello world,line 1`
- Delimiter `,`:
  - Field 1: the text before comma
  - Field 2: `line 1` etc.

Using `cut -d ';' -f 1` gave the numbers. Using `cut -d ';' -f 2 | cut -d ',' -f 1` extracted the string between the semicolon and the comma (e.g., `hello world`). This demonstrates **piping multiple `cut` commands** to progressively refine the output.

### 9.3 Combining `cut`, `head`, `tail`
To extract a specific piece of text, you can combine vertical and horizontal slicing:
```bash
cat fields.txt | cut -d ';' -f 2 | cut -d ',' -f 1 | head -n 2 | tail -n 1
```
This:
- Takes the second semicolon‑delimited field.
- Takes the first comma‑delimited field of that result.
- Keeps the first two lines.
- Takes the last of those two lines (i.e., the second line).

The result is the exact string “welcome command line”. This pipeline illustrates the Unix philosophy: each tool does one thing, and they can be chained to achieve complex extractions.

## 10. Practical Applications and Best Practices

1. **Always quote your pattern** with single quotes to prevent shell expansion.
2. **Test with `grep --color=auto`** or pipe to `less -R` to see what matched.
3. Use **`-E`** for complex patterns to avoid backslash clutter.
4. **Word boundaries** are your friend when you need to isolate numbers or words from surrounding text.
5. **`grep -v`** is excellent for removing unwanted lines (comments, blanks, headers).
6. **`cut`** is best for simple delimited data; for more complex field‑based transformations, `awk` is the next step.
7. **Regex greediness**: quantifiers are greedy by default. `.*` will match as much as possible. In POSIX regex, there is no non‑greedy option (Perl introduces `*?`).
8. **Locale awareness**: ranges like `[a-z]` depend on the locale. Use `LC_ALL=C` for predictable ASCII‑only matching when needed.
9. **Backreferences** are not part of strict ERE, but GNU `grep -E` supports them as an extension. Be aware of portability if writing POSIX‑compliant scripts.

## 11. Summary of Key Concepts

- **Regex** is a pattern language; **`grep`** is the search tool.
- **BRE** and **ERE** differ mainly in the need for backslashes and the availability of `+`, `?`, `|`.
- **Metacharacters** `. * [ ] ^ $ \` are the core building blocks.
- **Quantifiers** `*`, `+`, `?`, `{m,n}` specify how many times an element repeats.
- **Grouping** `()` captures for backreferences (BRE) or for applying quantifiers/alternation (ERE).
- **Anchors** `^`, `$`, `\b` fix the match to positions.
- **Character classes** `[...]` and POSIX named classes `[[:...:]]` provide flexible character sets.
- **`cut`** extracts columns by character positions or delimited fields.
- Pipelines of `grep`, `cut`, `head`, `tail`, `sort`, etc., can perform sophisticated text processing without writing a program.

With these skills, you can sift through log files, extract configuration values, validate input, and transform data streams – all from the comfort of the command line. The command line is not just a tool; it is a programmable environment where regular expressions and filters turn raw text into actionable information.
