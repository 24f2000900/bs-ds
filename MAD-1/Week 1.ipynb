## 1. What is an App?

### Definition and Evolution
The term “app” is a shorthand for **application software** – a computer program designed to carry out a specific task for the user. Originally, the word “application” referred to any software that ran on a desktop or laptop computer: word processors, spreadsheets, games, etc. The shortened form “app” gained massive popularity with the rise of **smartphones** and **app stores** (especially Apple’s App Store in 2008). Today, an app is commonly understood as:

> A self-contained piece of software, typically small and focused, that can be downloaded and installed on a mobile device, but increasingly also on desktops, web browsers, and embedded devices.

The Techopedia definition cited in the lecture captures this: *“An app is computer software or a program, most commonly a small, specific one used for mobile devices.”*

**Key characteristics of an app:**
- **Specific goal:** Unlike a full operating system, an app solves a particular problem (e.g., ordering food, editing photos, sending messages).
- **Self-contained distribution:** Apps are packaged as a single unit that users can install (via an app store or direct download).
- **User-centric interface:** They are designed for direct interaction, often through touch, keyboard/mouse, or voice.

### Apps vs. Applications: The Blurring Boundary
While “app” once implied a lightweight mobile program and “application” a heavyweight desktop program, the distinction has faded:
- Microsoft Word is available as a full desktop application, a mobile app, and a web app.
- Google Docs is primarily a web app, yet feels like a full application.
- Many “apps” today, like Adobe Photoshop Express, are scaled‑down versions of their desktop counterparts, but they still perform complex tasks.

Thus, in this course, the terms *app* and *application* are used interchangeably. The focus is on understanding the principles behind building any application, regardless of its final form factor.

### Categories of Apps by Platform
The lecture categorises apps by the **platform** on which they run. Each platform imposes different constraints and offers different capabilities.

#### 1. Desktop Apps
- **Environment:** Run on a personal computer (Windows, macOS, Linux) with a keyboard, mouse, and large display.
- **Characteristics:** Often **standalone** – can function without a network connection. Examples: text editors (VS Code, Notepad), word processors, IDEs, media players.
- **Development:** Traditionally built using OS‑specific **Software Development Kits (SDKs)** and APIs (e.g., Win32 API for Windows, Cocoa for macOS). This ties the application to a particular operating system.
- **Offline capability:** They rely heavily on local storage. Even an email client like Outlook downloads messages to the local machine (using POP3 or IMAP), allowing offline reading and composition; it only needs the network to send or sync.
- **Paradigm:** The desktop metaphor – files, folders, windows – mimics a physical desk, making it intuitive for office‑type work.

#### 2. Mobile Apps
- **Devices:** Smartphones and tablets (iOS, Android). The number of phones vastly exceeds that of tablets.
- **Constraints:**
  - **Screen size:** Limited real estate forces careful UI design.
  - **Input:** Touch replaces keyboard/mouse; no hover states. New modalities: multi‑touch gestures, accelerometer, gyroscope, GPS, cameras, microphones.
  - **Resources:** Less RAM, slower processors, and restricted power budgets because they run on batteries that must last a full day.
- **Network dependency:** Most mobile apps are inherently network‑oriented (social media, shopping, maps) because the device is always connected. Pure offline apps are rarer.
- **Development:**
  - **Native:** iOS apps are typically written in Swift/Objective‑C with Apple’s SDKs; Android apps in Kotlin/Java with Android SDK. They have deep access to hardware but are platform‑specific.
  - **Cross‑platform frameworks:** Technologies like React Native, Flutter, and Xamarin allow a single codebase to target both iOS and Android, trading some performance and native look‑and‑feel for faster development.

#### 3. Web Apps
- **What they are:** Applications that run inside a web browser. The browser (Chrome, Firefox, Safari, Edge) acts as a universal runtime.
- **Why the web is special:**
  - **Cross‑platform by nature:** Browsers are available on all major operating systems, including mobile. The web abstracts away the underlying hardware and OS differences.
  - **No installation:** Users just visit a URL; updates are instantaneous on the server side.
  - **Network‑centric:** Data and business logic typically reside on remote servers. This enables access from anywhere and easy collaboration.
- **Evolution:** The early web (Web 1.0) served static pages. With **Web 2.0** (mid‑2000s), dynamic, interactive web applications became possible thanks to technologies like **JavaScript (client‑side logic)** , **AJAX (asynchronous data loading)** , and powerful server‑side frameworks. Now, web apps can rival native apps in functionality (e.g., Google Docs, Figma).
- **This course’s focus:** The web will be the primary platform because it offers a clean, platform‑agnostic environment to teach the fundamental principles of app development: storage, computation, and presentation – without getting entangled in OS‑specific details.

#### 4. Embedded Apps
- **Definition:** Software that runs on dedicated, single‑function devices with limited resources: smart TVs, microwave ovens, refrigerators, digital watches, automotive control units.
- **Characteristics:**
  - **Fixed function:** The device does one thing (e.g., a camera takes pictures), often extremely well.
  - **Resource‑constrained:** Low memory, slow processors, minimal or no user interface.
  - **“Invisible” apps:** Many embedded systems have no screen; they communicate via simple LEDs, network protocols, or machine‑to‑machine interfaces.
- **Modern shift:** Devices are increasingly “smart” (e.g., smartwatches running Wear OS, smart fridges with grocery‑ordering apps). The line between embedded and mobile/web is blurring, but the design philosophy remains: minimal, efficient, and purpose‑built.

### Apps as Icons: The User’s View
The lecture shows familiar icons – Amazon, Twitter, Word, Chrome, VS Code, etc. Each icon represents an entry point into a distinct app. Notably, **browsers** and **text editors** are both apps themselves and tools for creating other apps. For example, a browser renders web apps; a text editor is used to write the code that becomes an app. The terminal (command line) is another critical tool that developers must master to run scripts, manage servers, and automate tasks.

## 2. Components of an App

Every application, regardless of platform, can be decomposed into three fundamental concerns: **storage**, **computation**, and **presentation**. This separation helps in designing, building, and maintaining software.

### 1. Storage (Data)
- **What is stored:** Any persistent information the app needs to retain between sessions – user profiles, documents, emails, settings, images, transaction records.
- **Where storage happens:**
  - **Local storage:** On the user’s device (files on disk, local databases like SQLite, browser localStorage). Desktop apps often rely heavily on local storage.
  - **Cloud/Server storage:** Data is kept on remote servers (“in the cloud”). Web apps and most mobile apps use this so that data is accessible from any device and never lost if the user’s device fails.
- **How storage is organized:**
  - **File systems:** Simple hierarchical folders and files (e.g., text files, CSVs, configuration files). Easy to understand but lack sophisticated querying.
  - **Databases:** Structured collections of data with efficient indexing and querying capabilities.
    - **Relational databases (SQL):** Use tables with rows and columns, enforce schemas, and support powerful queries (e.g., MySQL, PostgreSQL). Good for structured data with complex relationships.
    - **NoSQL databases:** Document stores (MongoDB), key‑value stores (Redis), graph databases (Neo4j). Offer flexibility in schema and scale horizontally.
  - **Proprietary formats:** Applications like Microsoft Outlook store emails in large, indexed binary files (PST/OST) that are essentially databases but not directly readable by other software. Other clients use open, text‑based formats like mbox (all mails in one file) or maildir (one file per message) for portability.
- **Storage design considerations:**
  - **Durability:** Data must survive power loss and hardware failures.
  - **Scalability:** As the amount of data grows, can the storage system handle it without degradation?
  - **Security:** Who can read/modify the data? Encryption at rest and in transit.
  - **Access patterns:** Is the data read‑heavy, write‑heavy, or balanced? This influences the choice of database technology.

### 2. Computation (Business Logic)
- **What it does:** The “brain” of the app. It processes data, enforces rules, performs calculations, and makes decisions. This is often called **business logic** – the core functionality that gives the app its value.
- **Examples from an email client:**
  - **Indexing:** When a new email arrives, the server must update its full‑text search index so the user can later search by keyword, sender, date, etc. Indexing involves tokenizing text, removing stop words, and building inverted indexes (like a book’s index but for words).
  - **Filtering/Spam detection:** Algorithms analyze incoming messages to decide whether they are spam, applying rules (e.g., Bayesian filtering, blacklists) that must run quickly and accurately.
  - **Sorting and threading:** Emails can be sorted by date, sender, subject. Threading reconstructs conversation trees, grouping replies together.
  - **Notifications:** When a new mail arrives, the server must determine which users are affected and push a notification to their devices.
- **Where computation happens:**
  - **Server‑side:** On the web server or application server. The client sends a request, the server does the heavy lifting, and returns the result. This is typical for web apps and any task that requires access to a large central database or sensitive algorithms.
  - **Client‑side:** In the browser (JavaScript) or in a mobile/desktop app. This reduces server load and latency, but is limited by the client’s computing power and security constraints. Often used for UI logic, form validation, and real‑time interactivity.
- **Real‑world analogy:** The “controller” in the MVC pattern (discussed later) is the part that orchestrates computation – it decides what to do with user input and how to update the model.

### 3. Presentation (User Interface)
- **What the user sees and interacts with:** The visual layout, text, images, buttons, forms, and all feedback (animations, sounds, haptics).
- **Responsibilities:**
  - **Rendering:** Taking the data from the model and displaying it in a human‑readable, aesthetically pleasing manner. For an email client, this means showing a list of emails with subject, sender, date, and an indicator for unread status; displaying the full email body with proper formatting (HTML emails with images).
  - **Interactivity:** Capturing user actions (clicks, taps, swipes, voice commands) and translating them into requests that the rest of the app can understand.
- **Presentation technologies:**
  - **Web:** HTML for structure, CSS for styling, and JavaScript for dynamic behavior. The browser interprets this trio to render web pages.
  - **Desktop:** Native UI toolkits (Windows Presentation Foundation, Qt, Cocoa) or cross‑platform frameworks (Electron, which uses web technologies).
  - **Mobile:** Android’s XML‑based layouts, iOS’s SwiftUI, or cross‑platform frameworks that compile to native components.
- **Separation of concerns:** Keeping presentation logic separate from business logic (as in MVC) allows designers to change the look and feel without touching core functionality, and vice‑versa. It also enables the same backend to serve different frontends (e.g., a mobile app and a web app sharing the same server).

### The Email Client Example – Putting It All Together
To illustrate the three components, consider a modern email service like Gmail:

| Component | Gmail Implementation |
|-----------|----------------------|
| **Storage** | All emails are stored on Google’s servers in a massively distributed database (likely a custom NoSQL system). Metadata (sender, recipient, labels, read status) is also stored and indexed. Attachments are saved in a separate blob store. |
| **Computation** | When you hit “Send,” the server checks recipient validity, appends the message to the sender’s Sent folder, creates a copy in each recipient’s Inbox, updates search indexes, and runs spam checks. Searching through years of emails requires lightning‑fast queries across petabytes of data. |
| **Presentation** | The Gmail web interface shows your inbox as a scrollable list. Clicking a message opens a detailed view. The interface is built with HTML, CSS, and a huge amount of JavaScript that updates the page dynamically without full reloads. The same backend serves the Gmail mobile app (which uses native rendering but the same APIs). |

## 3. Client-Server and Peer-to-Peer Architecture

When an app involves multiple devices communicating over a network, we need an **architecture** that defines how they connect and share responsibilities.

### Client-Server Architecture
This is the dominant model for most web and mobile apps today.

**The three entities:**
1. **Server:** A powerful computer (or a cluster) that **stores data** and **performs computation**. It listens for incoming requests from clients and responds with the requested data or the result of a computation.
2. **Client:** The end‑user’s device (laptop, phone, tablet) that initiates requests to the server. The client focuses on **presentation** and **user interaction**, sending user actions to the server and displaying the responses.
3. **Network:** The communication channel (the Internet, a local network) that connects clients and servers. Ideally, it acts as a transparent pipe – simply transporting data without modification.

**How it works:**
- The server is always on, waiting for requests.
- A client opens a connection, sends a structured request (e.g., “Get me the list of my unread emails”).
- The server processes the request, possibly querying a database or running business logic.
- The server sends back a response (e.g., the list of emails as data).
- The client receives the data and renders it for the user.

**Key characteristics:**
- **Explicit differentiation:** At any given moment, you can clearly identify who is the server and who is the client. The server does not initiate communication; it only responds.
- **Centralisation:** Data and logic are concentrated in the server. This makes it easier to secure, update, and scale (within limits).
- **Examples:**
  - **Email (IMAP/SMTP):** An email server (e.g., Gmail’s server) stores your mail; your email client (Outlook, Apple Mail, web browser) fetches and displays it.
  - **Web browsing:** The browser (client) requests a page from a web server (server).
  - **WhatsApp:** Messages are sent to WhatsApp’s servers, which then route them to the recipient(s). The server also stores undelivered messages.
  - **Databases:** A database server (MySQL, PostgreSQL) runs continuously; multiple client applications connect to it, send SQL queries, and receive results.
  - **Anti‑virus updates:** The anti‑virus software acts as a client, periodically checking a server for new virus definition files and downloading them without explicit user interaction.
- **Variations:**
  - **Single server, multiple clients:** The simplest model.
  - **Multi‑tier architecture:** The server itself might be split into a **web server**, an **application server**, and a **database server**, each running on different machines but collectively acting as “the server” to the client.
  - **Load balancers:** A front‑end server that distributes incoming client requests among several back‑end servers to handle high traffic.

### Peer-to-Peer (P2P) Architecture
In a P2P network, there is no central server that owns all resources. Instead, every participant (peer) acts as both client and server, sharing its own resources (storage, bandwidth, processing) and consuming resources from others.

**How it works:**
- Peers connect to each other directly or through a lightweight introducer service.
- Data is often fragmented and distributed across many peers. When you request a file, your client fetches pieces from multiple peers simultaneously, and you in turn upload pieces you already have to others.
- The system is self‑organising and can tolerate failures: if one peer goes offline, others can still serve the data.

**Key characteristics:**
- **No central authority:** There is no single point of control or failure. This makes the system resistant to censorship and server crashes.
- **Scalability:** The network becomes stronger as more peers join, because each new peer adds resources.
- **Trust and security challenges:** How do you ensure the data you receive is authentic and not corrupted? Solutions include cryptographic hashing, digital signatures, and blockchain‑style consensus.

**Examples:**
- **BitTorrent:** The classic file‑sharing protocol. A .torrent file points to a tracker (or uses a distributed hash table) that helps peers find each other. Files are downloaded in small pieces from many peers.
- **Blockchain and Cryptocurrencies:** A distributed ledger where each node holds a copy of the entire chain. No central bank; consensus algorithms validate transactions.
- **Distributed File Systems:** **IPFS (InterPlanetary File System)** and **Tahoe‑LAFS** allow files to be stored across a network of nodes, with content addressed by its hash. Data can be encrypted so that only authorised parties can decrypt it, even though it physically resides on strangers’ machines.
- **Skype (original):** Initially used a P2P network for voice calls, with supernodes helping to route traffic, before moving to a more centralised model.

### Comparison

| Aspect | Client-Server | Peer-to-Peer |
|--------|---------------|--------------|
| **Control** | Central server controls data and logic | Distributed; all peers are theoretically equal |
| **Scalability** | Limited by server capacity; requires load balancers and more servers to scale | Scales organically; each new peer adds capacity |
| **Reliability** | Server is a single point of failure; requires redundancy | Robust against failures; no single point of failure |
| **Security** | Easier to enforce access control and data integrity | Harder; requires cryptography and reputation systems |
| **Suitable for** | Applications needing central authority, user accounts, consistent data (email, e‑commerce, banking) | File sharing, content distribution, decentralised apps (dApps) |

In this course, we will focus exclusively on the **client-server** model because it aligns with how web applications are built: a browser (client) communicates with a web server (server) using HTTP.

## 4. Software Architectures (Design Patterns & MVC)

### What is a Design Pattern?
A **design pattern** is a general, reusable solution to a commonly occurring problem within a given context in software design. It is not a finished piece of code but a **template** or **best practice** that developers can adapt to their specific situation.

**Origins:**
- The concept was popularised by the “Gang of Four” (Gamma, Helm, Johnson, Vlissides) in their 1994 book *Design Patterns: Elements of Reusable Object‑Oriented Software*.
- Patterns arise from the collective experience of seasoned developers who have observed that certain structures lead to more maintainable, flexible, and understandable code.

**Benefits:**
- **Shared vocabulary:** Developers can communicate complex ideas succinctly (“We’ll use an Observer pattern here”).
- **Proven solutions:** They embody wisdom about what works and what pitfalls to avoid.
- **Design guidance:** They help beginners think about problems in a structured way.

### Model-View-Controller (MVC)
MVC is one of the most influential architectural patterns, originally introduced in the **Smalltalk‑80** programming environment in 1979. It partitions an application into three interconnected components:

1. **Model** – Manages the **data and business logic**.
   - Contains the core information that the application deals with.
   - Enforces the rules and constraints on that data (e.g., an email must have a valid recipient address).
   - Notifies the View when data changes so the View can update itself.
   - Examples: the list of emails and their metadata, the algorithms for search indexing, the rules for spam detection.

2. **View** – Handles the **presentation**.
   - Displays the data from the Model to the user in a specific format.
   - Sends user actions (clicks, taps) to the Controller.
   - Is passive: it does not modify the Model directly; it only reflects the Model’s state.
   - In an email app: the inbox list, the mail reading pane, the compose window.

3. **Controller** – Acts as the **intermediary**.
   - Receives input from the user (via the View).
   - Interprets the input and translates it into actions on the Model (e.g., “delete this email”, “mark as read”).
   - May also instruct the View to update.
   - Often where the “business logic” of what to do with user requests lives, though in some interpretations the heavy logic is in the Model and the Controller only coordinates.

**The MVC Flow:**
1. User interacts with the View (clicks “Delete”).
2. View notifies the Controller about the event.
3. Controller processes the event, possibly validating input, then calls appropriate methods on the Model (e.g., `email.markAsDeleted()`).
4. Model updates its state and notifies the View (directly or indirectly) that data has changed.
5. View re‑renders itself using the updated Model data.

**Why use MVC?**
- **Separation of concerns:** Developers can work on the Model, View, and Controller relatively independently. A designer can tweak the View without breaking the core logic.
- **Testability:** The Model can be tested in isolation without any UI; Controllers can be tested with mock Views.
- **Multiple Views:** The same Model can be displayed in different ways (e.g., a bar chart and a table of sales data) by using different Views.

**Limitations and Variations:**
- **MVC is not one‑size‑fits‑all:** In complex modern web apps, strict MVC can become messy because the Controller often gets bloated. Alternative patterns have emerged:
  - **Model-View-Presenter (MVP):** The Presenter replaces the Controller. It has a more direct hand in updating the View, and the View is completely passive (often used in Android development).
  - **Model-View-ViewModel (MVVM):** Common in frameworks like Angular, React (with state management), and WPF. The ViewModel exposes data and commands in a way that the View can bind to directly, reducing the need for a separate Controller.
  - **Hierarchical MVC:** Allows nesting of MVC triads, where a parent View can contain child Views that each have their own mini‑MVC.
- **In web development:** MVC is often applied with:
  - **Model:** The database and the server‑side code that interacts with it.
  - **View:** The HTML templates that render data.
  - **Controller:** The server‑side code that receives HTTP requests, calls Model methods, and selects the appropriate View to return.
  - Client‑side JavaScript frameworks (Angular, React, Vue) implement their own versions of MVC/MVVM.

For this course, we adopt the **MVC mindset** because it provides a clear, logical way to organise code and reason about the components of an app, even if we later use frameworks that deviate slightly from the pure pattern.

### Summary of Course Focus
- **Platform:** Web (browser + server)
- **Network Architecture:** Client-Server
- **Software Architecture:** Model-View-Controller
- **Core Technologies:** HTTP, HTML, CSS, JavaScript, and server‑side programming.

## 5. Introduction to the Web

The **World Wide Web** (“the web”) is an information system where documents and other resources are identified by URLs, interlinked by hypertext links, and accessible over the Internet. To understand the web as an application platform, we must first understand its foundations.

### Why the Web as a Platform?
- **Cross‑platform:** Works on Windows, macOS, Linux, iOS, Android, and even many embedded devices – any device with a modern browser. The browser abstracts hardware differences (screen size, input methods, OS).
- **Built on sound principles:** It has a simple, text‑based protocol (HTTP) and a declarative presentation language (HTML). These are well‑understood and have stood the test of time.
- **Access anywhere:** Data lives on servers, so a user can pick up any device, log in, and continue where they left off.
- **Cost‑effective deployment:** Updates are done on the server; users immediately see the latest version without downloading patches.
- **Constraints:** Web apps are limited by browser capabilities (no direct access to hardware like Bluetooth/USB without specific APIs), network dependence, and potential performance bottlenecks compared to native code. Nevertheless, for a huge class of applications, the web is ideal.

### Historical Foundations: From Circuit Switching to the Internet
To appreciate the web, we must understand the underlying network – the Internet.

**Circuit‑Switched Networks (Telephone model, 1890s)**
- When a call is made, a dedicated physical path (a circuit) is established end‑to‑end and reserved for the entire duration of the call.
- Advantages: guaranteed bandwidth, consistent latency.
- Disadvantages: horribly inefficient. Even during silence, the wires are tied up and cannot be used by anyone else. Scaling requires vast amounts of copper.

**Packet‑Switched Networks**
- Key idea: Break data into small **packets**. Each packet contains a payload (the actual data) and a header (source, destination, sequence number).
- Packets from many conversations share the same physical wires. Network devices (routers) read the destination and forward each packet hop‑by‑hop.
- Advantages: highly efficient use of resources. No dedicated circuit means the network can carry many times more traffic.
- Enables diverse data types – voice, video, text – all treated uniformly as bits.

**ARPANET (Late 1960s)**
- Funded by the US Advanced Research Projects Agency (ARPA).
- First large‑scale packet‑switched network. Connected universities and research labs.
- Introduced key concepts like host‑to‑host communication and the idea of an inter‑network.

**The Internet Protocol (IP) – January 1, 1983**
- Before IP, there were many incompatible networks (IBM’s SNA, DEC’s DECnet, Xerox’s XNS, etc.). They could not talk to each other.
- The Internet Protocol was designed as an **inter‑network** protocol: a “layer” on top of any underlying network technology. It defined a universal packet format and addressing scheme (IP addresses, originally 32‑bit values written as four decimal numbers: `192.168.1.1`).
- On “Flag Day”, January 1, 1983, ARPANET officially switched to TCP/IP, and the modern Internet was born.

**Transmission Control Protocol (TCP)**
- IP provides “best‑effort” delivery – packets can be lost, duplicated, or arrive out of order.
- TCP adds reliability on top of IP: it ensures data is delivered in order and without errors. It does this by:
  - **Acknowledgments:** The receiver sends back an ACK for every packet or group of packets. If the sender doesn’t get an ACK, it retransmits.
  - **Sequencing:** Each byte is numbered so the receiver can reassemble the stream correctly.
  - **Flow control:** Prevents a fast sender from overwhelming a slow receiver.
  - **Congestion control:** Algorithms (like TCP Reno) dynamically adjust sending rate to avoid network congestion.
- TCP/IP together form the fundamental transport layer of the Internet.

**Domain Name System (DNS) – 1985**
- IP addresses (e.g., `142.250.182.46`) are hard for humans to remember and type.
- DNS translates human‑friendly names (`google.com`) into IP addresses.
- It is a hierarchical, distributed database: root servers → top‑level domain servers (.com, .org) → authoritative servers for specific domains.
- DNS also allows for abstraction: the IP address of a server can change without the user ever knowing, and multiple IP addresses can be associated with one name for load balancing.

### The Birth of the Web: Hypertext and HTML
**Hypertext** is text that is not constrained to be linear. It contains **hyperlinks** that allow the reader to jump to other documents or sections. The idea predates computers (Vannevar Bush’s Memex, 1945; Ted Nelson coined the term in 1965).

**Tim Berners‑Lee at CERN (1989‑1990)** combined three key technologies to create the World Wide Web:
1. **HTML (HyperText Markup Language):** A simple language to structure documents. It uses tags to denote headings, paragraphs, lists, and most importantly, **links** (`<a href="...">`). HTML is plain text, so any text editor can create it.
2. **HTTP (HyperText Transfer Protocol):** A protocol for transferring HTML documents (and later, any file) over the Internet. It is a simple request‑response protocol: client sends a GET request with a URL, server responds with the file.
3. **URL (Uniform Resource Locator):** A string that uniquely identifies a resource on the web, e.g., `http://www.example.com/index.html`. It specifies the protocol, server, and path.

The first web browser, **WorldWideWeb** (later renamed Nexus), was also a WYSIWYG editor. The web was originally conceived as a collaborative medium for sharing research documents.

**Web 1.0 → Web 2.0**
- **Web 1.0 (early–mid 1990s):** Mostly static HTML pages. Servers simply served files. Interactivity was minimal; pages were read‑only brochures.
- **Web 2.0 (coined ~2004):** The web became a platform for **dynamic, interactive applications**.
  - **Server‑side processing:** Technologies like PHP, ASP.NET, Java Servlets allowed servers to generate HTML on‑the‑fly based on user input, database contents, etc.
  - **AJAX (Asynchronous JavaScript and XML):** Enabled web pages to fetch data in the background and update parts of the page without a full reload. This made applications like Google Maps possible.
  - **Rich client‑side logic:** JavaScript became a full‑fledged programming language, and frameworks (jQuery, then Angular, React, Vue) made it possible to build complex user interfaces entirely in the browser.
  - **HTTP as a universal transport:** HTTP was no longer just for hypertext; it carried JSON, images, videos, and API calls.

Today, the web is a mature application platform. Browsers are powerful execution environments with capabilities like 3D graphics (WebGL), real‑time communication (WebRTC), offline storage (Service Workers), and near‑native performance. This is why we can confidently use it to teach modern app development.

## 6. How the Web Works

Now we dive into the mechanics: what exactly happens when you type a URL into a browser?

### The Web Server
A **web server** is simply a computer program that listens for incoming HTTP requests and responds with the requested resources (or an error). Any general‑purpose computer can run web server software.

**Core function:**
- **Listen on a port:** A port is a 16‑bit number (0–65535) that identifies a specific service on a machine. HTTP conventionally uses **port 80**; HTTPS uses **443**. The server registers with the operating system to receive all TCP segments destined for that port.
- **Accept connections:** When a client connects, the OS creates a **socket** – an endpoint for communication. The server uses this socket to read the request and write the response.
- **Process the request:** The server reads the HTTP request line and headers, determines what the client wants (a specific file, a dynamic resource, etc.), and constructs a response.
- **Send the response:** It writes the HTTP status line, response headers, and then the response body back over the socket.

**How a simple static web server works:**
1. Parse the request to extract the requested path (e.g., `/index.html`).
2. Map that path to a file on the local filesystem (e.g., `/var/www/html/index.html`).
3. Check if the file exists and is readable. If not, return a **404 Not Found** status.
4. Determine the MIME type of the file (`.html` → `text/html`, `.jpg` → `image/jpeg`, etc.).
5. Construct headers: `Content-Type`, `Content-Length`, `Last-Modified`, etc.
6. Read the file contents and send them after the headers.

**Dynamic web servers** do much more: they might run a program (e.g., a Python script, a Node.js application) that generates the response on the fly, query a database, and return custom HTML.

### The HyperText Transfer Protocol (HTTP) in Detail
HTTP is a **request‑response** protocol. It is **stateless**: each request is independent; the server doesn’t inherently remember previous requests (state is maintained via cookies, sessions, or other mechanisms on top).

An HTTP message (request or response) consists of:
1. **Start‑line** (request line or status line).
2. **Headers** (zero or more key‑value pairs).
3. **Empty line** (a CRLF alone) to separate headers from body.
4. **Body** (optional data, e.g., HTML content, JSON).

**HTTP Request structure:**
```
GET /index.html HTTP/1.1
Host: www.example.com
User-Agent: Mozilla/5.0 ...
Accept: text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8
Accept-Language: en-us,en;q=0.5
Accept-Encoding: gzip, deflate
Connection: keep-alive

(empty line, no body for GET)
```
- **Method:** `GET` (retrieve resource), `POST` (submit data), `PUT`, `DELETE`, `HEAD`, `OPTIONS`, etc.
- **Path:** The local part of the URL (e.g., `/index.html`).
- **HTTP version:** `HTTP/1.1` or `HTTP/2`.
- **Headers:** `Host` is mandatory in HTTP/1.1 (allows multiple domains on one IP). `User-Agent` identifies the browser. `Accept` tells the server what MIME types the client can handle (with quality factors `q` for preference).

**HTTP Response structure:**
```
HTTP/1.1 200 OK
Date: Mon, 23 May 2022 22:38:34 GMT
Server: Apache/2.4.1 (Unix)
Content-Type: text/html; charset=UTF-8
Content-Length: 138
Last-Modified: Wed, 08 Jan 2022 23:11:55 GMT
Connection: close

<html>
<head><title>Example</title></head>
<body><p>Hello World!</p></body>
</html>
```
- **Status line:** `HTTP/1.1 200 OK`. The status code is a 3‑digit number:
  - `1xx` Informational
  - `2xx` Success (200 OK, 201 Created)
  - `3xx` Redirection (301 Moved Permanently, 302 Found)
  - `4xx` Client Error (400 Bad Request, 403 Forbidden, 404 Not Found)
  - `5xx` Server Error (500 Internal Server Error, 503 Service Unavailable)
- **Headers:** `Content-Type` specifies the MIME type (`text/html`, `image/png`, `application/json`, etc.). `Content-Length` is the size of the body in bytes, vital for the client to know when the transmission is complete.
- **Body:** The actual resource.

### Performance Factors at the Server Level
Understanding the flow of a web request reveals where performance bottlenecks can arise:

1. **Network latency:** Time for packets to travel physically. Affected by distance (speed of light in fiber ~200,000 km/s), router processing delays, and the number of hops.
2. **Bandwidth:** The maximum data rate of the connection (e.g., 100 Mbps). A response of 1 MB on a 100 Mbps link takes at least ~80 ms just for transmission (plus overhead).
3. **Concurrency:** How many simultaneous connections the server can handle. A simple server that forks a new process per connection can quickly exhaust memory (the C10K problem – handling 10,000 concurrent clients). Modern servers use event‑driven architectures (Node.js, Nginx) to handle massive concurrency with a single thread.
4. **Storage I/O:** Disk reads for static files or database queries for dynamic content. SSDs are orders of magnitude faster than HDDs, but still can be a bottleneck if not cached properly.
5. **CPU:** Generating dynamic content (template rendering, crypto, image processing) eats CPU cycles.

These will be revisited in the final section on performance.

## 7. A Simple Web Server in Action

To make the concepts concrete, the professor demonstrates two extremely simple web servers.

### The One‑Liner Bash Web Server
```bash
while true; do
    echo -e "HTTP/1.1 200 OK\r\n\r\n$(date)" | nc -l localhost 1500
done
```
**Breakdown:**
- `while true; do ... done`: An infinite loop. The server keeps running forever.
- `echo -e "..."`: Outputs a string. `-e` enables interpretation of escape sequences like `\r\n` (carriage return + line feed – the standard line ending in HTTP).
- `HTTP/1.1 200 OK\r\n\r\n`: The status line and a blank line to signal end of headers. There are no other headers; the body starts immediately after.
- `$(date)`: Command substitution. The current date and time is inserted.
- `| nc -l localhost 1500`: The pipe redirects `echo`'s output to `nc` (netcat). `nc -l` tells netcat to **listen** on the specified address and port. When a connection arrives, netcat sends the piped data and then closes the connection (the loop runs again for the next request).

**What this demonstrates:**
- A web server can be created with a single shell command.
- It completely ignores the client’s request; it always responds with the date.
- It is **dynamic**: the response changes each time (the date changes).
- It is **not robust**: no proper error handling, no MIME type, no concurrency (one request at a time).

**Testing with `curl`:**  
`curl` is a command‑line tool to transfer data using various protocols.  
`curl -v http://localhost:1500` makes a verbose HTTP request. The output shows the request sent (`> GET / HTTP/1.1`) and the response received (`< HTTP/1.1 200 OK`, date). This tool is invaluable for debugging APIs and web servers.

### Python’s Built‑in HTTP Server
```bash
python -m http.server
```
- This starts a simple but functional web server on port 8000 (default).
- It serves static files from the current working directory.
- If no file is requested (just `/`), it looks for `index.html` and serves it, or shows a directory listing.
- It correctly sets headers: `Content-Type`, `Content-Length`, `Last-Modified`.

**Experiment with `curl -v`:**
- `GET /` → returns `index.html` (or directory listing).
- `GET /somefile.sh` → returns the file with `Content-Type: application/x-sh`.
- The server parses the request, maps the path to a file, determines MIME type, and streams the file.

**Key Takeaways:**
- The protocol (HTTP) is simple and text‑based; you can read the entire conversation.
- A server’s job is to map a request to a resource and send it back with appropriate metadata.
- Real servers add layers of functionality: security, logging, caching, dynamic content generation, load balancing.

## 8. Understanding Protocols (HTTP in Detail)

A **protocol** is a set of rules governing the exchange of information between two parties. In human interaction, a handshake and greeting are a protocol. For computers, it’s a precise specification of message formats, sequencing, and actions.

### HTTP as a Text‑Based Protocol
HTTP/1.x messages are human‑readable. This was a deliberate design choice for simplicity and debugging. The trade‑off is that parsing text is slightly slower than parsing binary formats, but with modern hardware this is negligible for most applications. HTTP/2 uses a binary framing layer for efficiency, but the semantics remain the same.

### HTTP Methods (Verbs)
The method indicates the desired action on the resource.

| Method | Semantics | Typical Use |
|--------|-----------|-------------|
| **GET** | Retrieve a resource. Should be **safe** (no side effects) and **idempotent** (multiple identical requests have same effect as one). | Fetching a web page, search query (query in URL). |
| **POST** | Submit data to be processed. Often results in creation of a new resource or a side effect. **Not idempotent** (submitting a form twice may create two orders). | Form submissions, uploading a file, sending an email. |
| **PUT** | Replace a resource entirely, or create it if it doesn’t exist. **Idempotent**. | Updating a profile, putting a file at a specific URL. |
| **DELETE** | Remove a resource. **Idempotent**. | Deleting a user, removing a document. |
| **HEAD** | Like GET but only returns headers (no body). Useful for checking if a resource exists or its size. | Link validation, caching. |
| **OPTIONS** | Ask the server which methods are supported for a resource. | CORS preflight requests. |
| **PATCH** | Partially update a resource. | Modifying a single field in a JSON object. |

The lecture emphasises `GET` and `POST` as the most common, but modern **RESTful APIs** leverage all these methods to perform CRUD operations (Create=POST, Read=GET, Update=PUT/PATCH, Delete=DELETE) on resources.

### HTTP Status Codes
Status codes inform the client about the result. They are grouped by the first digit.

- **1xx Informational:** Request received, continuing process (rarely used directly by applications).
- **2xx Success:** The request was successfully received, understood, and accepted.
  - `200 OK`: Standard success.
  - `201 Created`: Resource created (POST).
  - `204 No Content`: Success but no body (e.g., after DELETE).
- **3xx Redirection:** Further action needs to be taken.
  - `301 Moved Permanently`: The resource has a new permanent URL. Clients should update bookmarks.
  - `302 Found` (temporary redirect).
  - `304 Not Modified`: The client’s cached version is still valid (sent in response to conditional GET).
- **4xx Client Error:** The request contains bad syntax or cannot be fulfilled.
  - `400 Bad Request`: Malformed request.
  - `401 Unauthorized`: Authentication required.
  - `403 Forbidden`: Server refuses to authorise (logged in but lacking permissions).
  - `404 Not Found`: The famous one.
- **5xx Server Error:** The server failed to fulfill a valid request.
  - `500 Internal Server Error`: Generic catch‑all; something crashed.
  - `503 Service Unavailable`: Server overloaded or under maintenance.

### Headers and MIME Types
Headers provide metadata. Some critical ones:

- **Request headers:**
  - `Host`: The domain name (mandatory in HTTP/1.1).
  - `User-Agent`: Client application info.
  - `Accept`: MIME types the client can understand (e.g., `text/html, application/json`).
  - `Cookie`: Sends stored cookies back to the server for session management.
- **Response headers:**
  - `Content-Type`: The MIME type of the body. This is how the client knows to render HTML, display an image, or download a file. Examples: `text/html`, `text/css`, `image/jpeg`, `application/json`, `application/pdf`.
  - `Content-Length`: Size of body in bytes.
  - `Cache-Control`: Directives for caching (e.g., `max-age=3600`).
  - `Set-Cookie`: Ask the client to store a cookie for state.

MIME (Multipurpose Internet Mail Extensions) was originally for email attachments but is now the universal content type system on the internet.

### The Python HTTP Server Example Revisited
When `curl -v http://localhost:8000` is run:
- The request sent is identical to the bash server example.
- The response now includes full headers:
  ```
  HTTP/1.0 200 OK
  Server: SimpleHTTP/0.6 Python/3.9.2
  Date: ...
  Content-Type: text/html
  Content-Length: 31
  Last-Modified: ...
  ```
- The body is the content of `index.html` (or directory listing). This is a proper static web server.

Requesting a specific file like `server.sh` yields `Content-Type: application/x-sh`, demonstrating how the server inspects the file extension to set the correct MIME type.

## 9. Performance of a Website

Building a functional app is one thing; building one that can handle real‑world traffic is another. The lecture presents back‑of‑the‑envelope calculations to highlight the physical limits that even the simplest web server faces.

### 1. Latency – The Speed of Light Limitation
- Speed of light in fiber ~2×10⁸ m/s. (200,000 km/s).
- Time per meter: 1 / (2×10⁸) = 5×10⁻⁹ s = **5 ns per meter**.
- 1,000 km → 5 ns/m × 1,000,000 m = **5 milliseconds** one‑way.
- For a 2,000 km round‑trip (Chennai to Delhi): ~20 ms.
- **Impact:** If the user’s request must complete before the next action (synchronous), the maximum number of sequential requests per second is 1 / 0.020 = **50 requests per second**.
- Realistically, browsers open multiple parallel connections and use asynchronous requests, but latency still sets a floor on perceived responsiveness. Content Delivery Networks (CDNs) mitigate this by placing servers physically closer to users.

### 2. Bandwidth – The Pipe Limitation
- Assume server has a 100 Mbps internet connection. 100 Mbps ≈ 10 MB/s (after protocol overhead).
- If each response is 1 KB of data (headers + HTML):
  - Max requests per second = 10,000,000 bytes/s ÷ 1,000 bytes = **10,000 requests/sec**.
- **Google’s homepage** (just the search box) transfers about **144 KB** of data (HTTP headers, HTML, CSS, JavaScript, tracking pixels).
  - With 144 KB per request, the same 100 Mbps pipe can handle only 10,000,000 / (144×1024) ≈ **68 requests/sec**.
- Google reportedly handles **60,000+ searches per second** globally.
  - Required aggregate bandwidth: 60,000 × 144 KB ≈ 8.6 GB/s = **69 Gbps**.
  - This is why Google has dozens of data centres worldwide, thousands of servers, and extensive caching layers. A single server could never cope.

### 3. Memory (RAM) – Concurrency Limitation
- The Python `http.server` process consumed ~6 MB of RAM when running.
- If the server spawns a new process (or thread) per connection, and we have long‑lived connections (e.g., streaming):
  - 2 million concurrent viewers (like a major live event on YouTube) × 6 MB = **12 TB of RAM**.
- Even if a single large server could hold 12 TB, it would be prohibitively expensive and a single point of failure.
- **Solution:** Scale **out**, not **up**. Distribute the load across thousands of smaller servers. With 1,000 servers, each handles 2,000 connections and needs only 12 GB of RAM – easily manageable.
- Modern asynchronous servers (Node.js, Nginx) handle many thousands of concurrent connections per process with a single thread and event loop, drastically reducing memory overhead.

### 4. Storage – The Data Scale
- Google’s web index is estimated at **100 petabytes** (100,000 TB, or 100 million GB).
- Even with 1 TB hard disks, that’s 100,000 disks. Managing this requires a distributed file system (Google File System/GFS, Hadoop HDFS) that spreads data across clusters, handles replication, and allows parallel processing (MapReduce).
- For an app developer, the lesson is: design your data model and access patterns with scalability in mind. What works for 1,000 users may collapse for 1 million.

### 5. System Design Implications
These physical constraints force developers to adopt certain strategies:

- **Caching:** Store frequently accessed data in memory (Memcached, Redis) to reduce database hits and computation.
- **CDNs:** Distribute static assets (images, CSS, JS) geographically.
- **Load balancing:** Distribute incoming requests among many servers.
- **Asynchronous processing:** Offload heavy work (e.g., sending emails) to background job queues.
- **Database sharding:** Split large databases across multiple servers by a key (e.g., user ID).
- **Efficient protocols:** Use HTTP/2 multiplexing to reduce connection overhead; compress responses (gzip).

The key takeaway is that **web apps must be designed for scale from the beginning**, even if the initial user base is small. Thinking about latency, bandwidth, memory, and storage early prevents painful rewrites later.

## 1. What is an App?

### Definition and Evolution
The term “app” is a shorthand for **application software** – a computer program designed to carry out a specific task for the user. Originally, the word “application” referred to any software that ran on a desktop or laptop computer: word processors, spreadsheets, games, etc. The shortened form “app” gained massive popularity with the rise of **smartphones** and **app stores** (especially Apple’s App Store in 2008). Today, an app is commonly understood as:

> A self-contained piece of software, typically small and focused, that can be downloaded and installed on a mobile device, but increasingly also on desktops, web browsers, and embedded devices.

The Techopedia definition cited in the lecture captures this: *“An app is computer software or a program, most commonly a small, specific one used for mobile devices.”*

**Key characteristics of an app:**
- **Specific goal:** Unlike a full operating system, an app solves a particular problem (e.g., ordering food, editing photos, sending messages).
- **Self-contained distribution:** Apps are packaged as a single unit that users can install (via an app store or direct download).
- **User-centric interface:** They are designed for direct interaction, often through touch, keyboard/mouse, or voice.

### Apps vs. Applications: The Blurring Boundary
While “app” once implied a lightweight mobile program and “application” a heavyweight desktop program, the distinction has faded:
- Microsoft Word is available as a full desktop application, a mobile app, and a web app.
- Google Docs is primarily a web app, yet feels like a full application.
- Many “apps” today, like Adobe Photoshop Express, are scaled‑down versions of their desktop counterparts, but they still perform complex tasks.

Thus, in this course, the terms *app* and *application* are used interchangeably. The focus is on understanding the principles behind building any application, regardless of its final form factor.

### Categories of Apps by Platform
The lecture categorises apps by the **platform** on which they run. Each platform imposes different constraints and offers different capabilities.

#### 1. Desktop Apps
- **Environment:** Run on a personal computer (Windows, macOS, Linux) with a keyboard, mouse, and large display.
- **Characteristics:** Often **standalone** – can function without a network connection. Examples: text editors (VS Code, Notepad), word processors, IDEs, media players.
- **Development:** Traditionally built using OS‑specific **Software Development Kits (SDKs)** and APIs (e.g., Win32 API for Windows, Cocoa for macOS). This ties the application to a particular operating system.
- **Offline capability:** They rely heavily on local storage. Even an email client like Outlook downloads messages to the local machine (using POP3 or IMAP), allowing offline reading and composition; it only needs the network to send or sync.
- **Paradigm:** The desktop metaphor – files, folders, windows – mimics a physical desk, making it intuitive for office‑type work.

#### 2. Mobile Apps
- **Devices:** Smartphones and tablets (iOS, Android). The number of phones vastly exceeds that of tablets.
- **Constraints:**
  - **Screen size:** Limited real estate forces careful UI design.
  - **Input:** Touch replaces keyboard/mouse; no hover states. New modalities: multi‑touch gestures, accelerometer, gyroscope, GPS, cameras, microphones.
  - **Resources:** Less RAM, slower processors, and restricted power budgets because they run on batteries that must last a full day.
- **Network dependency:** Most mobile apps are inherently network‑oriented (social media, shopping, maps) because the device is always connected. Pure offline apps are rarer.
- **Development:**
  - **Native:** iOS apps are typically written in Swift/Objective‑C with Apple’s SDKs; Android apps in Kotlin/Java with Android SDK. They have deep access to hardware but are platform‑specific.
  - **Cross‑platform frameworks:** Technologies like React Native, Flutter, and Xamarin allow a single codebase to target both iOS and Android, trading some performance and native look‑and‑feel for faster development.

#### 3. Web Apps
- **What they are:** Applications that run inside a web browser. The browser (Chrome, Firefox, Safari, Edge) acts as a universal runtime.
- **Why the web is special:**
  - **Cross‑platform by nature:** Browsers are available on all major operating systems, including mobile. The web abstracts away the underlying hardware and OS differences.
  - **No installation:** Users just visit a URL; updates are instantaneous on the server side.
  - **Network‑centric:** Data and business logic typically reside on remote servers. This enables access from anywhere and easy collaboration.
- **Evolution:** The early web (Web 1.0) served static pages. With **Web 2.0** (mid‑2000s), dynamic, interactive web applications became possible thanks to technologies like **JavaScript (client‑side logic)** , **AJAX (asynchronous data loading)** , and powerful server‑side frameworks. Now, web apps can rival native apps in functionality (e.g., Google Docs, Figma).
- **This course’s focus:** The web will be the primary platform because it offers a clean, platform‑agnostic environment to teach the fundamental principles of app development: storage, computation, and presentation – without getting entangled in OS‑specific details.

#### 4. Embedded Apps
- **Definition:** Software that runs on dedicated, single‑function devices with limited resources: smart TVs, microwave ovens, refrigerators, digital watches, automotive control units.
- **Characteristics:**
  - **Fixed function:** The device does one thing (e.g., a camera takes pictures), often extremely well.
  - **Resource‑constrained:** Low memory, slow processors, minimal or no user interface.
  - **“Invisible” apps:** Many embedded systems have no screen; they communicate via simple LEDs, network protocols, or machine‑to‑machine interfaces.
- **Modern shift:** Devices are increasingly “smart” (e.g., smartwatches running Wear OS, smart fridges with grocery‑ordering apps). The line between embedded and mobile/web is blurring, but the design philosophy remains: minimal, efficient, and purpose‑built.

### Apps as Icons: The User’s View
The lecture shows familiar icons – Amazon, Twitter, Word, Chrome, VS Code, etc. Each icon represents an entry point into a distinct app. Notably, **browsers** and **text editors** are both apps themselves and tools for creating other apps. For example, a browser renders web apps; a text editor is used to write the code that becomes an app. The terminal (command line) is another critical tool that developers must master to run scripts, manage servers, and automate tasks.

## 2. Components of an App

Every application, regardless of platform, can be decomposed into three fundamental concerns: **storage**, **computation**, and **presentation**. This separation helps in designing, building, and maintaining software.

### 1. Storage (Data)
- **What is stored:** Any persistent information the app needs to retain between sessions – user profiles, documents, emails, settings, images, transaction records.
- **Where storage happens:**
  - **Local storage:** On the user’s device (files on disk, local databases like SQLite, browser localStorage). Desktop apps often rely heavily on local storage.
  - **Cloud/Server storage:** Data is kept on remote servers (“in the cloud”). Web apps and most mobile apps use this so that data is accessible from any device and never lost if the user’s device fails.
- **How storage is organized:**
  - **File systems:** Simple hierarchical folders and files (e.g., text files, CSVs, configuration files). Easy to understand but lack sophisticated querying.
  - **Databases:** Structured collections of data with efficient indexing and querying capabilities.
    - **Relational databases (SQL):** Use tables with rows and columns, enforce schemas, and support powerful queries (e.g., MySQL, PostgreSQL). Good for structured data with complex relationships.
    - **NoSQL databases:** Document stores (MongoDB), key‑value stores (Redis), graph databases (Neo4j). Offer flexibility in schema and scale horizontally.
  - **Proprietary formats:** Applications like Microsoft Outlook store emails in large, indexed binary files (PST/OST) that are essentially databases but not directly readable by other software. Other clients use open, text‑based formats like mbox (all mails in one file) or maildir (one file per message) for portability.
- **Storage design considerations:**
  - **Durability:** Data must survive power loss and hardware failures.
  - **Scalability:** As the amount of data grows, can the storage system handle it without degradation?
  - **Security:** Who can read/modify the data? Encryption at rest and in transit.
  - **Access patterns:** Is the data read‑heavy, write‑heavy, or balanced? This influences the choice of database technology.

### 2. Computation (Business Logic)
- **What it does:** The “brain” of the app. It processes data, enforces rules, performs calculations, and makes decisions. This is often called **business logic** – the core functionality that gives the app its value.
- **Examples from an email client:**
  - **Indexing:** When a new email arrives, the server must update its full‑text search index so the user can later search by keyword, sender, date, etc. Indexing involves tokenizing text, removing stop words, and building inverted indexes (like a book’s index but for words).
  - **Filtering/Spam detection:** Algorithms analyze incoming messages to decide whether they are spam, applying rules (e.g., Bayesian filtering, blacklists) that must run quickly and accurately.
  - **Sorting and threading:** Emails can be sorted by date, sender, subject. Threading reconstructs conversation trees, grouping replies together.
  - **Notifications:** When a new mail arrives, the server must determine which users are affected and push a notification to their devices.
- **Where computation happens:**
  - **Server‑side:** On the web server or application server. The client sends a request, the server does the heavy lifting, and returns the result. This is typical for web apps and any task that requires access to a large central database or sensitive algorithms.
  - **Client‑side:** In the browser (JavaScript) or in a mobile/desktop app. This reduces server load and latency, but is limited by the client’s computing power and security constraints. Often used for UI logic, form validation, and real‑time interactivity.
- **Real‑world analogy:** The “controller” in the MVC pattern (discussed later) is the part that orchestrates computation – it decides what to do with user input and how to update the model.

### 3. Presentation (User Interface)
- **What the user sees and interacts with:** The visual layout, text, images, buttons, forms, and all feedback (animations, sounds, haptics).
- **Responsibilities:**
  - **Rendering:** Taking the data from the model and displaying it in a human‑readable, aesthetically pleasing manner. For an email client, this means showing a list of emails with subject, sender, date, and an indicator for unread status; displaying the full email body with proper formatting (HTML emails with images).
  - **Interactivity:** Capturing user actions (clicks, taps, swipes, voice commands) and translating them into requests that the rest of the app can understand.
- **Presentation technologies:**
  - **Web:** HTML for structure, CSS for styling, and JavaScript for dynamic behavior. The browser interprets this trio to render web pages.
  - **Desktop:** Native UI toolkits (Windows Presentation Foundation, Qt, Cocoa) or cross‑platform frameworks (Electron, which uses web technologies).
  - **Mobile:** Android’s XML‑based layouts, iOS’s SwiftUI, or cross‑platform frameworks that compile to native components.
- **Separation of concerns:** Keeping presentation logic separate from business logic (as in MVC) allows designers to change the look and feel without touching core functionality, and vice‑versa. It also enables the same backend to serve different frontends (e.g., a mobile app and a web app sharing the same server).

### The Email Client Example – Putting It All Together
To illustrate the three components, consider a modern email service like Gmail:

| Component | Gmail Implementation |
|-----------|----------------------|
| **Storage** | All emails are stored on Google’s servers in a massively distributed database (likely a custom NoSQL system). Metadata (sender, recipient, labels, read status) is also stored and indexed. Attachments are saved in a separate blob store. |
| **Computation** | When you hit “Send,” the server checks recipient validity, appends the message to the sender’s Sent folder, creates a copy in each recipient’s Inbox, updates search indexes, and runs spam checks. Searching through years of emails requires lightning‑fast queries across petabytes of data. |
| **Presentation** | The Gmail web interface shows your inbox as a scrollable list. Clicking a message opens a detailed view. The interface is built with HTML, CSS, and a huge amount of JavaScript that updates the page dynamically without full reloads. The same backend serves the Gmail mobile app (which uses native rendering but the same APIs). |

## 3. Client-Server and Peer-to-Peer Architecture

When an app involves multiple devices communicating over a network, we need an **architecture** that defines how they connect and share responsibilities.

### Client-Server Architecture
This is the dominant model for most web and mobile apps today.

**The three entities:**
1. **Server:** A powerful computer (or a cluster) that **stores data** and **performs computation**. It listens for incoming requests from clients and responds with the requested data or the result of a computation.
2. **Client:** The end‑user’s device (laptop, phone, tablet) that initiates requests to the server. The client focuses on **presentation** and **user interaction**, sending user actions to the server and displaying the responses.
3. **Network:** The communication channel (the Internet, a local network) that connects clients and servers. Ideally, it acts as a transparent pipe – simply transporting data without modification.

**How it works:**
- The server is always on, waiting for requests.
- A client opens a connection, sends a structured request (e.g., “Get me the list of my unread emails”).
- The server processes the request, possibly querying a database or running business logic.
- The server sends back a response (e.g., the list of emails as data).
- The client receives the data and renders it for the user.

**Key characteristics:**
- **Explicit differentiation:** At any given moment, you can clearly identify who is the server and who is the client. The server does not initiate communication; it only responds.
- **Centralisation:** Data and logic are concentrated in the server. This makes it easier to secure, update, and scale (within limits).
- **Examples:**
  - **Email (IMAP/SMTP):** An email server (e.g., Gmail’s server) stores your mail; your email client (Outlook, Apple Mail, web browser) fetches and displays it.
  - **Web browsing:** The browser (client) requests a page from a web server (server).
  - **WhatsApp:** Messages are sent to WhatsApp’s servers, which then route them to the recipient(s). The server also stores undelivered messages.
  - **Databases:** A database server (MySQL, PostgreSQL) runs continuously; multiple client applications connect to it, send SQL queries, and receive results.
  - **Anti‑virus updates:** The anti‑virus software acts as a client, periodically checking a server for new virus definition files and downloading them without explicit user interaction.
- **Variations:**
  - **Single server, multiple clients:** The simplest model.
  - **Multi‑tier architecture:** The server itself might be split into a **web server**, an **application server**, and a **database server**, each running on different machines but collectively acting as “the server” to the client.
  - **Load balancers:** A front‑end server that distributes incoming client requests among several back‑end servers to handle high traffic.

### Peer-to-Peer (P2P) Architecture
In a P2P network, there is no central server that owns all resources. Instead, every participant (peer) acts as both client and server, sharing its own resources (storage, bandwidth, processing) and consuming resources from others.

**How it works:**
- Peers connect to each other directly or through a lightweight introducer service.
- Data is often fragmented and distributed across many peers. When you request a file, your client fetches pieces from multiple peers simultaneously, and you in turn upload pieces you already have to others.
- The system is self‑organising and can tolerate failures: if one peer goes offline, others can still serve the data.

**Key characteristics:**
- **No central authority:** There is no single point of control or failure. This makes the system resistant to censorship and server crashes.
- **Scalability:** The network becomes stronger as more peers join, because each new peer adds resources.
- **Trust and security challenges:** How do you ensure the data you receive is authentic and not corrupted? Solutions include cryptographic hashing, digital signatures, and blockchain‑style consensus.

**Examples:**
- **BitTorrent:** The classic file‑sharing protocol. A .torrent file points to a tracker (or uses a distributed hash table) that helps peers find each other. Files are downloaded in small pieces from many peers.
- **Blockchain and Cryptocurrencies:** A distributed ledger where each node holds a copy of the entire chain. No central bank; consensus algorithms validate transactions.
- **Distributed File Systems:** **IPFS (InterPlanetary File System)** and **Tahoe‑LAFS** allow files to be stored across a network of nodes, with content addressed by its hash. Data can be encrypted so that only authorised parties can decrypt it, even though it physically resides on strangers’ machines.
- **Skype (original):** Initially used a P2P network for voice calls, with supernodes helping to route traffic, before moving to a more centralised model.

### Comparison

| Aspect | Client-Server | Peer-to-Peer |
|--------|---------------|--------------|
| **Control** | Central server controls data and logic | Distributed; all peers are theoretically equal |
| **Scalability** | Limited by server capacity; requires load balancers and more servers to scale | Scales organically; each new peer adds capacity |
| **Reliability** | Server is a single point of failure; requires redundancy | Robust against failures; no single point of failure |
| **Security** | Easier to enforce access control and data integrity | Harder; requires cryptography and reputation systems |
| **Suitable for** | Applications needing central authority, user accounts, consistent data (email, e‑commerce, banking) | File sharing, content distribution, decentralised apps (dApps) |

In this course, we will focus exclusively on the **client-server** model because it aligns with how web applications are built: a browser (client) communicates with a web server (server) using HTTP.

## 4. Software Architectures (Design Patterns & MVC)

### What is a Design Pattern?
A **design pattern** is a general, reusable solution to a commonly occurring problem within a given context in software design. It is not a finished piece of code but a **template** or **best practice** that developers can adapt to their specific situation.

**Origins:**
- The concept was popularised by the “Gang of Four” (Gamma, Helm, Johnson, Vlissides) in their 1994 book *Design Patterns: Elements of Reusable Object‑Oriented Software*.
- Patterns arise from the collective experience of seasoned developers who have observed that certain structures lead to more maintainable, flexible, and understandable code.

**Benefits:**
- **Shared vocabulary:** Developers can communicate complex ideas succinctly (“We’ll use an Observer pattern here”).
- **Proven solutions:** They embody wisdom about what works and what pitfalls to avoid.
- **Design guidance:** They help beginners think about problems in a structured way.

### Model-View-Controller (MVC)
MVC is one of the most influential architectural patterns, originally introduced in the **Smalltalk‑80** programming environment in 1979. It partitions an application into three interconnected components:

1. **Model** – Manages the **data and business logic**.
   - Contains the core information that the application deals with.
   - Enforces the rules and constraints on that data (e.g., an email must have a valid recipient address).
   - Notifies the View when data changes so the View can update itself.
   - Examples: the list of emails and their metadata, the algorithms for search indexing, the rules for spam detection.

2. **View** – Handles the **presentation**.
   - Displays the data from the Model to the user in a specific format.
   - Sends user actions (clicks, taps) to the Controller.
   - Is passive: it does not modify the Model directly; it only reflects the Model’s state.
   - In an email app: the inbox list, the mail reading pane, the compose window.

3. **Controller** – Acts as the **intermediary**.
   - Receives input from the user (via the View).
   - Interprets the input and translates it into actions on the Model (e.g., “delete this email”, “mark as read”).
   - May also instruct the View to update.
   - Often where the “business logic” of what to do with user requests lives, though in some interpretations the heavy logic is in the Model and the Controller only coordinates.

**The MVC Flow:**
1. User interacts with the View (clicks “Delete”).
2. View notifies the Controller about the event.
3. Controller processes the event, possibly validating input, then calls appropriate methods on the Model (e.g., `email.markAsDeleted()`).
4. Model updates its state and notifies the View (directly or indirectly) that data has changed.
5. View re‑renders itself using the updated Model data.

**Why use MVC?**
- **Separation of concerns:** Developers can work on the Model, View, and Controller relatively independently. A designer can tweak the View without breaking the core logic.
- **Testability:** The Model can be tested in isolation without any UI; Controllers can be tested with mock Views.
- **Multiple Views:** The same Model can be displayed in different ways (e.g., a bar chart and a table of sales data) by using different Views.

**Limitations and Variations:**
- **MVC is not one‑size‑fits‑all:** In complex modern web apps, strict MVC can become messy because the Controller often gets bloated. Alternative patterns have emerged:
  - **Model-View-Presenter (MVP):** The Presenter replaces the Controller. It has a more direct hand in updating the View, and the View is completely passive (often used in Android development).
  - **Model-View-ViewModel (MVVM):** Common in frameworks like Angular, React (with state management), and WPF. The ViewModel exposes data and commands in a way that the View can bind to directly, reducing the need for a separate Controller.
  - **Hierarchical MVC:** Allows nesting of MVC triads, where a parent View can contain child Views that each have their own mini‑MVC.
- **In web development:** MVC is often applied with:
  - **Model:** The database and the server‑side code that interacts with it.
  - **View:** The HTML templates that render data.
  - **Controller:** The server‑side code that receives HTTP requests, calls Model methods, and selects the appropriate View to return.
  - Client‑side JavaScript frameworks (Angular, React, Vue) implement their own versions of MVC/MVVM.

For this course, we adopt the **MVC mindset** because it provides a clear, logical way to organise code and reason about the components of an app, even if we later use frameworks that deviate slightly from the pure pattern.

### Summary of Course Focus
- **Platform:** Web (browser + server)
- **Network Architecture:** Client-Server
- **Software Architecture:** Model-View-Controller
- **Core Technologies:** HTTP, HTML, CSS, JavaScript, and server‑side programming.

## 5. Introduction to the Web

The **World Wide Web** (“the web”) is an information system where documents and other resources are identified by URLs, interlinked by hypertext links, and accessible over the Internet. To understand the web as an application platform, we must first understand its foundations.

### Why the Web as a Platform?
- **Cross‑platform:** Works on Windows, macOS, Linux, iOS, Android, and even many embedded devices – any device with a modern browser. The browser abstracts hardware differences (screen size, input methods, OS).
- **Built on sound principles:** It has a simple, text‑based protocol (HTTP) and a declarative presentation language (HTML). These are well‑understood and have stood the test of time.
- **Access anywhere:** Data lives on servers, so a user can pick up any device, log in, and continue where they left off.
- **Cost‑effective deployment:** Updates are done on the server; users immediately see the latest version without downloading patches.
- **Constraints:** Web apps are limited by browser capabilities (no direct access to hardware like Bluetooth/USB without specific APIs), network dependence, and potential performance bottlenecks compared to native code. Nevertheless, for a huge class of applications, the web is ideal.

### Historical Foundations: From Circuit Switching to the Internet
To appreciate the web, we must understand the underlying network – the Internet.

**Circuit‑Switched Networks (Telephone model, 1890s)**
- When a call is made, a dedicated physical path (a circuit) is established end‑to‑end and reserved for the entire duration of the call.
- Advantages: guaranteed bandwidth, consistent latency.
- Disadvantages: horribly inefficient. Even during silence, the wires are tied up and cannot be used by anyone else. Scaling requires vast amounts of copper.

**Packet‑Switched Networks**
- Key idea: Break data into small **packets**. Each packet contains a payload (the actual data) and a header (source, destination, sequence number).
- Packets from many conversations share the same physical wires. Network devices (routers) read the destination and forward each packet hop‑by‑hop.
- Advantages: highly efficient use of resources. No dedicated circuit means the network can carry many times more traffic.
- Enables diverse data types – voice, video, text – all treated uniformly as bits.

**ARPANET (Late 1960s)**
- Funded by the US Advanced Research Projects Agency (ARPA).
- First large‑scale packet‑switched network. Connected universities and research labs.
- Introduced key concepts like host‑to‑host communication and the idea of an inter‑network.

**The Internet Protocol (IP) – January 1, 1983**
- Before IP, there were many incompatible networks (IBM’s SNA, DEC’s DECnet, Xerox’s XNS, etc.). They could not talk to each other.
- The Internet Protocol was designed as an **inter‑network** protocol: a “layer” on top of any underlying network technology. It defined a universal packet format and addressing scheme (IP addresses, originally 32‑bit values written as four decimal numbers: `192.168.1.1`).
- On “Flag Day”, January 1, 1983, ARPANET officially switched to TCP/IP, and the modern Internet was born.

**Transmission Control Protocol (TCP)**
- IP provides “best‑effort” delivery – packets can be lost, duplicated, or arrive out of order.
- TCP adds reliability on top of IP: it ensures data is delivered in order and without errors. It does this by:
  - **Acknowledgments:** The receiver sends back an ACK for every packet or group of packets. If the sender doesn’t get an ACK, it retransmits.
  - **Sequencing:** Each byte is numbered so the receiver can reassemble the stream correctly.
  - **Flow control:** Prevents a fast sender from overwhelming a slow receiver.
  - **Congestion control:** Algorithms (like TCP Reno) dynamically adjust sending rate to avoid network congestion.
- TCP/IP together form the fundamental transport layer of the Internet.

**Domain Name System (DNS) – 1985**
- IP addresses (e.g., `142.250.182.46`) are hard for humans to remember and type.
- DNS translates human‑friendly names (`google.com`) into IP addresses.
- It is a hierarchical, distributed database: root servers → top‑level domain servers (.com, .org) → authoritative servers for specific domains.
- DNS also allows for abstraction: the IP address of a server can change without the user ever knowing, and multiple IP addresses can be associated with one name for load balancing.

### The Birth of the Web: Hypertext and HTML
**Hypertext** is text that is not constrained to be linear. It contains **hyperlinks** that allow the reader to jump to other documents or sections. The idea predates computers (Vannevar Bush’s Memex, 1945; Ted Nelson coined the term in 1965).

**Tim Berners‑Lee at CERN (1989‑1990)** combined three key technologies to create the World Wide Web:
1. **HTML (HyperText Markup Language):** A simple language to structure documents. It uses tags to denote headings, paragraphs, lists, and most importantly, **links** (`<a href="...">`). HTML is plain text, so any text editor can create it.
2. **HTTP (HyperText Transfer Protocol):** A protocol for transferring HTML documents (and later, any file) over the Internet. It is a simple request‑response protocol: client sends a GET request with a URL, server responds with the file.
3. **URL (Uniform Resource Locator):** A string that uniquely identifies a resource on the web, e.g., `http://www.example.com/index.html`. It specifies the protocol, server, and path.

The first web browser, **WorldWideWeb** (later renamed Nexus), was also a WYSIWYG editor. The web was originally conceived as a collaborative medium for sharing research documents.

**Web 1.0 → Web 2.0**
- **Web 1.0 (early–mid 1990s):** Mostly static HTML pages. Servers simply served files. Interactivity was minimal; pages were read‑only brochures.
- **Web 2.0 (coined ~2004):** The web became a platform for **dynamic, interactive applications**.
  - **Server‑side processing:** Technologies like PHP, ASP.NET, Java Servlets allowed servers to generate HTML on‑the‑fly based on user input, database contents, etc.
  - **AJAX (Asynchronous JavaScript and XML):** Enabled web pages to fetch data in the background and update parts of the page without a full reload. This made applications like Google Maps possible.
  - **Rich client‑side logic:** JavaScript became a full‑fledged programming language, and frameworks (jQuery, then Angular, React, Vue) made it possible to build complex user interfaces entirely in the browser.
  - **HTTP as a universal transport:** HTTP was no longer just for hypertext; it carried JSON, images, videos, and API calls.

Today, the web is a mature application platform. Browsers are powerful execution environments with capabilities like 3D graphics (WebGL), real‑time communication (WebRTC), offline storage (Service Workers), and near‑native performance. This is why we can confidently use it to teach modern app development.

## 6. How the Web Works

Now we dive into the mechanics: what exactly happens when you type a URL into a browser?

### The Web Server
A **web server** is simply a computer program that listens for incoming HTTP requests and responds with the requested resources (or an error). Any general‑purpose computer can run web server software.

**Core function:**
- **Listen on a port:** A port is a 16‑bit number (0–65535) that identifies a specific service on a machine. HTTP conventionally uses **port 80**; HTTPS uses **443**. The server registers with the operating system to receive all TCP segments destined for that port.
- **Accept connections:** When a client connects, the OS creates a **socket** – an endpoint for communication. The server uses this socket to read the request and write the response.
- **Process the request:** The server reads the HTTP request line and headers, determines what the client wants (a specific file, a dynamic resource, etc.), and constructs a response.
- **Send the response:** It writes the HTTP status line, response headers, and then the response body back over the socket.

**How a simple static web server works:**
1. Parse the request to extract the requested path (e.g., `/index.html`).
2. Map that path to a file on the local filesystem (e.g., `/var/www/html/index.html`).
3. Check if the file exists and is readable. If not, return a **404 Not Found** status.
4. Determine the MIME type of the file (`.html` → `text/html`, `.jpg` → `image/jpeg`, etc.).
5. Construct headers: `Content-Type`, `Content-Length`, `Last-Modified`, etc.
6. Read the file contents and send them after the headers.

**Dynamic web servers** do much more: they might run a program (e.g., a Python script, a Node.js application) that generates the response on the fly, query a database, and return custom HTML.

### The HyperText Transfer Protocol (HTTP) in Detail
HTTP is a **request‑response** protocol. It is **stateless**: each request is independent; the server doesn’t inherently remember previous requests (state is maintained via cookies, sessions, or other mechanisms on top).

An HTTP message (request or response) consists of:
1. **Start‑line** (request line or status line).
2. **Headers** (zero or more key‑value pairs).
3. **Empty line** (a CRLF alone) to separate headers from body.
4. **Body** (optional data, e.g., HTML content, JSON).

**HTTP Request structure:**
```
GET /index.html HTTP/1.1
Host: www.example.com
User-Agent: Mozilla/5.0 ...
Accept: text/html,application/xhtml+xml,application/xml;q=0.9,*/*;q=0.8
Accept-Language: en-us,en;q=0.5
Accept-Encoding: gzip, deflate
Connection: keep-alive

(empty line, no body for GET)
```
- **Method:** `GET` (retrieve resource), `POST` (submit data), `PUT`, `DELETE`, `HEAD`, `OPTIONS`, etc.
- **Path:** The local part of the URL (e.g., `/index.html`).
- **HTTP version:** `HTTP/1.1` or `HTTP/2`.
- **Headers:** `Host` is mandatory in HTTP/1.1 (allows multiple domains on one IP). `User-Agent` identifies the browser. `Accept` tells the server what MIME types the client can handle (with quality factors `q` for preference).

**HTTP Response structure:**
```
HTTP/1.1 200 OK
Date: Mon, 23 May 2022 22:38:34 GMT
Server: Apache/2.4.1 (Unix)
Content-Type: text/html; charset=UTF-8
Content-Length: 138
Last-Modified: Wed, 08 Jan 2022 23:11:55 GMT
Connection: close

<html>
<head><title>Example</title></head>
<body><p>Hello World!</p></body>
</html>
```
- **Status line:** `HTTP/1.1 200 OK`. The status code is a 3‑digit number:
  - `1xx` Informational
  - `2xx` Success (200 OK, 201 Created)
  - `3xx` Redirection (301 Moved Permanently, 302 Found)
  - `4xx` Client Error (400 Bad Request, 403 Forbidden, 404 Not Found)
  - `5xx` Server Error (500 Internal Server Error, 503 Service Unavailable)
- **Headers:** `Content-Type` specifies the MIME type (`text/html`, `image/png`, `application/json`, etc.). `Content-Length` is the size of the body in bytes, vital for the client to know when the transmission is complete.
- **Body:** The actual resource.

### Performance Factors at the Server Level
Understanding the flow of a web request reveals where performance bottlenecks can arise:

1. **Network latency:** Time for packets to travel physically. Affected by distance (speed of light in fiber ~200,000 km/s), router processing delays, and the number of hops.
2. **Bandwidth:** The maximum data rate of the connection (e.g., 100 Mbps). A response of 1 MB on a 100 Mbps link takes at least ~80 ms just for transmission (plus overhead).
3. **Concurrency:** How many simultaneous connections the server can handle. A simple server that forks a new process per connection can quickly exhaust memory (the C10K problem – handling 10,000 concurrent clients). Modern servers use event‑driven architectures (Node.js, Nginx) to handle massive concurrency with a single thread.
4. **Storage I/O:** Disk reads for static files or database queries for dynamic content. SSDs are orders of magnitude faster than HDDs, but still can be a bottleneck if not cached properly.
5. **CPU:** Generating dynamic content (template rendering, crypto, image processing) eats CPU cycles.

These will be revisited in the final section on performance.

## 7. A Simple Web Server in Action

To make the concepts concrete, the professor demonstrates two extremely simple web servers.

### The One‑Liner Bash Web Server
```bash
while true; do
    echo -e "HTTP/1.1 200 OK\r\n\r\n$(date)" | nc -l localhost 1500
done
```
**Breakdown:**
- `while true; do ... done`: An infinite loop. The server keeps running forever.
- `echo -e "..."`: Outputs a string. `-e` enables interpretation of escape sequences like `\r\n` (carriage return + line feed – the standard line ending in HTTP).
- `HTTP/1.1 200 OK\r\n\r\n`: The status line and a blank line to signal end of headers. There are no other headers; the body starts immediately after.
- `$(date)`: Command substitution. The current date and time is inserted.
- `| nc -l localhost 1500`: The pipe redirects `echo`'s output to `nc` (netcat). `nc -l` tells netcat to **listen** on the specified address and port. When a connection arrives, netcat sends the piped data and then closes the connection (the loop runs again for the next request).

**What this demonstrates:**
- A web server can be created with a single shell command.
- It completely ignores the client’s request; it always responds with the date.
- It is **dynamic**: the response changes each time (the date changes).
- It is **not robust**: no proper error handling, no MIME type, no concurrency (one request at a time).

**Testing with `curl`:**  
`curl` is a command‑line tool to transfer data using various protocols.  
`curl -v http://localhost:1500` makes a verbose HTTP request. The output shows the request sent (`> GET / HTTP/1.1`) and the response received (`< HTTP/1.1 200 OK`, date). This tool is invaluable for debugging APIs and web servers.

### Python’s Built‑in HTTP Server
```bash
python -m http.server
```
- This starts a simple but functional web server on port 8000 (default).
- It serves static files from the current working directory.
- If no file is requested (just `/`), it looks for `index.html` and serves it, or shows a directory listing.
- It correctly sets headers: `Content-Type`, `Content-Length`, `Last-Modified`.

**Experiment with `curl -v`:**
- `GET /` → returns `index.html` (or directory listing).
- `GET /somefile.sh` → returns the file with `Content-Type: application/x-sh`.
- The server parses the request, maps the path to a file, determines MIME type, and streams the file.

**Key Takeaways:**
- The protocol (HTTP) is simple and text‑based; you can read the entire conversation.
- A server’s job is to map a request to a resource and send it back with appropriate metadata.
- Real servers add layers of functionality: security, logging, caching, dynamic content generation, load balancing.

## 8. Understanding Protocols (HTTP in Detail)

A **protocol** is a set of rules governing the exchange of information between two parties. In human interaction, a handshake and greeting are a protocol. For computers, it’s a precise specification of message formats, sequencing, and actions.

### HTTP as a Text‑Based Protocol
HTTP/1.x messages are human‑readable. This was a deliberate design choice for simplicity and debugging. The trade‑off is that parsing text is slightly slower than parsing binary formats, but with modern hardware this is negligible for most applications. HTTP/2 uses a binary framing layer for efficiency, but the semantics remain the same.

### HTTP Methods (Verbs)
The method indicates the desired action on the resource.

| Method | Semantics | Typical Use |
|--------|-----------|-------------|
| **GET** | Retrieve a resource. Should be **safe** (no side effects) and **idempotent** (multiple identical requests have same effect as one). | Fetching a web page, search query (query in URL). |
| **POST** | Submit data to be processed. Often results in creation of a new resource or a side effect. **Not idempotent** (submitting a form twice may create two orders). | Form submissions, uploading a file, sending an email. |
| **PUT** | Replace a resource entirely, or create it if it doesn’t exist. **Idempotent**. | Updating a profile, putting a file at a specific URL. |
| **DELETE** | Remove a resource. **Idempotent**. | Deleting a user, removing a document. |
| **HEAD** | Like GET but only returns headers (no body). Useful for checking if a resource exists or its size. | Link validation, caching. |
| **OPTIONS** | Ask the server which methods are supported for a resource. | CORS preflight requests. |
| **PATCH** | Partially update a resource. | Modifying a single field in a JSON object. |

The lecture emphasises `GET` and `POST` as the most common, but modern **RESTful APIs** leverage all these methods to perform CRUD operations (Create=POST, Read=GET, Update=PUT/PATCH, Delete=DELETE) on resources.

### HTTP Status Codes
Status codes inform the client about the result. They are grouped by the first digit.

- **1xx Informational:** Request received, continuing process (rarely used directly by applications).
- **2xx Success:** The request was successfully received, understood, and accepted.
  - `200 OK`: Standard success.
  - `201 Created`: Resource created (POST).
  - `204 No Content`: Success but no body (e.g., after DELETE).
- **3xx Redirection:** Further action needs to be taken.
  - `301 Moved Permanently`: The resource has a new permanent URL. Clients should update bookmarks.
  - `302 Found` (temporary redirect).
  - `304 Not Modified`: The client’s cached version is still valid (sent in response to conditional GET).
- **4xx Client Error:** The request contains bad syntax or cannot be fulfilled.
  - `400 Bad Request`: Malformed request.
  - `401 Unauthorized`: Authentication required.
  - `403 Forbidden`: Server refuses to authorise (logged in but lacking permissions).
  - `404 Not Found`: The famous one.
- **5xx Server Error:** The server failed to fulfill a valid request.
  - `500 Internal Server Error`: Generic catch‑all; something crashed.
  - `503 Service Unavailable`: Server overloaded or under maintenance.

### Headers and MIME Types
Headers provide metadata. Some critical ones:

- **Request headers:**
  - `Host`: The domain name (mandatory in HTTP/1.1).
  - `User-Agent`: Client application info.
  - `Accept`: MIME types the client can understand (e.g., `text/html, application/json`).
  - `Cookie`: Sends stored cookies back to the server for session management.
- **Response headers:**
  - `Content-Type`: The MIME type of the body. This is how the client knows to render HTML, display an image, or download a file. Examples: `text/html`, `text/css`, `image/jpeg`, `application/json`, `application/pdf`.
  - `Content-Length`: Size of body in bytes.
  - `Cache-Control`: Directives for caching (e.g., `max-age=3600`).
  - `Set-Cookie`: Ask the client to store a cookie for state.

MIME (Multipurpose Internet Mail Extensions) was originally for email attachments but is now the universal content type system on the internet.

### The Python HTTP Server Example Revisited
When `curl -v http://localhost:8000` is run:
- The request sent is identical to the bash server example.
- The response now includes full headers:
  ```
  HTTP/1.0 200 OK
  Server: SimpleHTTP/0.6 Python/3.9.2
  Date: ...
  Content-Type: text/html
  Content-Length: 31
  Last-Modified: ...
  ```
- The body is the content of `index.html` (or directory listing). This is a proper static web server.

Requesting a specific file like `server.sh` yields `Content-Type: application/x-sh`, demonstrating how the server inspects the file extension to set the correct MIME type.

## 9. Performance of a Website

Building a functional app is one thing; building one that can handle real‑world traffic is another. The lecture presents back‑of‑the‑envelope calculations to highlight the physical limits that even the simplest web server faces.

### 1. Latency – The Speed of Light Limitation
- Speed of light in fiber ~2×10⁸ m/s. (200,000 km/s).
- Time per meter: 1 / (2×10⁸) = 5×10⁻⁹ s = **5 ns per meter**.
- 1,000 km → 5 ns/m × 1,000,000 m = **5 milliseconds** one‑way.
- For a 2,000 km round‑trip (Chennai to Delhi): ~20 ms.
- **Impact:** If the user’s request must complete before the next action (synchronous), the maximum number of sequential requests per second is 1 / 0.020 = **50 requests per second**.
- Realistically, browsers open multiple parallel connections and use asynchronous requests, but latency still sets a floor on perceived responsiveness. Content Delivery Networks (CDNs) mitigate this by placing servers physically closer to users.

### 2. Bandwidth – The Pipe Limitation
- Assume server has a 100 Mbps internet connection. 100 Mbps ≈ 10 MB/s (after protocol overhead).
- If each response is 1 KB of data (headers + HTML):
  - Max requests per second = 10,000,000 bytes/s ÷ 1,000 bytes = **10,000 requests/sec**.
- **Google’s homepage** (just the search box) transfers about **144 KB** of data (HTTP headers, HTML, CSS, JavaScript, tracking pixels).
  - With 144 KB per request, the same 100 Mbps pipe can handle only 10,000,000 / (144×1024) ≈ **68 requests/sec**.
- Google reportedly handles **60,000+ searches per second** globally.
  - Required aggregate bandwidth: 60,000 × 144 KB ≈ 8.6 GB/s = **69 Gbps**.
  - This is why Google has dozens of data centres worldwide, thousands of servers, and extensive caching layers. A single server could never cope.

### 3. Memory (RAM) – Concurrency Limitation
- The Python `http.server` process consumed ~6 MB of RAM when running.
- If the server spawns a new process (or thread) per connection, and we have long‑lived connections (e.g., streaming):
  - 2 million concurrent viewers (like a major live event on YouTube) × 6 MB = **12 TB of RAM**.
- Even if a single large server could hold 12 TB, it would be prohibitively expensive and a single point of failure.
- **Solution:** Scale **out**, not **up**. Distribute the load across thousands of smaller servers. With 1,000 servers, each handles 2,000 connections and needs only 12 GB of RAM – easily manageable.
- Modern asynchronous servers (Node.js, Nginx) handle many thousands of concurrent connections per process with a single thread and event loop, drastically reducing memory overhead.

### 4. Storage – The Data Scale
- Google’s web index is estimated at **100 petabytes** (100,000 TB, or 100 million GB).
- Even with 1 TB hard disks, that’s 100,000 disks. Managing this requires a distributed file system (Google File System/GFS, Hadoop HDFS) that spreads data across clusters, handles replication, and allows parallel processing (MapReduce).
- For an app developer, the lesson is: design your data model and access patterns with scalability in mind. What works for 1,000 users may collapse for 1 million.

### 5. System Design Implications
These physical constraints force developers to adopt certain strategies:

- **Caching:** Store frequently accessed data in memory (Memcached, Redis) to reduce database hits and computation.
- **CDNs:** Distribute static assets (images, CSS, JS) geographically.
- **Load balancing:** Distribute incoming requests among many servers.
- **Asynchronous processing:** Offload heavy work (e.g., sending emails) to background job queues.
- **Database sharding:** Split large databases across multiple servers by a key (e.g., user ID).
- **Efficient protocols:** Use HTTP/2 multiplexing to reduce connection overhead; compress responses (gzip).

The key takeaway is that **web apps must be designed for scale from the beginning**, even if the initial user base is small. Thinking about latency, bandwidth, memory, and storage early prevents painful rewrites later.
