## 1. Information Representation in Machines

### 1.1 The Digital Foundation: Bits and Binary
At the most fundamental level, a modern computer is a digital electronic machine. It stores and processes information using **binary digits (bits)**, which can take only two values: 0 or 1. Electrically, these correspond to two distinct voltage levels (e.g., 0 V for logic 0, a higher voltage like 1 V or 3.3 V for logic 1). The use of only two states makes digital circuits highly immune to noise, temperature variations, and manufacturing imperfections, because a small drift in voltage does not change the interpretation of a signal as long as it stays within a well‑defined tolerance band.

Everything a computer works with – numbers, text, images, sound, video, program instructions – must ultimately be encoded as sequences of bits. This encoding is not inherent in the bits themselves; it relies on **conventions** (standards) that assign meaning to specific bit patterns. Without such conventions, a sequence like `0100 0001` is just eight bits with no fixed interpretation.

### 1.2 Representing Numbers
Numbers are among the easiest data types to encode in binary because there is a natural mathematical mapping.

- **Unsigned integers** use the standard place‑value (positional) system. For example, the 4‑bit sequence `0110` is interpreted as:
  \[
  0 \times 2^3 + 1 \times 2^2 + 1 \times 2^1 + 0 \times 2^0 = 0 + 4 + 2 + 0 = 6_{10}
  \]
  In general, an \(n\)-bit unsigned number can represent values from \(0\) to \(2^n - 1\).

- **Negative numbers** cannot be represented using a simple minus sign because the only available symbols are 0 and 1. The most common solution in modern computers is **two’s complement** representation. In an \(n\)-bit two’s complement system, the most significant bit (MSB) has a negative weight of \(-2^{n-1}\). For example, the 4‑bit pattern `1010` in two’s complement calculates as:
  \[
  1 \times (-2^3) + 0 \times 2^2 + 1 \times 2^1 + 0 \times 2^0 = -8 + 0 + 2 + 0 = -6_{10}
  \]
  If the same pattern were interpreted as unsigned, it would be \(10_{10}\). This illustrates the crucial role of **context**: a given bit sequence may mean different things depending on whether the program treats it as signed, unsigned, a character, or something else entirely.

### 1.3 The Problem of Representing Text
Unlike numbers, letters of the alphabet have no natural mapping to numbers. To store, transmit, and display text, a community of computer users must agree on a **character encoding** – a table that assigns a unique numeric code point to each character.

The need for such encodings arose very early in the history of digital computing, driven by:
- **Machine‑to‑machine communication** (e.g., computer to printer, or over a network).
- **Human–machine interaction** (e.g., keyboards, screens).

A character encoding answers the question: “When I see the byte `0100 0001` in memory that is supposed to be text, which letter should I display?”

### 1.4 ASCII – The American Standard Code for Information Interchange
Developed in the early 1960s, ASCII was originally a **7‑bit** code, capable of encoding \(2^7 = 128\) distinct characters. This sufficed for the basic needs of English text and simple control functions. The 128 code points were divided into:

- **Control characters** (0–31 and 127): non‑printable characters like carriage return (CR), line feed (LF), tab, bell, escape, etc.
- **Printable characters** (32–126): space, digits 0‑9, uppercase A‑Z (65–90), lowercase a‑z (97–122), punctuation marks, and a few mathematical symbols.

Notably, the codes for uppercase and lowercase letters differ by exactly 32 (the value of bit 5, `0010 0000`), which simplifies case‑conversion algorithms.

The 7‑bit ASCII was soon extended to an **8‑bit** format (Extended ASCII) by utilising the additional bit to represent another 128 characters (128–255). Different regions and manufacturers defined their own extensions for European accented characters, line‑drawing symbols, and mathematical symbols. This led to a proliferation of incompatible “code pages” (e.g., ISO 8859‑1 for Western Europe, ISO 8859‑5 for Cyrillic, etc.), making international text interchange problematic.

### 1.5 Unicode and the Universal Character Set
To unify the chaotic world of character encodings, the **Unicode Consortium** was formed, eventually producing the **Universal Character Set (UCS)**. The goal was to assign a unique numeric identifier – a **code point** – to every character of every writing system in the world (living, historical, and even some artificial scripts), plus symbols, emoji, etc.

Unicode code points are usually written as `U+` followed by a hexadecimal number, e.g.:
- `U+0041` – Latin capital letter A
- `U+05D0` – Hebrew letter Alef
- `U+597D` – CJK ideograph (Chinese/Japanese/Korean)
- `U+1F600` – 😀 (grinning face emoji)

The standard defines well over 140,000 characters, and code points range from `U+0000` to `U+10FFFF` (requiring up to 21 bits). The original vision considered two fixed‑width encoding forms:

- **UCS‑2** (2 bytes per character): could represent the first 65,536 code points (the Basic Multilingual Plane, BMP). It was simple but insufficient for historic scripts and emoji.
- **UCS‑4 / UTF‑32** (4 bytes per character): directly stores each code point as a 32‑bit integer, covering the entire Unicode range trivially. It is extremely simple to process but quadruples the memory and storage requirements for texts that are mostly ASCII.

Because most real‑world text is dominated by ASCII characters (especially on the web), a fixed 4‑byte encoding is highly inefficient. This motivated the development of **variable‑width encodings** that are backwards‑compatible with ASCII.

### 1.6 UTF‑8: The Dominant Variable‑Length Encoding
**UTF‑8** (Unicode Transformation Format – 8‑bit) encodes each code point in 1 to 4 bytes, using a **prefix‑code** scheme that self‑synchronises and never confuses the start of a character with a continuation byte.

The encoding rules for UTF‑8 are:

| Code point range (hex) | Byte 1 (binary) | Byte 2 | Byte 3 | Byte 4 |
|------------------------|-----------------|--------|--------|--------|
| U+0000 – U+007F       | 0xxxxxxx        | –      | –      | –      |
| U+0080 – U+07FF       | 110xxxxx        | 10xxxxxx | – | – |
| U+0800 – U+FFFF       | 1110xxxx        | 10xxxxxx | 10xxxxxx | – |
| U+10000 – U+10FFFF    | 11110xxx        | 10xxxxxx | 10xxxxxx | 10xxxxxx |

- **1‑byte sequences**: If the first bit is `0`, the remaining 7 bits are the ASCII code point. This guarantees that every valid ASCII text is also a valid UTF‑8 text, byte‑for‑byte identical.
- **Multi‑byte sequences**: The first byte begins with a sequence of `1` bits equal to the total number of bytes, followed by a `0`. All continuation bytes start with `10`. The `x` positions are the payload bits from the code point, distributed across the bytes.
- The `10` prefix on continuation bytes ensures that if you randomly jump into the middle of a multi‑byte character, you can immediately recognise that you are not at a character boundary and skip back to the next valid start. This property is vital for robust text processing and recovery from errors.

The example from the lecture illustrates the encoding:
- `U+0041` (A) → `0x41` (1 byte)
- `U+05D0` (Alef) → binary `0000 0101 1101 0000`, fitted into the 11 payload bits of a 2‑byte sequence → `0xD7 0x90`
- `U+597D` (好) → payload needs 16 bits, thus 3 bytes: `0xE5 0xA5 0xBD`
- `U+233B4` (a rare CJK character) → payload requires 18 bits, so 4 bytes: `0xF0 0xA3 0x8E 0xB4`

UTF‑8 strikes an excellent balance: ASCII text stays compact; most European languages need only a few 2‑byte characters; scripts like Chinese, Japanese, and Korean use 3 bytes; and the highest planes (including emoji) use 4. This made UTF‑8 the dominant encoding for the web and for many file formats and programming languages.

### 1.7 UTF‑16 and UTF‑32
- **UTF‑16** uses 2 bytes for characters in the BMP (U+0000 to U+FFFF) and a **surrogate pair** (two 2‑byte code units) for characters outside that range. It is the native encoding of many Windows APIs and the Java Virtual Machine.
- **UTF‑32** uses a fixed 4 bytes per character, simplifying random access but wasting space.

Both are valid Unicode encodings, but UTF‑8’s backwards compatibility and efficiency for English‑heavy text have made it the de facto standard for web content. The common practice is to declare the encoding in HTML documents using `<meta charset="UTF-8">`.

## 2. Efficiency of Encoding

The choice of character encoding has a direct impact on storage size, memory consumption, and transmission bandwidth. The lecture explores how to think about encoding efficiency.

### 2.1 Fixed‑Width vs. Variable‑Width
Assume a document of 5,000 characters (roughly 1,000 English words). The storage required under different encodings would be:

| Encoding | Bits per character | Total bits | Notes |
|----------|-------------------|------------|-------|
| UTF‑32 (UCS‑4) | 32 | 160,000 | Simple but wasteful for ASCII text |
| ASCII (8‑bit) | 8 | 40,000 | Covers only English; 4× smaller than UTF‑32 |
| 7‑bit ASCII | 7 | 35,000 | Slight extra saving, but uncommon in practice |
| 6‑bit custom | 6 | 30,000 | Hypothetical, if we reduced character set further |

Even more aggressive savings are possible if we exploit **character frequency**. In English, the letter ‘E’ appears far more often than ‘Z’. A **Huffman coding** scheme assigns shorter bit sequences to more frequent characters, potentially reducing the total size to 10,000–20,000 bits for the same 5,000‑character text. This is essentially what general‑purpose compression algorithms (like ZIP, gzip) do, but they operate on arbitrary byte sequences, not just characters.

### 2.2 Absolute Optimal Encoding Depends on Content
The theoretical minimum number of bits needed to represent a message is given by its **information entropy**. A document that repeats the same word 1,000 times can be compressed drastically (e.g., by run‑length encoding). Thus, there is no single “optimal” encoding for all documents; it depends on the actual statistical properties of the text. This is why compression programs are data‑dependent and why fixed encodings like UTF‑8 are a pragmatic compromise: they give reasonable efficiency for a wide range of languages without requiring per‑document analysis.

### 2.3 UTF‑8 as a Practical Compromise
UTF‑8 achieves partial efficiency by using 1 byte for the most frequent characters (ASCII) and progressively more bytes for less common ones, mimicking a static Huffman tree for the world’s scripts. It is not as compact as a custom compression of a specific document, but it has the enormous advantage of being a universal standard: any UTF‑8 text can be read by any compliant system without a separate decompression step.

## 3. What is Markup?

### 3.1 The Concept of Markup
**Markup** is a system for annotating a document with information that is syntactically distinguishable from the content itself, and that describes how the content should be structured, formatted, or otherwise processed. The term originates from the traditional publishing workflow where editors would physically write (“mark up”) instructions on a manuscript for the typesetter – e.g., “make this a heading”, “indent this paragraph”, “use italics here”.

In digital form, markup languages embed these annotations as **tags** or **codes** within the text. The same raw content can thus be presented in multiple ways, or processed automatically by software that understands the markup.

### 3.2 Presentational, Procedural, and Descriptive Markup
The seminal 1987 paper by Coombs, Renear, and DeRose (“Markup systems and the future of scholarly text processing”) classified markup into three categories:

1. **Presentational markup** (often associated with WYSIWYG editors like Microsoft Word):
   - The markup directly specifies the visual appearance: “make this text 14pt bold Helvetica, centred”.
   - The codes are typically binary or proprietary, not meant for human reading. The user sees the final rendered form while editing (“What You See Is What You Get”).
   - **Drawback:** The document’s logical structure (what is a heading, what is a citation) is lost or implicit. Automated extraction of meaning (e.g., for a screen reader, or to republish in a different medium) becomes extremely difficult.

2. **Procedural markup**:
   - The document includes explicit processing instructions: “skip 2 lines, change font to Times 12pt, print the following text, then skip 1 line”.
   - This gives very fine‑grained control, but the markup is tied to a particular output device or formatter. It mixes logical structure and presentation in a way that is hard to maintain.

3. **Descriptive (semantic) markup**:
   - The markup describes the **meaning** or **logical role** of parts of the document: “this is a level‑1 heading”, “this is a paragraph”, “this is an emphasised word”.
   - The actual visual rendering is left to a separate style sheet or to the formatting engine, which can apply different designs for different media (screen, print, speech).
   - This separation of content from presentation is the foundation of HTML, XML, LaTeX, and modern web development. It makes documents more accessible, maintainable, and reusable.

### 3.3 Semantics vs. Presentation
The distinction is crucial:
- **Semantic markup** captures the *meaning* or *purpose* of a piece of content. For example, `<strong>` indicates that the enclosed text has strong importance, but does not demand that it be rendered in bold. A screen reader might increase the volume instead.
- **Presentational markup** specifies the *look*. The old `<b>` tag, for instance, simply means “bold”, irrespective of whether the text is important.

Good HTML practices favour semantic elements (`<strong>`, `<em>`, `<h1>`–`<h6>`, `<nav>`, `<article>`) over presentational ones (`<b>`, `<i>`, `<font>`, `<center>`), which are either deprecated or should be used sparingly. This allows the same document to be:
- Styled differently via CSS.
- Processed by search engines to extract structure (headings, navigation).
- Read aloud by screen readers with appropriate intonation.

## 4. Introduction to HTML

### 4.1 Historical Context
HTML (HyperText Markup Language) is a descriptive markup language designed for creating hypertext documents on the World Wide Web. It was invented by Tim Berners‑Lee at CERN around 1990. Its syntax was influenced by SGML (Standard Generalized Markup Language), a much more complex meta‑language for defining document types.

Key milestones:
- **HTML 2.0** (1995) – first formal standard.
- **HTML 3.2, 4.0** (1997) – introduced tables, style sheets, scripting, and improved forms.
- **XHTML 1.0** (2000) – a reformulation of HTML 4 as an application of XML, enforcing stricter syntax (all tags must be closed, case‑sensitive, etc.).
- **HTML5** (2014, W3C Recommendation) – the current standard, designed to support rich web applications without plugins, with new semantic elements, multimedia, and APIs. HTML5 also accommodates the “forgiving” nature of browsers by specifying detailed parsing rules even for invalid markup.

A core design philosophy of HTML from the start was **fault tolerance**: browsers should do their best to render a page even if the HTML is not perfectly well‑formed. This contrasts sharply with strict XML parsing, which would reject the entire document on a single error. The practical reason is that most HTML was (and is) written by humans, and it is far better to show something rather than an error message.

### 4.2 Anatomy of an HTML Document
A minimal HTML5 document:

```html
<!DOCTYPE html>
<html>
  <head>
    <title>My First Page</title>
  </head>
  <body>
    <h1>My First Heading</h1>
    <p>My first paragraph.</p>
  </body>
</html>
```

- `<!DOCTYPE html>`: Declares the document type as HTML5. In older HTML versions, the DOCTYPE was a long, complicated string referencing a DTD. The HTML5 version is extremely simple.
- `<html>`: The root element that wraps the entire document.
- `<head>`: Contains metadata – title, character encoding, links to style sheets and scripts, etc. It is not directly displayed.
- `<body>`: Contains the visible content of the page.

### 4.3 Tags, Elements, and Attributes
- **Tags** are the core of HTML markup. They are enclosed in angle brackets: `<tagname>`. Most elements consist of a start tag and an end tag, e.g., `<p>...</p>`. The end tag includes a forward slash before the tag name.
- **Elements** are the complete unit: start tag + content + end tag. Some elements are empty (e.g., `<img>`, `<br>`) and may be written as `<br />` in XHTML, but HTML5 allows just `<br>`.
- **Attributes** provide additional information about an element. They are written inside the start tag as `name="value"` pairs, e.g., `<a href="http://example.com">link</a>`.

Tags are case‑insensitive (`<H1>` and `<h1>` are equivalent), but the convention is to use lowercase.

### 4.4 Nesting and Document Structure
HTML elements can be nested inside one another to create a hierarchical structure. However, tags must be properly nested: the last opened tag must be the first closed. Valid:
```html
<strong><em>bold italic</em></strong>
```
Invalid (but browsers will often fix it silently):
```html
<strong><em>bold italic</strong></em>
```

This hierarchical structure is formalised as the **Document Object Model (DOM)**, a tree representation of the document. The DOM is what the browser constructs in memory after parsing the HTML. Every element, attribute, and piece of text becomes a node in the tree. The DOM can then be manipulated programmatically using JavaScript – adding, removing, or modifying nodes – which is the foundation of dynamic web applications.

### 4.5 Semantic vs. Presentational Tags in HTML
- **Semantic:** `<h1>`–`<h6>` (headings), `<p>` (paragraph), `<strong>` (strong importance), `<em>` (emphasis), `<blockquote>`, `<code>`, `<nav>`, `<article>`, `<section>`, `<aside>`, `<header>`, `<footer>`.
- **Presentational (mostly deprecated or replaced by CSS):** `<b>` (bold), `<i>` (italic), `<font>`, `<center>`, `<big>`, `<small>`. (Note: `<small>` has been given semantic meaning in HTML5 for fine print/side comments.)

Using semantic elements improves **accessibility** (screen readers can announce “navigation”, “heading level 1”) and **SEO** (search engines better understand content hierarchy). 

### 4.6 Block vs. Inline Elements
- **Block‑level elements** (e.g., `<div>`, `<p>`, `<h1>`) start on a new line and take up the full available width. They are used for major structural sections.
- **Inline elements** (e.g., `<span>`, `<a>`, `<strong>`) do not start a new line; they flow within the text.

`<div>` is a generic block container, and `<span>` a generic inline container. Both have no inherent semantic meaning, making them versatile for grouping content for styling or scripting, but they should be used judiciously; semantic alternatives are preferred when applicable.

### 4.7 HTML5 and New Capabilities
HTML5 introduced native support for multimedia (`<audio>`, `<video>`), vector graphics (`<canvas>`, `<svg>`), and new form input types (`date`, `email`, `range`). It also defined a rich set of JavaScript APIs, including:
- **Canvas API** for 2D/3D drawing.
- **Web Storage** (localStorage, sessionStorage) for client‑side persistent data.
- **Offline web applications** via Service Workers and the Application Cache (now superseded).
- **Drag‑and‑drop** API.
- **Geolocation**, **WebSockets**, **Web Workers**, etc.

These features transformed the web from a document viewer into a full‑fledged application platform, rivalling native desktop and mobile apps.

## 5. Introduction to Styling & CSS

### 5.1 The Need for Style Sheets
HTML alone defines structure and semantics, but has minimal control over visual appearance. The same heading `<h1>Welcome</h1>` could be rendered in countless ways depending on the device, browser, and user preferences. To give designers control over layout, colours, fonts, and spacing, **Cascading Style Sheets (CSS)** were introduced.

The key principle is **separation of concerns**: HTML contains the content and its meaning; CSS contains the presentation instructions. This allows:
- Consistent styling across many pages by sharing a single style sheet.
- Multiple “themes” (skins) applied to the same HTML.
- Adaptation to different screen sizes and media types (responsive design).
- Easier maintenance: change a colour site‑wide by editing one CSS rule.

### 5.2 The Cascade
CSS is called “cascading” because multiple style sheets can influence the same element, and a deterministic cascade algorithm resolves conflicts. Styles can come from:
- **Browser default styles** (user‑agent stylesheet).
- **User styles** (set in browser preferences; rarely used today).
- **Author styles** (written by the web developer), which can be inline, embedded, or external.

The cascade assigns a weight to each rule based on its origin and specificity. In general, author styles override browser defaults, and more specific selectors override less specific ones. The `!important` flag can force a rule to take precedence.

### 5.3 CSS Syntax
A CSS rule consists of a **selector** and a **declaration block**:
```css
h1 {
  color: blue;
  font-size: 2em;
  margin-left: 40px;
}
```
- `h1` is a **type selector** (selects all `<h1>` elements).
- The block contains **declarations**, each a **property** followed by a **value**, separated by a colon and terminated by a semicolon.

Selectors can target elements by tag name, class (`.classname`), ID (`#idname`), attribute, pseudo‑class (`:hover`, `:nth-child()`), and can be combined to form complex selection patterns.

### 5.4 Ways to Apply CSS
1. **Inline CSS** (using the `style` attribute directly on an element):
   ```html
   <h1 style="color: red; text-align: center;">Hello</h1>
   ```
   This is the most specific but least reusable; it tightly couples content and presentation, defeating the purpose of CSS. It is generally discouraged except for very limited, one‑off adjustments.

2. **Internal (Embedded) CSS**: a `<style>` element inside the `<head>` of the HTML document:
   ```html
   <head>
     <style>
       body { background-color: linen; }
       h1 { color: maroon; margin-left: 40px; }
     </style>
   </head>
   ```
   This is better than inline because all rules are in one place and can affect multiple elements. However, the styles are still confined to that single page.

3. **External CSS**: a separate `.css` file linked from the HTML:
   ```html
   <head>
     <link rel="stylesheet" href="style.css">
   </head>
   ```
   This is the recommended approach for any non‑trivial website. Benefits:
   - Reuse across hundreds of pages.
   - Browsers can cache the CSS file, reducing bandwidth and load time for subsequent page views.
   - Clean separation of content and presentation, enabling team members to work on HTML and CSS independently.

## 6. Types of CSS Styling & Responsive Websites

### 6.1 The Cascade in Action: Overriding Defaults
The browser’s built‑in stylesheet provides sensible defaults (e.g., `<h1>` is large and bold, `<a>` is blue and underlined). By writing author styles, we override parts of these defaults. The cascading nature means that if we only specify `color`, the `font-size` and `font-weight` from the browser default still apply. This layering is why CSS is “cascading”.

### 6.2 Responsive Web Design
With the explosion of mobile devices, websites must adapt to a vast range of screen sizes and input methods. **Responsive design** uses CSS to automatically rearrange and resize content based on viewport characteristics. Key techniques include:
- **Fluid grids**: using relative units (percentages, `em`, `rem`, `vw`, `vh`) instead of fixed pixels.
- **Flexible images**: `max-width: 100%` so images scale down within their containers.
- **Media queries**: CSS rules that apply only when certain conditions are met, e.g.:
  ```css
  @media (max-width: 600px) {
    body { font-size: 14px; }
    .sidebar { display: none; }
  }
  ```
  This makes it possible to have entirely different layouts for desktop, tablet, and phone, all while keeping the same HTML.

The lecture demonstrates how the IIT Madras EE department website adapts from a wide desktop layout to a narrower tablet layout and finally to a stacked mobile layout, hiding some blocks, converting horizontal navigation to a hamburger menu, and adjusting font sizes and spacing – all driven by CSS without changes to the HTML.

### 6.3 CSS Frameworks (Bootstrap)
Building a responsive, aesthetically pleasing design from scratch is labour‑intensive. **CSS frameworks** like Bootstrap, Foundation, Tailwind, etc., provide pre‑written CSS and often JavaScript components. Bootstrap, originally developed at Twitter, is one of the most popular. It offers:
- A responsive 12‑column grid system.
- Pre‑styled components: buttons, navbars, forms, cards, modals.
- Consistent, modern typography and colour schemes.
- Utility classes for spacing, alignment, visibility.
- A “mobile‑first” approach: base styles target small screens, and media queries add enhancements for larger ones.

Using such a framework allows developers to focus on functionality rather than design details, and ensures a baseline of accessibility and responsiveness.

### 6.4 The Role of JavaScript
HTML and CSS together handle content and presentation, but they are not programming languages (CSS can perform some conditional styling, but is not Turing‑complete in a practical sense). **JavaScript** is a full‑fledged, interpreted programming language that runs in the browser. It can:
- Manipulate the DOM dynamically (add, remove, modify elements in response to user actions).
- Communicate asynchronously with the server (AJAX) to update parts of a page without reloading.
- Validate form input on the client side before submission.
- Create complex interactive experiences (games, drawing, maps).

In the broader “modern application development” context, JavaScript is often used heavily on the client side, but this course’s focus remains on the fundamentals of server‑side logic, HTTP, databases, and the structural separation provided by HTML and CSS. JavaScript will be introduced later as the glue for interactivity.

## 7. Practical CSS Implementation (Screencasts)

These screencasts translate theory into hands‑on examples, using a local Python web server to serve an HTML file (`jnanpith.html`) that displays information about Jnanpith awardees in a table. The goal is to style the table progressively, moving from inline styles to external CSS.

### 7.1 Setup
- **Environment:** Ubuntu with Chrome/Firefox, gedit text editor, terminal.
- **Server:** `python3 -m http.server` serves files on `http://localhost:8000`.
- **Initial HTML:** A page with a title, description, a `<div>` containing an `<h1>`, a table (`<table>`) with `<thead>` and `<tbody>`, and a `<footer>`.

### 7.2 Inline CSS: The Quick but Messy Way
The `style` attribute is added directly to an HTML tag:
```html
<th style="background-color: lightgreen;">Awardees</th>
```
- CSS properties are written as `property: value;` inside double quotes.
- Colours can be named (`lightgreen`) or specified as hex RGB (`#00FF00`).
- Multiple declarations are separated by semicolons.

**Demonstrations:**
- Adding `background-color: lightgreen` to `<th>` makes the header row green.
- Adding `background-color: lightgray` to every other `<tr>` (manually) creates alternating row colours.
- Adding `border: 1px solid black` to the `<table>` tag draws a border around the entire table.
- Adding `border: 1px solid black` to individual `<td>` or `<th>` elements adds borders around each cell. However, this results in a double border where cells meet. The fix is `border-collapse: collapse` on the table, which merges adjacent cell borders into a single border.
- `text-align: center` on the `<table>` centres all cell text. A specific `<tr>` can override this with `text-align: left`, illustrating the cascade: more specific (lower‑level) rules beat higher‑level ones.

**Drawbacks of inline CSS:**
- Repetition: every cell that needs a border must have the attribute.
- Hard to maintain: to change the border colour, you must edit every occurrence.
- Mixes content and presentation, violating the separation of concerns.
- Not scalable across multiple pages.

### 7.3 Internal CSS: Centralising Styles in `<style>`
The `<style>` element is placed inside `<head>`. Instead of inline attributes, CSS rules with selectors are written.

- **Type selector** `table { ... }` applies to all `<table>` elements. The `border` and `text-align` can be moved here, instantly affecting the whole table.
- **Grouping selectors** `td, th { border: 1px solid black; }` applies the border to both table data cells and table header cells in one rule.
- **Pseudo‑class selector** `tr:nth-child(even) { background-color: lightgray; }` targets every even row, achieving alternating row colours without manual markup. `:nth-child(odd)` would target odd rows.
- The `<thead>` background colour can be set with `thead { background-color: lightgreen; }`.
- Overrides: a lower‑level rule (e.g., on a specific `<tr>` via inline or a more specific selector) can override a higher‑level one due to cascade specificity.

Internal CSS avoids repetition within a single page, but is still page‑specific.

### 7.4 External CSS: The Optimal Approach
All style rules are moved to a separate file, e.g., `style.css`:
```css
table {
  border: 1px solid black;
  border-collapse: collapse;
  text-align: center;
}
td, th {
  border: 1px solid black;
}
thead {
  background-color: lightgreen;
}
tr:nth-child(even) {
  background-color: lightgray;
}
```
The HTML file links to it:
```html
<link rel="stylesheet" href="style.css">
```

When the page loads, the browser first fetches `jnanpith.html`, then sees the `<link>`, fetches `style.css`, parses it, and applies the styles before rendering the page (this can be observed in the browser’s Developer Tools “Network” tab as a “waterfall”). The CSS file is cacheable, so subsequent page loads or navigation to other pages using the same stylesheet avoid a redundant download.

**Key benefits:**
- Complete separation: HTML contains only content; all presentation is in one place.
- Reusability: the same `style.css` can be linked from multiple HTML files, ensuring a consistent look.
- Maintainability: a single change updates the entire site.
- Performance: reduced HTML size and caching.

## 1. Information Representation in Machines

### 1.1 The Digital Foundation: Bits and Binary
At the most fundamental level, a modern computer is a digital electronic machine. It stores and processes information using **binary digits (bits)**, which can take only two values: 0 or 1. Electrically, these correspond to two distinct voltage levels (e.g., 0 V for logic 0, a higher voltage like 1 V or 3.3 V for logic 1). The use of only two states makes digital circuits highly immune to noise, temperature variations, and manufacturing imperfections, because a small drift in voltage does not change the interpretation of a signal as long as it stays within a well‑defined tolerance band.

Everything a computer works with – numbers, text, images, sound, video, program instructions – must ultimately be encoded as sequences of bits. This encoding is not inherent in the bits themselves; it relies on **conventions** (standards) that assign meaning to specific bit patterns. Without such conventions, a sequence like `0100 0001` is just eight bits with no fixed interpretation.

### 1.2 Representing Numbers
Numbers are among the easiest data types to encode in binary because there is a natural mathematical mapping.

- **Unsigned integers** use the standard place‑value (positional) system. For example, the 4‑bit sequence `0110` is interpreted as:
  \[
  0 \times 2^3 + 1 \times 2^2 + 1 \times 2^1 + 0 \times 2^0 = 0 + 4 + 2 + 0 = 6_{10}
  \]
  In general, an \(n\)-bit unsigned number can represent values from \(0\) to \(2^n - 1\).

- **Negative numbers** cannot be represented using a simple minus sign because the only available symbols are 0 and 1. The most common solution in modern computers is **two’s complement** representation. In an \(n\)-bit two’s complement system, the most significant bit (MSB) has a negative weight of \(-2^{n-1}\). For example, the 4‑bit pattern `1010` in two’s complement calculates as:
  \[
  1 \times (-2^3) + 0 \times 2^2 + 1 \times 2^1 + 0 \times 2^0 = -8 + 0 + 2 + 0 = -6_{10}
  \]
  If the same pattern were interpreted as unsigned, it would be \(10_{10}\). This illustrates the crucial role of **context**: a given bit sequence may mean different things depending on whether the program treats it as signed, unsigned, a character, or something else entirely.

### 1.3 The Problem of Representing Text
Unlike numbers, letters of the alphabet have no natural mapping to numbers. To store, transmit, and display text, a community of computer users must agree on a **character encoding** – a table that assigns a unique numeric code point to each character.

The need for such encodings arose very early in the history of digital computing, driven by:
- **Machine‑to‑machine communication** (e.g., computer to printer, or over a network).
- **Human–machine interaction** (e.g., keyboards, screens).

A character encoding answers the question: “When I see the byte `0100 0001` in memory that is supposed to be text, which letter should I display?”

### 1.4 ASCII – The American Standard Code for Information Interchange
Developed in the early 1960s, ASCII was originally a **7‑bit** code, capable of encoding \(2^7 = 128\) distinct characters. This sufficed for the basic needs of English text and simple control functions. The 128 code points were divided into:

- **Control characters** (0–31 and 127): non‑printable characters like carriage return (CR), line feed (LF), tab, bell, escape, etc.
- **Printable characters** (32–126): space, digits 0‑9, uppercase A‑Z (65–90), lowercase a‑z (97–122), punctuation marks, and a few mathematical symbols.

Notably, the codes for uppercase and lowercase letters differ by exactly 32 (the value of bit 5, `0010 0000`), which simplifies case‑conversion algorithms.

The 7‑bit ASCII was soon extended to an **8‑bit** format (Extended ASCII) by utilising the additional bit to represent another 128 characters (128–255). Different regions and manufacturers defined their own extensions for European accented characters, line‑drawing symbols, and mathematical symbols. This led to a proliferation of incompatible “code pages” (e.g., ISO 8859‑1 for Western Europe, ISO 8859‑5 for Cyrillic, etc.), making international text interchange problematic.

### 1.5 Unicode and the Universal Character Set
To unify the chaotic world of character encodings, the **Unicode Consortium** was formed, eventually producing the **Universal Character Set (UCS)**. The goal was to assign a unique numeric identifier – a **code point** – to every character of every writing system in the world (living, historical, and even some artificial scripts), plus symbols, emoji, etc.

Unicode code points are usually written as `U+` followed by a hexadecimal number, e.g.:
- `U+0041` – Latin capital letter A
- `U+05D0` – Hebrew letter Alef
- `U+597D` – CJK ideograph (Chinese/Japanese/Korean)
- `U+1F600` – 😀 (grinning face emoji)

The standard defines well over 140,000 characters, and code points range from `U+0000` to `U+10FFFF` (requiring up to 21 bits). The original vision considered two fixed‑width encoding forms:

- **UCS‑2** (2 bytes per character): could represent the first 65,536 code points (the Basic Multilingual Plane, BMP). It was simple but insufficient for historic scripts and emoji.
- **UCS‑4 / UTF‑32** (4 bytes per character): directly stores each code point as a 32‑bit integer, covering the entire Unicode range trivially. It is extremely simple to process but quadruples the memory and storage requirements for texts that are mostly ASCII.

Because most real‑world text is dominated by ASCII characters (especially on the web), a fixed 4‑byte encoding is highly inefficient. This motivated the development of **variable‑width encodings** that are backwards‑compatible with ASCII.

### 1.6 UTF‑8: The Dominant Variable‑Length Encoding
**UTF‑8** (Unicode Transformation Format – 8‑bit) encodes each code point in 1 to 4 bytes, using a **prefix‑code** scheme that self‑synchronises and never confuses the start of a character with a continuation byte.

The encoding rules for UTF‑8 are:

| Code point range (hex) | Byte 1 (binary) | Byte 2 | Byte 3 | Byte 4 |
|------------------------|-----------------|--------|--------|--------|
| U+0000 – U+007F       | 0xxxxxxx        | –      | –      | –      |
| U+0080 – U+07FF       | 110xxxxx        | 10xxxxxx | – | – |
| U+0800 – U+FFFF       | 1110xxxx        | 10xxxxxx | 10xxxxxx | – |
| U+10000 – U+10FFFF    | 11110xxx        | 10xxxxxx | 10xxxxxx | 10xxxxxx |

- **1‑byte sequences**: If the first bit is `0`, the remaining 7 bits are the ASCII code point. This guarantees that every valid ASCII text is also a valid UTF‑8 text, byte‑for‑byte identical.
- **Multi‑byte sequences**: The first byte begins with a sequence of `1` bits equal to the total number of bytes, followed by a `0`. All continuation bytes start with `10`. The `x` positions are the payload bits from the code point, distributed across the bytes.
- The `10` prefix on continuation bytes ensures that if you randomly jump into the middle of a multi‑byte character, you can immediately recognise that you are not at a character boundary and skip back to the next valid start. This property is vital for robust text processing and recovery from errors.

The example from the lecture illustrates the encoding:
- `U+0041` (A) → `0x41` (1 byte)
- `U+05D0` (Alef) → binary `0000 0101 1101 0000`, fitted into the 11 payload bits of a 2‑byte sequence → `0xD7 0x90`
- `U+597D` (好) → payload needs 16 bits, thus 3 bytes: `0xE5 0xA5 0xBD`
- `U+233B4` (a rare CJK character) → payload requires 18 bits, so 4 bytes: `0xF0 0xA3 0x8E 0xB4`

UTF‑8 strikes an excellent balance: ASCII text stays compact; most European languages need only a few 2‑byte characters; scripts like Chinese, Japanese, and Korean use 3 bytes; and the highest planes (including emoji) use 4. This made UTF‑8 the dominant encoding for the web and for many file formats and programming languages.

### 1.7 UTF‑16 and UTF‑32
- **UTF‑16** uses 2 bytes for characters in the BMP (U+0000 to U+FFFF) and a **surrogate pair** (two 2‑byte code units) for characters outside that range. It is the native encoding of many Windows APIs and the Java Virtual Machine.
- **UTF‑32** uses a fixed 4 bytes per character, simplifying random access but wasting space.

Both are valid Unicode encodings, but UTF‑8’s backwards compatibility and efficiency for English‑heavy text have made it the de facto standard for web content. The common practice is to declare the encoding in HTML documents using `<meta charset="UTF-8">`.

## 2. Efficiency of Encoding

The choice of character encoding has a direct impact on storage size, memory consumption, and transmission bandwidth. The lecture explores how to think about encoding efficiency.

### 2.1 Fixed‑Width vs. Variable‑Width
Assume a document of 5,000 characters (roughly 1,000 English words). The storage required under different encodings would be:

| Encoding | Bits per character | Total bits | Notes |
|----------|-------------------|------------|-------|
| UTF‑32 (UCS‑4) | 32 | 160,000 | Simple but wasteful for ASCII text |
| ASCII (8‑bit) | 8 | 40,000 | Covers only English; 4× smaller than UTF‑32 |
| 7‑bit ASCII | 7 | 35,000 | Slight extra saving, but uncommon in practice |
| 6‑bit custom | 6 | 30,000 | Hypothetical, if we reduced character set further |

Even more aggressive savings are possible if we exploit **character frequency**. In English, the letter ‘E’ appears far more often than ‘Z’. A **Huffman coding** scheme assigns shorter bit sequences to more frequent characters, potentially reducing the total size to 10,000–20,000 bits for the same 5,000‑character text. This is essentially what general‑purpose compression algorithms (like ZIP, gzip) do, but they operate on arbitrary byte sequences, not just characters.

### 2.2 Absolute Optimal Encoding Depends on Content
The theoretical minimum number of bits needed to represent a message is given by its **information entropy**. A document that repeats the same word 1,000 times can be compressed drastically (e.g., by run‑length encoding). Thus, there is no single “optimal” encoding for all documents; it depends on the actual statistical properties of the text. This is why compression programs are data‑dependent and why fixed encodings like UTF‑8 are a pragmatic compromise: they give reasonable efficiency for a wide range of languages without requiring per‑document analysis.

### 2.3 UTF‑8 as a Practical Compromise
UTF‑8 achieves partial efficiency by using 1 byte for the most frequent characters (ASCII) and progressively more bytes for less common ones, mimicking a static Huffman tree for the world’s scripts. It is not as compact as a custom compression of a specific document, but it has the enormous advantage of being a universal standard: any UTF‑8 text can be read by any compliant system without a separate decompression step.

## 3. What is Markup?

### 3.1 The Concept of Markup
**Markup** is a system for annotating a document with information that is syntactically distinguishable from the content itself, and that describes how the content should be structured, formatted, or otherwise processed. The term originates from the traditional publishing workflow where editors would physically write (“mark up”) instructions on a manuscript for the typesetter – e.g., “make this a heading”, “indent this paragraph”, “use italics here”.

In digital form, markup languages embed these annotations as **tags** or **codes** within the text. The same raw content can thus be presented in multiple ways, or processed automatically by software that understands the markup.

### 3.2 Presentational, Procedural, and Descriptive Markup
The seminal 1987 paper by Coombs, Renear, and DeRose (“Markup systems and the future of scholarly text processing”) classified markup into three categories:

1. **Presentational markup** (often associated with WYSIWYG editors like Microsoft Word):
   - The markup directly specifies the visual appearance: “make this text 14pt bold Helvetica, centred”.
   - The codes are typically binary or proprietary, not meant for human reading. The user sees the final rendered form while editing (“What You See Is What You Get”).
   - **Drawback:** The document’s logical structure (what is a heading, what is a citation) is lost or implicit. Automated extraction of meaning (e.g., for a screen reader, or to republish in a different medium) becomes extremely difficult.

2. **Procedural markup**:
   - The document includes explicit processing instructions: “skip 2 lines, change font to Times 12pt, print the following text, then skip 1 line”.
   - This gives very fine‑grained control, but the markup is tied to a particular output device or formatter. It mixes logical structure and presentation in a way that is hard to maintain.

3. **Descriptive (semantic) markup**:
   - The markup describes the **meaning** or **logical role** of parts of the document: “this is a level‑1 heading”, “this is a paragraph”, “this is an emphasised word”.
   - The actual visual rendering is left to a separate style sheet or to the formatting engine, which can apply different designs for different media (screen, print, speech).
   - This separation of content from presentation is the foundation of HTML, XML, LaTeX, and modern web development. It makes documents more accessible, maintainable, and reusable.

### 3.3 Semantics vs. Presentation
The distinction is crucial:
- **Semantic markup** captures the *meaning* or *purpose* of a piece of content. For example, `<strong>` indicates that the enclosed text has strong importance, but does not demand that it be rendered in bold. A screen reader might increase the volume instead.
- **Presentational markup** specifies the *look*. The old `<b>` tag, for instance, simply means “bold”, irrespective of whether the text is important.

Good HTML practices favour semantic elements (`<strong>`, `<em>`, `<h1>`–`<h6>`, `<nav>`, `<article>`) over presentational ones (`<b>`, `<i>`, `<font>`, `<center>`), which are either deprecated or should be used sparingly. This allows the same document to be:
- Styled differently via CSS.
- Processed by search engines to extract structure (headings, navigation).
- Read aloud by screen readers with appropriate intonation.

## 4. Introduction to HTML

### 4.1 Historical Context
HTML (HyperText Markup Language) is a descriptive markup language designed for creating hypertext documents on the World Wide Web. It was invented by Tim Berners‑Lee at CERN around 1990. Its syntax was influenced by SGML (Standard Generalized Markup Language), a much more complex meta‑language for defining document types.

Key milestones:
- **HTML 2.0** (1995) – first formal standard.
- **HTML 3.2, 4.0** (1997) – introduced tables, style sheets, scripting, and improved forms.
- **XHTML 1.0** (2000) – a reformulation of HTML 4 as an application of XML, enforcing stricter syntax (all tags must be closed, case‑sensitive, etc.).
- **HTML5** (2014, W3C Recommendation) – the current standard, designed to support rich web applications without plugins, with new semantic elements, multimedia, and APIs. HTML5 also accommodates the “forgiving” nature of browsers by specifying detailed parsing rules even for invalid markup.

A core design philosophy of HTML from the start was **fault tolerance**: browsers should do their best to render a page even if the HTML is not perfectly well‑formed. This contrasts sharply with strict XML parsing, which would reject the entire document on a single error. The practical reason is that most HTML was (and is) written by humans, and it is far better to show something rather than an error message.

### 4.2 Anatomy of an HTML Document
A minimal HTML5 document:

```html
<!DOCTYPE html>
<html>
  <head>
    <title>My First Page</title>
  </head>
  <body>
    <h1>My First Heading</h1>
    <p>My first paragraph.</p>
  </body>
</html>
```

- `<!DOCTYPE html>`: Declares the document type as HTML5. In older HTML versions, the DOCTYPE was a long, complicated string referencing a DTD. The HTML5 version is extremely simple.
- `<html>`: The root element that wraps the entire document.
- `<head>`: Contains metadata – title, character encoding, links to style sheets and scripts, etc. It is not directly displayed.
- `<body>`: Contains the visible content of the page.

### 4.3 Tags, Elements, and Attributes
- **Tags** are the core of HTML markup. They are enclosed in angle brackets: `<tagname>`. Most elements consist of a start tag and an end tag, e.g., `<p>...</p>`. The end tag includes a forward slash before the tag name.
- **Elements** are the complete unit: start tag + content + end tag. Some elements are empty (e.g., `<img>`, `<br>`) and may be written as `<br />` in XHTML, but HTML5 allows just `<br>`.
- **Attributes** provide additional information about an element. They are written inside the start tag as `name="value"` pairs, e.g., `<a href="http://example.com">link</a>`.

Tags are case‑insensitive (`<H1>` and `<h1>` are equivalent), but the convention is to use lowercase.

### 4.4 Nesting and Document Structure
HTML elements can be nested inside one another to create a hierarchical structure. However, tags must be properly nested: the last opened tag must be the first closed. Valid:
```html
<strong><em>bold italic</em></strong>
```
Invalid (but browsers will often fix it silently):
```html
<strong><em>bold italic</strong></em>
```

This hierarchical structure is formalised as the **Document Object Model (DOM)**, a tree representation of the document. The DOM is what the browser constructs in memory after parsing the HTML. Every element, attribute, and piece of text becomes a node in the tree. The DOM can then be manipulated programmatically using JavaScript – adding, removing, or modifying nodes – which is the foundation of dynamic web applications.

### 4.5 Semantic vs. Presentational Tags in HTML
- **Semantic:** `<h1>`–`<h6>` (headings), `<p>` (paragraph), `<strong>` (strong importance), `<em>` (emphasis), `<blockquote>`, `<code>`, `<nav>`, `<article>`, `<section>`, `<aside>`, `<header>`, `<footer>`.
- **Presentational (mostly deprecated or replaced by CSS):** `<b>` (bold), `<i>` (italic), `<font>`, `<center>`, `<big>`, `<small>`. (Note: `<small>` has been given semantic meaning in HTML5 for fine print/side comments.)

Using semantic elements improves **accessibility** (screen readers can announce “navigation”, “heading level 1”) and **SEO** (search engines better understand content hierarchy). 

### 4.6 Block vs. Inline Elements
- **Block‑level elements** (e.g., `<div>`, `<p>`, `<h1>`) start on a new line and take up the full available width. They are used for major structural sections.
- **Inline elements** (e.g., `<span>`, `<a>`, `<strong>`) do not start a new line; they flow within the text.

`<div>` is a generic block container, and `<span>` a generic inline container. Both have no inherent semantic meaning, making them versatile for grouping content for styling or scripting, but they should be used judiciously; semantic alternatives are preferred when applicable.

### 4.7 HTML5 and New Capabilities
HTML5 introduced native support for multimedia (`<audio>`, `<video>`), vector graphics (`<canvas>`, `<svg>`), and new form input types (`date`, `email`, `range`). It also defined a rich set of JavaScript APIs, including:
- **Canvas API** for 2D/3D drawing.
- **Web Storage** (localStorage, sessionStorage) for client‑side persistent data.
- **Offline web applications** via Service Workers and the Application Cache (now superseded).
- **Drag‑and‑drop** API.
- **Geolocation**, **WebSockets**, **Web Workers**, etc.

These features transformed the web from a document viewer into a full‑fledged application platform, rivalling native desktop and mobile apps.

## 5. Introduction to Styling & CSS

### 5.1 The Need for Style Sheets
HTML alone defines structure and semantics, but has minimal control over visual appearance. The same heading `<h1>Welcome</h1>` could be rendered in countless ways depending on the device, browser, and user preferences. To give designers control over layout, colours, fonts, and spacing, **Cascading Style Sheets (CSS)** were introduced.

The key principle is **separation of concerns**: HTML contains the content and its meaning; CSS contains the presentation instructions. This allows:
- Consistent styling across many pages by sharing a single style sheet.
- Multiple “themes” (skins) applied to the same HTML.
- Adaptation to different screen sizes and media types (responsive design).
- Easier maintenance: change a colour site‑wide by editing one CSS rule.

### 5.2 The Cascade
CSS is called “cascading” because multiple style sheets can influence the same element, and a deterministic cascade algorithm resolves conflicts. Styles can come from:
- **Browser default styles** (user‑agent stylesheet).
- **User styles** (set in browser preferences; rarely used today).
- **Author styles** (written by the web developer), which can be inline, embedded, or external.

The cascade assigns a weight to each rule based on its origin and specificity. In general, author styles override browser defaults, and more specific selectors override less specific ones. The `!important` flag can force a rule to take precedence.

### 5.3 CSS Syntax
A CSS rule consists of a **selector** and a **declaration block**:
```css
h1 {
  color: blue;
  font-size: 2em;
  margin-left: 40px;
}
```
- `h1` is a **type selector** (selects all `<h1>` elements).
- The block contains **declarations**, each a **property** followed by a **value**, separated by a colon and terminated by a semicolon.

Selectors can target elements by tag name, class (`.classname`), ID (`#idname`), attribute, pseudo‑class (`:hover`, `:nth-child()`), and can be combined to form complex selection patterns.

### 5.4 Ways to Apply CSS
1. **Inline CSS** (using the `style` attribute directly on an element):
   ```html
   <h1 style="color: red; text-align: center;">Hello</h1>
   ```
   This is the most specific but least reusable; it tightly couples content and presentation, defeating the purpose of CSS. It is generally discouraged except for very limited, one‑off adjustments.

2. **Internal (Embedded) CSS**: a `<style>` element inside the `<head>` of the HTML document:
   ```html
   <head>
     <style>
       body { background-color: linen; }
       h1 { color: maroon; margin-left: 40px; }
     </style>
   </head>
   ```
   This is better than inline because all rules are in one place and can affect multiple elements. However, the styles are still confined to that single page.

3. **External CSS**: a separate `.css` file linked from the HTML:
   ```html
   <head>
     <link rel="stylesheet" href="style.css">
   </head>
   ```
   This is the recommended approach for any non‑trivial website. Benefits:
   - Reuse across hundreds of pages.
   - Browsers can cache the CSS file, reducing bandwidth and load time for subsequent page views.
   - Clean separation of content and presentation, enabling team members to work on HTML and CSS independently.

## 6. Types of CSS Styling & Responsive Websites

### 6.1 The Cascade in Action: Overriding Defaults
The browser’s built‑in stylesheet provides sensible defaults (e.g., `<h1>` is large and bold, `<a>` is blue and underlined). By writing author styles, we override parts of these defaults. The cascading nature means that if we only specify `color`, the `font-size` and `font-weight` from the browser default still apply. This layering is why CSS is “cascading”.

### 6.2 Responsive Web Design
With the explosion of mobile devices, websites must adapt to a vast range of screen sizes and input methods. **Responsive design** uses CSS to automatically rearrange and resize content based on viewport characteristics. Key techniques include:
- **Fluid grids**: using relative units (percentages, `em`, `rem`, `vw`, `vh`) instead of fixed pixels.
- **Flexible images**: `max-width: 100%` so images scale down within their containers.
- **Media queries**: CSS rules that apply only when certain conditions are met, e.g.:
  ```css
  @media (max-width: 600px) {
    body { font-size: 14px; }
    .sidebar { display: none; }
  }
  ```
  This makes it possible to have entirely different layouts for desktop, tablet, and phone, all while keeping the same HTML.

The lecture demonstrates how the IIT Madras EE department website adapts from a wide desktop layout to a narrower tablet layout and finally to a stacked mobile layout, hiding some blocks, converting horizontal navigation to a hamburger menu, and adjusting font sizes and spacing – all driven by CSS without changes to the HTML.

### 6.3 CSS Frameworks (Bootstrap)
Building a responsive, aesthetically pleasing design from scratch is labour‑intensive. **CSS frameworks** like Bootstrap, Foundation, Tailwind, etc., provide pre‑written CSS and often JavaScript components. Bootstrap, originally developed at Twitter, is one of the most popular. It offers:
- A responsive 12‑column grid system.
- Pre‑styled components: buttons, navbars, forms, cards, modals.
- Consistent, modern typography and colour schemes.
- Utility classes for spacing, alignment, visibility.
- A “mobile‑first” approach: base styles target small screens, and media queries add enhancements for larger ones.

Using such a framework allows developers to focus on functionality rather than design details, and ensures a baseline of accessibility and responsiveness.

### 6.4 The Role of JavaScript
HTML and CSS together handle content and presentation, but they are not programming languages (CSS can perform some conditional styling, but is not Turing‑complete in a practical sense). **JavaScript** is a full‑fledged, interpreted programming language that runs in the browser. It can:
- Manipulate the DOM dynamically (add, remove, modify elements in response to user actions).
- Communicate asynchronously with the server (AJAX) to update parts of a page without reloading.
- Validate form input on the client side before submission.
- Create complex interactive experiences (games, drawing, maps).

In the broader “modern application development” context, JavaScript is often used heavily on the client side, but this course’s focus remains on the fundamentals of server‑side logic, HTTP, databases, and the structural separation provided by HTML and CSS. JavaScript will be introduced later as the glue for interactivity.

## 7. Practical CSS Implementation (Screencasts)

These screencasts translate theory into hands‑on examples, using a local Python web server to serve an HTML file (`jnanpith.html`) that displays information about Jnanpith awardees in a table. The goal is to style the table progressively, moving from inline styles to external CSS.

### 7.1 Setup
- **Environment:** Ubuntu with Chrome/Firefox, gedit text editor, terminal.
- **Server:** `python3 -m http.server` serves files on `http://localhost:8000`.
- **Initial HTML:** A page with a title, description, a `<div>` containing an `<h1>`, a table (`<table>`) with `<thead>` and `<tbody>`, and a `<footer>`.

### 7.2 Inline CSS: The Quick but Messy Way
The `style` attribute is added directly to an HTML tag:
```html
<th style="background-color: lightgreen;">Awardees</th>
```
- CSS properties are written as `property: value;` inside double quotes.
- Colours can be named (`lightgreen`) or specified as hex RGB (`#00FF00`).
- Multiple declarations are separated by semicolons.

**Demonstrations:**
- Adding `background-color: lightgreen` to `<th>` makes the header row green.
- Adding `background-color: lightgray` to every other `<tr>` (manually) creates alternating row colours.
- Adding `border: 1px solid black` to the `<table>` tag draws a border around the entire table.
- Adding `border: 1px solid black` to individual `<td>` or `<th>` elements adds borders around each cell. However, this results in a double border where cells meet. The fix is `border-collapse: collapse` on the table, which merges adjacent cell borders into a single border.
- `text-align: center` on the `<table>` centres all cell text. A specific `<tr>` can override this with `text-align: left`, illustrating the cascade: more specific (lower‑level) rules beat higher‑level ones.

**Drawbacks of inline CSS:**
- Repetition: every cell that needs a border must have the attribute.
- Hard to maintain: to change the border colour, you must edit every occurrence.
- Mixes content and presentation, violating the separation of concerns.
- Not scalable across multiple pages.

### 7.3 Internal CSS: Centralising Styles in `<style>`
The `<style>` element is placed inside `<head>`. Instead of inline attributes, CSS rules with selectors are written.

- **Type selector** `table { ... }` applies to all `<table>` elements. The `border` and `text-align` can be moved here, instantly affecting the whole table.
- **Grouping selectors** `td, th { border: 1px solid black; }` applies the border to both table data cells and table header cells in one rule.
- **Pseudo‑class selector** `tr:nth-child(even) { background-color: lightgray; }` targets every even row, achieving alternating row colours without manual markup. `:nth-child(odd)` would target odd rows.
- The `<thead>` background colour can be set with `thead { background-color: lightgreen; }`.
- Overrides: a lower‑level rule (e.g., on a specific `<tr>` via inline or a more specific selector) can override a higher‑level one due to cascade specificity.

Internal CSS avoids repetition within a single page, but is still page‑specific.

### 7.4 External CSS: The Optimal Approach
All style rules are moved to a separate file, e.g., `style.css`:
```css
table {
  border: 1px solid black;
  border-collapse: collapse;
  text-align: center;
}
td, th {
  border: 1px solid black;
}
thead {
  background-color: lightgreen;
}
tr:nth-child(even) {
  background-color: lightgray;
}
```
The HTML file links to it:
```html
<link rel="stylesheet" href="style.css">
```

When the page loads, the browser first fetches `jnanpith.html`, then sees the `<link>`, fetches `style.css`, parses it, and applies the styles before rendering the page (this can be observed in the browser’s Developer Tools “Network” tab as a “waterfall”). The CSS file is cacheable, so subsequent page loads or navigation to other pages using the same stylesheet avoid a redundant download.

**Key benefits:**
- Complete separation: HTML contains only content; all presentation is in one place.
- Reusability: the same `style.css` can be linked from multiple HTML files, ensuring a consistent look.
- Maintainability: a single change updates the entire site.
- Performance: reduced HTML size and caching.

## Summary of Week 2

- **Information is represented in computers as bits, using agreed‑upon encodings.** Numbers use place‑value systems (two’s complement for signed integers). Text requires character encodings. ASCII (7‑bit) was the first widely adopted standard. Unicode provides a universal code point for every character, and UTF‑8 is the dominant variable‑length encoding that is both efficient and backwards‑compatible with ASCII.
- **Markup is a way of embedding instructions in text.** Descriptive (semantic) markup focuses on meaning (headings, paragraphs, emphasis) rather than visual presentation, enabling separation of content and style.
- **HTML is the semantic markup language of the web.** It defines a tree‑like structure (the DOM) using tags and attributes. Modern HTML5 introduced new semantic elements and APIs.
- **CSS controls presentation.** It can be applied inline, internally, or externally (preferred). The cascade determines which rules apply when multiple styles conflict.
- **Responsive design uses CSS media queries** to adapt layout to different screen sizes, essential for mobile usability.
- **Frameworks like Bootstrap** provide pre‑built CSS components and grids, accelerating development.
- **Practical exercises** with a simple table demonstrated moving from inline styles to external CSS, highlighting the advantages of separation and reusability.

This foundation paves the way for building web applications where the UI is crafted with HTML+CSS and the dynamic behaviour and data handling are handled by server‑side code and, later, JavaScript.
