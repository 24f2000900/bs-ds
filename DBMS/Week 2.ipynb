## Module 6: Introduction to Relational Model – Part 1

### 6.1. The Relational Model: A Foundation for Structured Data

The relational model is the cornerstone of modern database systems. It provides a simple yet powerful way to organize data into structures that are easy to understand, query, and maintain. The fundamental idea is to represent data as a collection of **relations**, which are essentially tables.

#### 6.1.1. What is a Relation?

A relation is a two-dimensional table with the following characteristics:

- **Columns:** Each column represents an **attribute**. Attributes have a name and a data type. The order of columns is irrelevant; what matters is their names.
- **Rows:** Each row is a **tuple** (or record). A tuple holds a value for each attribute in the relation.
- **Set Semantics:** A relation is a set of tuples. This implies:
  - The order of tuples does not matter.
  - No two tuples can be identical (duplicates are not allowed).
- **Atomic Values:** Each attribute value within a tuple must be atomic (indivisible). You cannot have a list, array, or another table as a value for a single attribute in the pure relational model.

**Example: The `instructor` Relation**

The transcript shows an instructor relation with columns: `ID`, `name`, `dept_name`, and `salary`. Each row contains information about a specific instructor.

---

### 6.2. Attributes and Their Characteristics

An attribute is a named column of a relation that describes a particular property of the entity being modeled.

#### 6.2.1. Atomicity of Attributes

An attribute must hold a single, indivisible value. This means you cannot store multiple pieces of information in one attribute in the relational model (e.g., you cannot have an attribute `phone_numbers` that contains a list). If you need to store multiple values, you must either create a separate relation or normalize the data.

**Example of a Bad Design:** An attribute `name` that contains both "First Last" in a single field is not atomic. Instead, you should have `first_name` and `last_name` as separate attributes.

#### 6.2.2. Attribute Types and Domains

Every attribute is associated with a **domain** – the set of permissible values it can take. A domain is defined by a data type and, optionally, constraints.

**Common Data Types:**

- **Alphanumeric String:** A sequence of characters (letters and digits). Example: `roll_number` as "14CS92P01".
- **Alphabetic String:** A sequence of letters only. Example: `first_name` as "Partha".
- **Date:** A standard date representation, often with format like `mm-dd-yy` or `yyyy-mmm-dd`. Example: `27-Mar-1997`.
- **String:** A fixed or variable-length sequence of characters. Example: `passport_number` as "A1234567".
- **Number:** Integer or decimal. Example: `aadhaar_number` as 12-digit integer.
- **Nullable:** The ability to allow a special `NULL` value indicating "unknown" or "does not exist". Example: `passport_number` may be `NULL` if a student does not have a passport.

**Domain Constraints:**

- `roll_number` must be an alphanumeric string; you cannot put a value like "CSE235" if the domain says it must start with "14" and follow a specific pattern.
- `date_of_birth` must be a valid date; you cannot put "27th March, 1997" (a string) because the domain is `DATE`.

**Key Point:** The domain ensures data integrity at the schema level. The DBMS enforces that any value entered for an attribute is a member of its declared domain.

---

### 6.3. Relational Schema and Instance

#### 6.3.1. Relational Schema

The **relational schema** describes the structure of a relation. It specifies the relation's name and the set of its attributes, along with their domains.

- **Mathematical Representation:** A relation with attributes `A1, A2, ..., An` is denoted as:
  ```
  R(A1, A2, ..., An)
  ```
  This is the schema. It defines the "blueprint" for the table.

- **Example:** `instructor(ID: char(5), name: varchar(20), dept_name: varchar(20), salary: numeric(8,2))` is a schema.

- **Formal Definition:** If each attribute `Ai` has a domain `Di`, then the schema is `R ⊆ D1 × D2 × ... × Dn`. This means the relation `R` is a subset of the Cartesian product of all domains, representing all possible tuple combinations.

#### 6.3.2. Relational Instance

The **relational instance** is the actual data stored in the relation at a particular moment in time. It is a set of tuples that conform to the schema.

- **Schema vs. Instance:**
  - Schema: The structure; changes rarely.
  - Instance: The content; changes frequently with inserts, updates, deletes.

- **Important Properties of Instances:**
  1. **Order of tuples is irrelevant.** The relation is a set, so rearranging rows does not change the relation.
  2. **All tuples must be distinct.** No two rows can have identical values for all attributes. This is guaranteed by the primary key, which we will discuss shortly.

**Example:**
An instance of the `instructor` relation might contain:
```
(10101, Srinivasan, Comp. Sci., 65000)
(12121, Wu, Finance, 90000)
(15151, Mozart, Music, 40000)
```

### 6.4. Keys: Uniquely Identifying Tuples

Keys are fundamental to the relational model because they ensure that every tuple can be uniquely identified. This is necessary to retrieve, update, or delete specific records.

#### 6.4.1. Superkey

A **superkey** is a set of one or more attributes that, taken collectively, allow us to uniquely identify a tuple in a relation.

- **Definition:** A set of attributes `K ⊆ R` is a superkey of `R` if, for any two distinct tuples `t1` and `t2` in `R`, `t1[K] ≠ t2[K]`. That is, no two tuples can have the same combination of values for the attributes in `K`.
- **Example:** In `instructor`, `{ID}` is a superkey. Also `{ID, name}` is a superkey, and `{ID, name, dept_name}` is a superkey. Any superset of a superkey is also a superkey.

#### 6.4.2. Candidate Key

A **candidate key** is a **minimal** superkey. It is a superkey for which no proper subset is also a superkey.

- **Minimality Requirement:** If `K` is a candidate key, then removing any attribute from `K` would cause it to lose its uniqueness property.
- **Example:** `{ID}` is a candidate key because it is a superkey and has no proper subset (other than empty set) that is also a superkey. `{ID, name}` is not a candidate key because `{ID}` alone is a superkey.
- **Multiple Candidate Keys:** A relation can have more than one candidate key. For example, in a student relation, both `roll_number` and `aadhaar_number` might be candidate keys.

#### 6.4.3. Primary Key

The **primary key** is one candidate key chosen by the database designer to be the main means of uniquely identifying tuples.

- **Choice Criteria:**
  - **Uniqueness:** It must be unique for all tuples.
  - **Stability:** Should not change over time.
  - **Simplicity:** Prefer a single attribute over a composite one if possible.
  - **Non-null:** By default, primary key attributes cannot be `NULL`.
- **Design Considerations from Transcript:**
  - `roll_number` chosen over `aadhaar_number` because it encodes useful information (year, department, category, type, serial number). This avoids the need for separate attributes for these fields.
  - `first_name, last_name` pair is a candidate key but rejected because duplicate names are common.
  - `passport_number` is not a candidate key because it can be `NULL` for some students.

#### 6.4.4. Secondary/Alternate Key

Any candidate key that is not chosen as the primary key is called a **secondary** or **alternate** key. They still enforce uniqueness but are not the primary identification method.

#### 6.4.5. Surrogate/Synthetic Key

A **surrogate key** is an artificially generated key, often a sequential number or a unique identifier (like `order_number`), that has no business meaning. It is used when no natural key exists or when natural keys are unstable or too complex.

- **Example:** An e-commerce order. Two orders might have identical items, quantities, prices, and buyer, but they are distinct orders. The `order_number` is generated by the system to distinguish them.
- **Difference from Natural Key:** A natural key (like `roll_number` or `aadhaar_number`) is derived from real-world data. A surrogate key is created solely for identification within the database.

#### 6.4.6. Composite Key and Simple Key

- **Simple Key:** A key consisting of a single attribute. Example: `roll_number`.
- **Composite Key:** A key consisting of two or more attributes. Example: `first_name, last_name`.

#### 6.4.7. Foreign Key

A **foreign key** is an attribute (or set of attributes) in one relation that refers to the primary key of another relation. It establishes a link between the two relations.

- **Purpose:** Ensures referential integrity – you cannot reference a non-existent entity.
- **Example:** In the `student` relation, `dept_name` is a foreign key referencing the primary key `dept_name` of the `department` relation. This ensures that every student belongs to a valid department.
- **Important:** A foreign key need not be a primary key in its own relation. It is the primary key in the referenced relation.

### 6.5. Relational Query Languages: An Overview

To retrieve and manipulate data in a relational database, we need a query language. There are two broad categories:

#### 6.5.1. Procedural Languages

- **Definition:** Specify *how* to obtain the result, step by step.
- **Example:** Languages like C, Java, Python. You write algorithms to loop through data, check conditions, and compute results.
- **Example: Finding Square Root Procedurally**
  ```
  Guess a value x close to sqrt(n).
  while (true) {
    x = (x + n/x) / 2;
    if (abs(x*x - n) < epsilon) break;
  }
  return x;
  ```
  This tells the computer exactly what operations to perform.

#### 6.5.2. Declarative Languages

- **Definition:** Specify *what* you want, not how to get it.
- **Example:** SQL is declarative. You state the desired result conditions, and the DBMS figures out the most efficient way to execute it.
- **Example: Finding Square Root Declaratively**
  ```
  Find m such that m * m = n.
  ```
  You don't specify the algorithm; the system (if it supported this) would solve it.

#### 6.5.3. Relational Algebra, Tuple Calculus, Domain Calculus

These are formal, theoretical query languages for the relational model. They are not Turing-complete, meaning there are some computational problems they cannot express. They provide the foundation for SQL.

- **Relational Algebra:** Procedural. Uses operators (select, project, join) to build expressions.
- **Tuple Relational Calculus:** Declarative. Uses variables that range over tuples.
- **Domain Relational Calculus:** Declarative. Uses variables that range over domains (attribute values).

SQL is based on a combination of these formal languages, primarily influenced by relational algebra and tuple calculus.

## Module 7: Introduction to Relational Model – Part 2: Relational Algebra

### 7.1. Relational Algebra: The Foundation of Query Processing

Relational algebra is a formal procedural query language. It consists of a set of operations that take one or two relations as input and produce a new relation as output. Since the output is a relation, the operations can be composed (nested) to build complex queries.

**Two Crucial Set Properties to Remember:**

1. **Order is irrelevant.** Relations are sets, so tuple order does not matter.
2. **Duplicates are not allowed.** Every tuple in a relation must be distinct.

These properties are enforced by the operations where applicable.

---

### 7.2. Fundamental Operators

#### 7.2.1. Select (σ)

The **select** operation filters rows from a relation based on a given predicate (condition).

- **Notation:** `σ_p(r)` where `r` is a relation, `p` is a selection predicate (Boolean condition).
- **Definition:** Returns all tuples in `r` for which the predicate `p` evaluates to true.
- **Predicate:** Can involve comparisons (`=`, `≠`, `<`, `>`, `≤`, `≥`) and logical connectives (`AND`, `OR`, `NOT`).

**Example:**
Consider relation `r(A, B, C, D)` with tuples.
`σ_{A=B AND D>5}(r)`
This selects tuples where the value of attribute `A` equals the value of `B` **AND** the value of `D` is greater than 5. It removes the other rows.

**Key Points:**
- Select is a horizontal filter (takes a subset of rows).
- The result relation has the same schema (attributes) as the input relation.
- The predicate is applied to each tuple individually.

#### 7.2.2. Project (π)

The **project** operation selects specific columns from a relation, removing others.

- **Notation:** `π_{A1, A2, ..., An}(r)`
- **Definition:** Returns a relation with only the attributes `A1, A2, ..., An`. The other attributes are discarded.
- **Duplicate Elimination:** Since the result is a set, duplicate rows (that may arise after removing columns) are eliminated. For example, if you project only the `A` column and multiple tuples had the same `A` value, the result contains each distinct value only once.

**Example:**
`π_{A,C}(r)` returns a relation with columns A and C. If two tuples `(alpha, 1)` and `(alpha, 2)` both have A=alpha, after projecting on A and C they might become duplicate `(alpha, x)` if x is same, so duplicates removed.

**Key Points:**
- Project is a vertical filter (takes a subset of columns).
- It enforces set semantics (no duplicates).

#### 7.2.3. Union (∪)

The **union** operation combines two relations with identical schemas.

- **Condition:** Both input relations must have the same number of attributes, and corresponding attributes must have compatible domains.
- **Definition:** Returns a relation containing all tuples that are in either relation `r` or relation `s`.
- **Duplicate Elimination:** Duplicates are removed because the result is a set.

**Example:**
`r` has tuples `(a1, b1), (a1, b2)`, `s` has `(a1, b1), (a2, b2)`. `r ∪ s = (a1, b1), (a1, b2), (a2, b2)`.

#### 7.2.4. Set Difference (−)

The **set difference** operation returns tuples that are in one relation but not in another.

- **Notation:** `r - s`
- **Condition:** Same schema compatibility as union.
- **Definition:** Returns all tuples that are in `r` but not in `s`.

**Example:** If `r` has `(a1, b1), (a1, b2), (a2, b1)`, `s` has `(a1, b1)`, then `r - s = (a1, b2), (a2, b1)`.

#### 7.2.5. Intersection (∩)

The **intersection** operation returns tuples that are common to both relations.

- **Notation:** `r ∩ s`
- **Definition:** Returns all tuples that are in both `r` and `s`.
- **Redundancy:** Intersection is not a fundamental operator because it can be expressed using set difference:
  `r ∩ s = r - (r - s)`

#### 7.2.6. Cartesian Product (×)

The **Cartesian product** (also called cross-product) combines every tuple of one relation with every tuple of another.

- **Notation:** `r × s`
- **Definition:** If `r` has `m` tuples and `s` has `n` tuples, the result has `m × n` tuples. The schema of the result is the concatenation of the schemas of `r` and `s`.
- **Attribute Name Conflicts:** If both relations have an attribute with the same name (e.g., both have `B`), the conflicting names are qualified by the relation name: `r.B` and `s.B` to make them unique in the result.
- **Use Case:** Often used as a starting point for joins.

**Example:** If `r` has 2 tuples and `s` has 4, result has 8 tuples. Each tuple from `r` is paired with each tuple from `s`.

#### 7.2.7. Rename (ρ)

The **rename** operation is used to change the name of a relation or its attributes. This is especially useful for self-joins or to give more meaningful names.

- **Notation:** `ρ_{S(A1, A2, ..., An)}(r)` renames relation `r` to `S` and its attributes to `A1, ..., An`.
- **Importance:**
  - Avoids name conflicts.
  - Allows self-join (joining a relation with itself) by creating two copies with different names.
  - Makes queries more readable.

**Example:** To find all pairs of instructors where one earns more than the other, you need to compare the `instructor` table with itself. You rename one copy as `T` and another as `S`, then write:
`σ_{T.salary > S.salary} (ρ_T(instructor) × ρ_S(instructor))`

---

### 7.3. Composition of Operations

Relational algebra operations can be nested to form complex expressions. Since each operation returns a relation, you can apply another operator to its result.

**Example:** `π_{A} (σ_{D>5}(r))` first selects rows where D>5, then projects onto A.

---

### 7.4. The Natural Join (⋈)

The **natural join** is a specialized, powerful operation that combines two relations based on equality of common attributes.

- **Definition:** `r ⋈ s`
- **How it works:**
  1. Form the Cartesian product of `r` and `s`.
  2. Select tuples where the values of attributes that are common to both relations (i.e., have the same names) are equal.
  3. Project out one copy of each common attribute (i.e., remove duplicate columns).

- **Schema of Result:** `R ∪ S` (union of attributes), with common attributes appearing only once.
- **Example from Transcript:**
  - `r(A, B, C, D)` and `s(B, D, E)`.
  - `R ∩ S = {B, D}` (common attributes).
  - The join only keeps tuple pairs where `r.B = s.B` and `r.D = s.D`.
  - In the result, only one copy of `B` and one copy of `D` are kept.

- **Why it's powerful:** It automatically combines related tuples from different tables based on matching attribute names. It is the most common operation for combining tables in queries, often expressed in SQL as `JOIN ... ON ...` or implicitly via `WHERE`.

**Note:** The natural join requires that common attribute names have the same meaning across tables. If they don't, you must use renaming or theta join (a Cartesian product with a selection condition).

---

### 7.5. Aggregate Operators

Beyond the fundamental relational operators, SQL and relational algebra extensions include **aggregate operators** that compute a single value from a set of values.

- **Common Aggregates:**
  - `AVG(column)`: Average value.
  - `MIN(column)`: Minimum value.
  - `MAX(column)`: Maximum value.
  - `SUM(column)`: Sum of values.
  - `COUNT(column)`: Number of non-null values.

- **Difference from Relational Operators:**
  - Input: A relation (or a subset).
  - Output: A single scalar value (not a relation).
  - They operate on a column, not on tuples.

**Example:**
Given relation `r(A, B)` with values `(1,2), (3,4), (5,6)`:
- `SUM(B)` returns `12`.
- `MAX(A)` returns `5`.

Aggregates are essential for reporting and analysis. They are typically used with `GROUP BY` (discussed later) to compute per-group summaries.

## Module 8: Introduction to SQL – Part 1: History, DDL, and Basic Query Structure

### 8.1. SQL: The Universal Relational Language

SQL (Structured Query Language), originally called SEQUEL (Structured English Query Language), is the standard language for interacting with relational database management systems.

#### 8.1.1. History and Standardization

- **Origins:** Developed at IBM's San Jose Research Laboratory in the 1970s as part of System R.
- **Renaming:** Initially SEQUEL, renamed to SQL (Structured Query Language).
- **Standardization:**
  - **SQL-86:** First ANSI standard.
  - **SQL-89:** Added integrity constraints.
  - **SQL-92 (ISO 9075):** Major international standard, widely implemented.
  - **SQL:1999:** Added regular expression matching, recursive queries, triggers, procedural language support, and embedded SQL in Java (SQLJ).
  - **SQL:2003:** Added XML support.
  - **SQL:2006:** Enhanced XML import/export, manipulation.
  - **SQL:2019:** Latest version with more enhancements.

- **Compliance:** Not all systems fully comply with the latest standard. Most support SQL-92 core features; many support later features partially. Therefore, syntax may vary slightly between MySQL, PostgreSQL, Oracle, SQL Server, etc.

#### 8.1.2. Alternatives and Derivatives

- **Alternatives:** There are practically no serious alternatives to SQL for relational databases. Some early languages like QUEL (based on tuple calculus) existed but faded.
- **Front-ends:** Many languages provide libraries that embed SQL (e.g., JDBC for Java, psycopg2 for Python, ActiveRecord for Ruby).
- **Derivatives:** SPARQL (SPARQL Protocol and RDF Query Language) is a query language for RDF (Resource Description Framework) data, often used in graph and semantic web databases. It is influenced by SQL but works on triples.

---

### 8.2. SQL Data Definition Language (DDL)

DDL is used to define and manage the structure of database objects (tables, schemas, indexes, etc.).

#### 8.2.1. Data Types in SQL

SQL provides a variety of data types. Common ones include:

- **CHAR(n):** Fixed-length character string. If a shorter string is stored, it is padded with spaces to length `n`.
- **VARCHAR(n):** Variable-length character string up to `n` characters.
- **INT/INTEGER:** Integer.
- **SMALLINT:** Small integer.
- **NUMERIC(p, d):** Fixed-point number with `p` total digits and `d` digits after decimal.
- **FLOAT(n):** Floating-point number.
- **DATE:** Date value.
- **TIME:** Time value.
- **TIMESTAMP:** Date and time.

#### 8.2.2. Creating a Table: `CREATE TABLE`

The `CREATE TABLE` statement defines a new relation, its attributes, types, and integrity constraints.

**General Syntax:**
```sql
CREATE TABLE table_name (
    attribute1 datatype [constraints],
    attribute2 datatype [constraints],
    ...
    [table-level constraints]
);
```

**Example: Creating the `instructor` table**
```sql
CREATE TABLE instructor (
    ID          CHAR(5),
    name        VARCHAR(20) NOT NULL,
    dept_name   VARCHAR(20),
    salary      NUMERIC(8,2),
    PRIMARY KEY (ID),
    FOREIGN KEY (dept_name) REFERENCES department(dept_name)
);
```

**Explanation:**
- `ID CHAR(5)`: Attribute ID of fixed length 5 characters.
- `name VARCHAR(20) NOT NULL`: name can vary up to 20 chars; cannot be NULL.
- `dept_name VARCHAR(20)`: Department name, can be NULL if unknown? Here it's not specified NOT NULL, but as a foreign key it could be NULL only if allowed.
- `salary NUMERIC(8,2)`: Salary with up to 8 digits total, 2 after decimal.
- `PRIMARY KEY (ID)`: Sets ID as primary key (implies NOT NULL and unique).
- `FOREIGN KEY (dept_name) REFERENCES department(dept_name)`: Ensures referential integrity.

**Other table examples:**

- **Student Table:**
```sql
CREATE TABLE student (
    ID            VARCHAR(10) PRIMARY KEY,
    name          VARCHAR(50) NOT NULL,
    dept_name     VARCHAR(20),
    tot_cred      NUMERIC(3,0),
    FOREIGN KEY (dept_name) REFERENCES department(dept_name)
);
```

- **Courses Table:**
```sql
CREATE TABLE course (
    course_id   VARCHAR(7) PRIMARY KEY,
    title       VARCHAR(50),
    dept_name   VARCHAR(20),
    credits     NUMERIC(2,0),
    FOREIGN KEY (dept_name) REFERENCES department(dept_name)
);
```

- **Takes Table (for many-to-many relationship between student and course sections):**
```sql
CREATE TABLE takes (
    ID        VARCHAR(10),
    course_id VARCHAR(7),
    sec_id    VARCHAR(8),
    semester  VARCHAR(6),
    year      NUMERIC(4,0),
    grade     VARCHAR(2),
    PRIMARY KEY (ID, course_id, sec_id, semester, year),
    FOREIGN KEY (ID) REFERENCES student(ID),
    FOREIGN KEY (course_id) REFERENCES course(course_id)
    -- section foreign key omitted for simplicity
);
```

#### 8.2.3. Altering a Table: `ALTER TABLE`

Used to modify an existing table's schema.

- **Add a new attribute:**
  ```sql
  ALTER TABLE student ADD COLUMN email VARCHAR(100);
  ```
  This adds a new column. Existing tuples will have NULL for this new column.

- **Drop an attribute:**
  ```sql
  ALTER TABLE student DROP COLUMN email;
  ```
  This removes the column and its data from all tuples.

#### 8.2.4. Dropping a Table: `DROP TABLE`

Removes the entire table definition and its data.

```sql
DROP TABLE student;
```

#### 8.2.5. Deleting All Rows: `DELETE FROM`

Removes all data from a table but keeps the table structure.

```sql
DELETE FROM student;
```

This is often used to clear a table for testing.

### 8.3. SQL Data Manipulation Language (DML): The Basic Query Structure

The most common DML operation is the query. The basic query structure in SQL is the **SELECT-FROM-WHERE** block.

```sql
SELECT [DISTINCT] attribute_list
FROM relation_list
WHERE predicate;
```

This maps to relational algebra as: **Projection (SELECT), Cartesian product (FROM), Selection (WHERE)**.

#### 8.3.1. The SELECT Clause

- Specifies which columns to include in the result.
- It corresponds to the **projection** operation.
- **Example:** `SELECT name FROM instructor;` returns a single-column table of all instructor names.
- **Case Insensitivity:** SQL keywords and identifiers are case-insensitive (though string literals are case-sensitive).
- **Duplicates:** SQL does **not** automatically eliminate duplicates (unlike relational algebra). It uses multiset semantics. To remove duplicates, use `DISTINCT`:
  ```sql
  SELECT DISTINCT dept_name FROM instructor;
  ```
- **Wildcard `*`:** `SELECT * FROM instructor;` returns all columns.
- **Arithmetic Expressions:** You can compute values:
  ```sql
  SELECT ID, name, salary/12 AS monthly_salary FROM instructor;
  ```
  This divides salary by 12 and renames the result column to `monthly_salary`.
- **Literal Values:** `SELECT '423' AS foo;` returns a single-row, single-column table with value '423'.

#### 8.3.2. The FROM Clause

- Specifies the relations involved in the query.
- If multiple relations are listed, the implicit operation is the **Cartesian product** of those relations.
- **Example:** `SELECT * FROM instructor, teaches;` produces the Cartesian product of `instructor` and `teaches`.
- **Attribute Ambiguity:** If both tables have a column named `ID`, you must qualify it as `instructor.ID` and `teaches.ID`.

#### 8.3.3. The WHERE Clause

- Specifies a Boolean predicate to filter rows.
- Corresponds to the **selection** operation.
- **Example:** `SELECT name FROM instructor WHERE dept_name = 'Comp. Sci.';` returns names of CS instructors.
- **Complex Predicates:** Can use `AND`, `OR`, `NOT`, and comparisons.
  ```sql
  SELECT name FROM instructor
  WHERE dept_name = 'Comp. Sci.' AND salary > 80000;
  ```

- **Joining via WHERE:** The combination of multiple tables in FROM and an equality condition in WHERE effectively performs a **join** (not necessarily natural join, but a theta join).
  ```sql
  SELECT instructor.name, teaches.course_id
  FROM instructor, teaches
  WHERE instructor.ID = teaches.ID;
  ```
  This returns instructor names and the courses they teach.

## Module 9: Introduction to SQL – Part 2: Additional Operations

### 9.1. The Natural Join in SQL

The transcript reinforces the concept of joining tables. The manual way is Cartesian product + WHERE condition. SQL provides explicit join syntax for clarity.

**Implicit (Theta) Join:**
```sql
SELECT name, course_id
FROM instructor, teaches
WHERE instructor.ID = teaches.ID;
```

**Explicit `JOIN ... ON`:**
```sql
SELECT name, course_id
FROM instructor JOIN teaches ON instructor.ID = teaches.ID;
```

**Natural Join (`NATURAL JOIN`):**
```sql
SELECT name, course_id
FROM instructor NATURAL JOIN teaches;
```
- This automatically joins on all common column names. In this case, `ID` is common. It eliminates one of the duplicate `ID` columns in the result.
- **Caution:** Natural join can be dangerous if there are unexpected common column names. It's better to use explicit joins for clarity.

**Example: Finding courses taught by instructors in Art department:**
```sql
SELECT name, course_id
FROM instructor JOIN teaches ON instructor.ID = teaches.ID
WHERE instructor.dept_name = 'Art';
```

### 9.2. Renaming and Aliasing (`AS`)

The `AS` keyword is used to rename a table or column within a query. This is crucial for self-joins and readability.

**Syntax:**
- `table_name AS alias_name`
- `column_name AS alias_name`
- The `AS` keyword is optional in many SQL dialects.

**Example: Self-Join to find instructors with salary greater than some CS instructor:**
```sql
SELECT T.name
FROM instructor T, instructor S
WHERE T.salary > S.salary AND S.dept_name = 'Comp. Sci.';
```
Here `T` and `S` are aliases for two copies of `instructor`. The condition matches every instructor `T` with a CS instructor `S` whose salary is lower; it selects those `T` names. (Note: this may produce duplicates unless `DISTINCT` is used.)

**Example from transcript:**
To find supervisor's supervisor, you need to join `emp_super` with itself. Aliases `e1`, `e2` would be used.

### 9.3. String Matching with `LIKE`

String comparisons often require pattern matching. The `LIKE` operator supports patterns using two special characters:

- **`%` (Percent sign):** Matches any sequence of characters, including zero characters.
- **`_` (Underscore):** Matches exactly one character.

**Examples:**
- `name LIKE '%dar%'`: Matches any name containing "dar" anywhere (e.g., Darwin, Majumdar, Sardar).
- `name LIKE 'dar%'`: Matches names starting with "dar".
- `name LIKE '%dar'`: Matches names ending with "dar".
- `name LIKE '_dar%'`: Matches names where "dar" starts at the second character.
- `name LIKE '____'`: Matches names exactly four characters long.

**Case Sensitivity:** String patterns are case-sensitive. `LIKE 'Darwin'` will not match `'darwin'`.

**Other String Functions:**
- Concatenation: `||` (standard) or `+` (SQL Server) or `CONCAT()`
- `UPPER()`, `LOWER()`, `LENGTH()`, `SUBSTRING()`, etc., vary by DBMS.

### 9.4. Ordering Results with `ORDER BY`

The `ORDER BY` clause sorts the output rows.

- **Syntax:** `ORDER BY attribute1 [ASC|DESC], attribute2 [ASC|DESC], ...`
- **Default:** Ascending order.
- **Example:**
  ```sql
  SELECT DISTINCT name
  FROM instructor
  ORDER BY name DESC;
  ```
  Sorts names in descending lexicographic order.
- **Multiple Attributes:** Sorts by first attribute, then second, etc.
- **Note:** `ORDER BY` is applied after the result is computed. It does not affect the set/multiset property, only presentation.

### 9.5. Limiting Results: `TOP`, `LIMIT`, `FETCH`

To retrieve only a portion of the result set (e.g., for pagination), SQL provides mechanisms that vary by DBMS:

- **SQL Server / MS Access:** `SELECT TOP 10 name FROM instructor;`
- **MySQL:** `SELECT name FROM instructor LIMIT 10;`
- **Oracle / PostgreSQL (standard):** `SELECT name FROM instructor FETCH FIRST 10 ROWS ONLY;`

This is useful for showing the first N rows after ordering.

### 9.6. Convenience Features in WHERE Clause

#### 9.6.1. `BETWEEN`

Simplifies range conditions.

- **Syntax:** `attribute BETWEEN lower AND upper` (inclusive).
- **Equivalent to:** `attribute >= lower AND attribute <= upper`.
- **Example:**
  ```sql
  SELECT name FROM instructor WHERE salary BETWEEN 90000 AND 100000;
  ```

#### 9.6.2. Tuple Comparison

Allows comparing multiple attributes at once.

- **Syntax:** `(a1, a2) = (b1, b2)`
- **Example:**
  ```sql
  SELECT name
  FROM instructor, teaches
  WHERE (instructor.ID, instructor.dept_name) = (teaches.ID, 'Biology');
  ```
  This is equivalent to `instructor.ID = teaches.ID AND instructor.dept_name = 'Biology'`. It is more compact but less common.

#### 9.6.3. `IN` and `NOT IN`

Used to check if an attribute value is in a list or subquery.

- **Syntax:** `attribute IN (value1, value2, ...)`
- **Example:**
  ```sql
  SELECT name FROM instructor WHERE dept_name IN ('Comp. Sci.', 'Biology');
  ```
  Equivalent to `dept_name = 'Comp. Sci.' OR dept_name = 'Biology'`.
- `NOT IN` is the negation.

### 9.7. Duplicate Handling: Multiset Semantics

SQL, unlike pure relational algebra, operates on **multisets** (bags) where duplicates are allowed. This is a practical choice for performance and convenience.

- **Why?** Eliminating duplicates automatically after every operation would be costly and sometimes you need duplicates (e.g., counting rows with duplicates).
- **Effect on Operations:**
  - **Select:** Duplicates in the input are preserved in the output.
  - **Project:** Duplicates are not eliminated unless `DISTINCT` is used.
  - **Union:** If you use `UNION ALL`, duplicates are preserved; `UNION` (without ALL) eliminates duplicates.
  - **Intersection, Except:** Similar to UNION, `ALL` variant keeps multiplicities.
- **Multiset algebra rules:** For duplicate counts:
  - If tuple appears `m` times in `r` and `n` times in `s`:
    - `UNION ALL` gives `m + n`.
    - `INTERSECT ALL` gives `min(m, n)`.
    - `EXCEPT ALL` gives `max(0, m - n)`.

**Example from transcript:**
Relation `r1` with duplicates, projection gives two `a`'s. Then cross product with relation `r2` having two `3`'s yields 4 duplicates of `a3`. This is multiset behavior.

**To get set semantics, use `DISTINCT` or set operators without `ALL`.**

## Module 10: Introduction to SQL – Part 3: Set Operations, Null Values, and Aggregation

### 10.1. Set Operations in SQL

SQL supports union, intersection, and difference operations on query results. These operate on relations (treating them as sets).

#### 10.1.1. `UNION`, `INTERSECT`, `EXCEPT`

- **`UNION`:** Returns all rows from either query, no duplicates.
- **`INTERSECT`:** Returns rows common to both queries.
- **`EXCEPT`:** Returns rows in first query but not in second (set difference).

**Conditions:** Both queries must have the same number of columns, and corresponding columns must have compatible data types.

**Example: Find courses that ran in Fall 2009 or Spring 2010**
```sql
(SELECT course_id FROM section WHERE semester='Fall' AND year=2009)
UNION
(SELECT course_id FROM section WHERE semester='Spring' AND year=2010);
```

**Find courses that ran in both Fall 2009 and Spring 2010:**
```sql
(SELECT course_id FROM section WHERE semester='Fall' AND year=2009)
INTERSECT
(SELECT course_id FROM section WHERE semester='Spring' AND year=2010);
```

**Find courses that ran in Fall 2009 but not in Spring 2010:**
```sql
(SELECT course_id FROM section WHERE semester='Fall' AND year=2009)
EXCEPT
(SELECT course_id FROM section WHERE semester='Spring' AND year=2010);
```

#### 10.1.2. `ALL` Variant for Duplicates

- `UNION ALL`: Keeps all duplicates.
- `INTERSECT ALL`: Keeps duplicates based on minimum count.
- `EXCEPT ALL`: Keeps duplicates based on maximum(0, m-n).

**Usage:** When you need to preserve multiplicities.

#### 10.1.3. Example: Finding the Highest Salary Using Set Difference

Without aggregate functions, one can find the highest salary as follows:

```sql
(SELECT DISTINCT salary FROM instructor)
EXCEPT
(SELECT DISTINCT T.salary
 FROM instructor T, instructor S
 WHERE T.salary < S.salary);
```
This works by excluding all salaries that are less than some other salary, leaving the maximum.

### 10.2. Null Values in SQL

**Null** represents missing or unknown information. It is a special marker, not a value.

#### 10.2.1. Reasons for Null

- **Value does not exist:** Example: passport number for a student without passport.
- **Value unknown:** Example: total credits for a newly admitted student before any courses are taken.
- **No value assigned:** When adding a new column to an existing table, all existing rows get `NULL` for that column.

#### 10.2.2. Properties of Null

- **Arithmetic:** Any arithmetic expression involving `NULL` yields `NULL`. (e.g., `5 + NULL = NULL`).
- **Comparison:** `NULL = NULL` is not true; it evaluates to `UNKNOWN`. You cannot compare with `NULL` using `=`. Instead, use `IS NULL` or `IS NOT NULL`.
- **Logical operations (Three-Valued Logic):**
  - Boolean logic extended with `UNKNOWN`.
  - Truth tables:
    - `OR`:
      - `TRUE OR UNKNOWN` = `TRUE`
      - `FALSE OR UNKNOWN` = `UNKNOWN`
      - `UNKNOWN OR UNKNOWN` = `UNKNOWN`
    - `AND`:
      - `TRUE AND UNKNOWN` = `UNKNOWN`
      - `FALSE AND UNKNOWN` = `FALSE`
      - `UNKNOWN AND UNKNOWN` = `UNKNOWN`
    - `NOT UNKNOWN` = `UNKNOWN`
- **In WHERE clause:** A predicate that evaluates to `UNKNOWN` is treated as `FALSE`, so the row is not selected.

**Example:**
```sql
SELECT name FROM instructor WHERE salary IS NULL;
```
Returns instructors whose salary is not known.

#### 10.2.3. Null and Aggregates

- Aggregates generally ignore `NULL` values for the attribute being aggregated.
- `COUNT(*)` counts rows including those with nulls; `COUNT(column)` counts non-null values.
- If all values are `NULL`:
  - `COUNT` returns 0.
  - Other aggregates (`SUM`, `AVG`, `MIN`, `MAX`) return `NULL`.

### 10.3. Aggregate Functions in SQL

Aggregate functions compute a single value from a set of rows.

#### 10.3.1. Common Aggregate Functions

- **`AVG(column)`**: Average of non-null values.
- **`MIN(column)`**: Minimum value.
- **`MAX(column)`**: Maximum value.
- **`SUM(column)`**: Sum of non-null values.
- **`COUNT(column)`**: Number of non-null values.
- **`COUNT(*)`**: Number of rows.

#### 10.3.2. Using Aggregates in SELECT

**Example: Average salary of CS instructors**
```sql
SELECT AVG(salary) AS avg_salary
FROM instructor
WHERE dept_name = 'Comp. Sci.';
```

**Example: Total number of instructors**
```sql
SELECT COUNT(*) AS total_instructors FROM instructor;
```

#### 10.3.3. `DISTINCT` in Aggregates

To count distinct values:
```sql
SELECT COUNT(DISTINCT dept_name) FROM instructor;
```
This counts the number of different departments.

### 10.4. Grouping: `GROUP BY`

Often you need aggregate values per group (e.g., average salary per department). The `GROUP BY` clause partitions the rows into groups based on the values of specified attributes, then applies aggregates to each group.

**Example: Average salary per department**
```sql
SELECT dept_name, AVG(salary) AS avg_salary
FROM instructor
GROUP BY dept_name;
```
This groups rows by `dept_name`, then computes the average salary within each group.

**Rules for `GROUP BY`:**
- Attributes in the `SELECT` clause must be either in the `GROUP BY` clause or be arguments of aggregate functions.
- Example of invalid query:
  ```sql
  SELECT ID, AVG(salary) FROM instructor GROUP BY dept_name;
  ```
  `ID` is not in `GROUP BY` nor an aggregate, so it is invalid.

### 10.5. Filtering Groups: `HAVING`

The `HAVING` clause is similar to `WHERE` but applied **after** grouping. It filters groups based on aggregate properties.

**Example: Departments with average salary > 70000**
```sql
SELECT dept_name, AVG(salary) AS avg_salary
FROM instructor
GROUP BY dept_name
HAVING AVG(salary) > 70000;
```

**Difference between `WHERE` and `HAVING`:**
- `WHERE` filters individual rows before grouping.
- `HAVING` filters groups after grouping.

**Example combining both:**
```sql
SELECT dept_name, AVG(salary)
FROM instructor
WHERE salary > 50000
GROUP BY dept_name
HAVING AVG(salary) > 70000;
```
Here, first rows with salary <= 50000 are excluded, then groups formed, then groups with average <= 70000 are excluded.
