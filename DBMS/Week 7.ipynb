### 31.2. Unity Behind Diversity

Despite the enormous diversity in domain, functionality, user base, response time, and scalability, all these applications share a **common architecture**:

1. **Data-intensive**: All use RDBMS (Oracle, DB2, MySQL, PostgreSQL) to manage data.
2. **Three-layer structure**: Every application can be functionally decomposed into three layers (or tiers).

### 31.3. The Three-Tier Architecture

Every database-backed application, regardless of domain, has three conceptual layers:

#### 31.3.1. Frontend Layer (Presentation Tier)

- **Purpose**: Directly interacts with the user through a browser or mobile app.
- **Examples**: Viewing emails in Gmail, browsing items in Amazon, filling a form.
- **Key functions**:
  - Displaying information to the user
  - Capturing user input
  - Executing user workflows (e.g., choose item → add to cart → checkout → payment → track order)
- **Technologies**: HTML, CSS, JavaScript, browser rendering.

#### 31.3.2. Middle Layer (Application/Business Logic Layer)

- **Purpose**: Implements the actual functionality of the application; links frontend and backend.
- **Key functions**:
  - **Authentication**: Verifying user identity (password, OTP, captcha)
  - **Search and browse logic**: Deciding what to show and in what order
  - **Pricing logic**: Computing costs
  - **Cart management**: Adding/removing items
  - **Payment gateway integration**: Interfacing with banks, UPI, credit cards
  - **Order management**: Confirmation, cancellation, tracking
  - **Workflow execution**: Maintaining the sequence of steps for tasks
- **Technologies**: Java, Python, C++, PHP, Ruby, server-side frameworks.

#### 31.3.3. Backend Layer (Data Access Layer)

- **Purpose**: Manages persistent data.
- **Key functions**:
  - Storing user data, inventory data, order data, vendor data
  - Efficient access to data
  - Security and consistency
- **Technologies**: Relational databases (MySQL, PostgreSQL, Oracle), SQL.

### 31.4. Detailed View of the Three Layers

#### 31.4.1. Presentation Layer: MVC Architecture

The presentation layer typically follows the **Model-View-Controller (MVC)** pattern:

- **Model**: Represents the business logic that must be presented in the frontend (e.g., authentication model, data validation model).
- **View**: How the data is displayed (checkboxes, edit boxes, radio buttons, static text).
- **Controller**: Manages the interaction between View and Model. It receives messages from the user, executes actions, and returns a view.

**Workflow:**
1. User submits a form.
2. Controller receives the submission.
3. Controller sends the data to the business logic layer.
4. Business logic layer processes and returns a response.
5. Controller instructs the View to render the response.
6. View generates the presentation and sends it back to the user.

#### 31.4.2. Business Logic Layer

- Provides a **high-level view of data** using an **object data model** corresponding to relational tables.
- Hides details of the storage schema.
- Selectively exposes views of data.
- Contains **complex business rules** that cannot be implemented in the database.
  - Example: "A student can enroll in a class only if she has completed prerequisites and paid her tuition fees."
- Supports **workflows**: The sequence of steps to accomplish a task, with branching and exception handling.

#### 31.4.3. Data Access Layer

- Interfaces between the business logic layer and the underlying database.
- Provides **Object-Relational Mapping (ORM)**: Maps object models (classes in Java/Python) to relational tables.
- Translates SQL results into programming language objects and vice versa.
- Handles the mismatch between the object-oriented paradigm (business layer) and relational paradigm (database).

### 31.5. End-to-End Request Flow

A typical request flows through all three tiers:

1. **User** sends an HTTP request from the web browser.
2. **Presentation layer (MVC)** interprets the request.
3. **Business logic layer** processes the request (authenticates, applies business rules).
4. **Data access layer** translates the high-level data needs into SQL queries.
5. **Database server** executes the SQL query.
6. **Result** (a table) is returned to the data access layer.
7. **Data access layer** maps the relational result back to objects.
8. **Business logic layer** post-processes (e.g., adds computed fields).
9. **Presentation layer** renders the result as HTML.
10. **User** sees the HTML response in the browser.

### 31.6. Evolution of Application Architectures

#### 31.6.1. One-Tier Architecture (1960s-1970s)

- All three layers (presentation, logic, data) reside on the same computer (often a mainframe).
- Users access via dumb terminals.
- Simple but not scalable.

#### 31.6.2. Two-Tier Architecture (Client-Server, 1980s)

- Client tier (frontend + some application logic) and database tier (data layer).
- Direct communication between client and server.
- Application logic is distributed between client and server.
- Limitations: No clear separation of concerns, difficult to manage business logic.

#### 31.6.3. Three-Tier Architecture (1990s-Present)

- Clear separation: Presentation tier, Business Logic Tier (BLL), Data Access Layer (DAL).
- Allows independent scaling and development of each tier.
- Most common architecture for modern web applications.

#### 31.6.4. N-Tier Architecture

- Functionally similar to three-tier, but implemented with multiple servers at each tier.
- Multiple application servers, multiple database servers.
- Web services as separate entities.
- Physically distributed across different machines.
- Conceptually still follows the same three-layer model.

## Module 32: Application Design and Development – Part 2: Web Applications

### 32.1. Web Fundamentals

#### 32.1.1. The World Wide Web

A **distributed information system** based on **hypertext**. Documents are:
- Formatted using **HTML (Hypertext Markup Language)** – a W3C standard.
- Contain text, formatting specifications, links to other documents, and forms for data collection.

#### 32.1.2. URL (Uniform Resource Locator)

A URL identifies and locates a resource on the web.

**Structure:**
```
protocol://hostname/path?query_parameters
```

**Example:**
```
http://www.example.com/search?q=silberschatz
```

- **Protocol**: `http` (Hypertext Transfer Protocol) – how to interact with the document.
- **Hostname**: `www.example.com` – which machine to find the document.
- **Path**: `/search` – location of the document within the machine.
- **Query parameters**: `q=silberschatz` – additional parameters for the request.

**Alternative protocols:**
- `file:` – access local file system.
- `https:` – secure HTTP.

**URI vs URN vs URL:**
- **URI (Uniform Resource Identifier)**: The general name/identifier.
- **URN (Uniform Resource Name)**: Like a person's name.
- **URL (Uniform Resource Locator)**: Like a person's address (how to find them).

We commonly use URL.

#### 32.1.4. HTTP Protocol

**HTTP (Hypertext Transfer Protocol)** is:
- **Connectionless**: Each request is independent; no persistent connection between client and server.
- **Request-Response model**: Client sends request, server sends response.

**Problem**: For stateful applications (like net banking), the server needs to know who the client is across multiple requests.

**Solution: Cookies and Sessions**

- **Cookie**: A small piece of information (encrypted text) stored on the client machine, issued by the server after authentication.
- Every subsequent request from the client includes the cookie.
- The server validates the cookie to identify the user.
- **Session**: The duration for which the cookie remains valid.
- **Timeout**: Cookies expire after inactivity to protect user security.

### 32.4. Scripting

**Scripting** refers to programming languages that are:
- Text-based (interpreted, not compiled).
- Typically small and lightweight.
- Executed within another program (interpreted line by line).

**Why scripting for web applications?**
- Many tasks are small (validate date, check input).
- Need to generate HTML dynamically.
- Faster development cycle (no compile step).

#### 32.4.1. Client-Side Scripting

Runs in the web browser. Used for:
- Data validation (e.g., checking if a date is valid before submitting to server)
- Auto-fill
- Dynamic page content without server round-trips
- Interactive UI elements

**Common technologies:**
- **JavaScript**: The most common client-side scripting language.
- Flash, SVG scripting, Java Applets (less common now).

**Security**: Client-side scripts run in a **sandbox** – they cannot access the client's disk or execute programs outside the browser.

**JavaScript Example:**
```html
<script type="text/javascript">
function validate() {
    var credits = document.getElementById("credits").value;
    if (isNaN(credits) || credits <= 0 || credits >= 16) {
        alert("Credits must be a number between 1 and 15");
        return false;
    }
    return true;
}
</script>
```
This validates input locally without sending a request to the server.

#### 32.4.2. Server-Side Scripting

Runs on the server. Used for:
- Connecting to databases
- Extracting data
- Generating HTML dynamically
- Implementing business logic

**Common technologies:**
- **JSP (Java Server Pages)**
- **PHP**
- **ASP (Active Server Pages)** – now largely replaced
- **Python** (as a scripting language)
- **Ruby on Rails**

**How it works:**
1. Server receives HTTP request.
2. A preprocessor selects the appropriate script.
3. The scripting engine executes the script.
4. The script queries the database, computes, and fills an HTML template.
5. The final HTML is returned to the client.

## Module 33: Application Design and Development – Part 3: SQL and Native Language

### 33.1. The Core Challenge

The three-tier architecture has:
- **Frontend**: HTML/CSS/JavaScript (in browser)
- **Business Logic**: Java, Python, C++ (object-oriented)
- **Data Layer**: SQL (relational)

The challenge is **bridging the gap** between the object-oriented programming languages and the relational database with SQL.

**Key steps:**
1. Connect to the database server with credentials.
2. Send SQL commands.
3. Fetch results (tuples) into the programming language.
4. Process the results.
5. Close the connection.

### 33.2. Two Approaches to Database Connectivity

#### 33.2.1. Connectionist Framework

- Uses a **driver** (API library) to connect the native language to the database.
- The SQL commands are passed as strings through API calls.
- Examples:
  - **ODBC (Open Database Connectivity)**: Works with almost all programming languages.
  - **JDBC (Java Database Connectivity)**: Java-specific.
  - **OLEDB, ADO.NET**: Microsoft alternatives.

#### 33.2.2. Embedded SQL

- SQL commands are embedded directly within the native language program using special syntax (e.g., `EXEC SQL` or `#sql`).
- A preprocessor translates the embedded SQL into API calls.

### 33.3. ODBC (Open Database Connectivity)

**Definition**: A standard API for accessing databases that is both database-system and operating-system independent.

**History:**
- Developed by Microsoft and Simba Technology.
- Later incorporated into the SQL standard (1990s).

**Workflow:**
1. Open a connection.
2. Send queries and updates.
3. Get results.

**Python ODBC Example (pyodbc):**
```python
import pyodbc

# Connect
conn = pyodbc.connect(
    'DRIVER={SQL Server};SERVER=localhost;DATABASE=mydb;UID=user;PWD=password'
)

# Create cursor
cursor = conn.cursor()

# Execute SQL
cursor.execute("CREATE TABLE products (id INT, name VARCHAR(20))")
cursor.execute("INSERT INTO products VALUES (1, 'Product A')")
cursor.execute("SELECT * FROM products")

# Fetch results
row = cursor.fetchone()
while row:
    print(row)
    row = cursor.fetchone()

# Close
cursor.close()
conn.close()
```

### 33.4. JDBC (Java Database Connectivity)

**Definition**: Java-specific API for database access, part of Java Standard Edition.

**Key features:**
- Completely object-based.
- Uses `Statement` object as handle.
- Has exception handling via try-catch.
- Platform independent (because Java is platform independent).

**Workflow:**
1. Load driver and get connection.
2. Create statement object.
3. Execute query using statement.
4. Iterate through result set.
5. Close connection.

**Java JDBC Example:**
```java
import java.sql.*;

public class DBExample {
    public static void main(String[] args) {
        String url = "jdbc:postgresql://localhost:5432/mydb";
        String user = "user";
        String password = "password";

        try {
            Connection conn = DriverManager.getConnection(url, user, password);
            Statement stmt = conn.createStatement();

            // Create table
            stmt.execute("CREATE TABLE products (id INT, name VARCHAR(20))");

            // Insert data
            stmt.execute("INSERT INTO products VALUES (1, 'Product A')");

            // Query
            ResultSet rs = stmt.executeQuery("SELECT * FROM products");
            while (rs.next()) {
                System.out.println(rs.getInt("id") + ": " + rs.getString("name"));
            }

            conn.close();
        } catch (SQLException e) {
            e.printStackTrace();
        }
    }
}
```

### 33.5. Bridge Drivers

To connect frameworks that don't natively support each other:

- **ODBC to JDBC Bridge**: Application uses ODBC, but database supports JDBC.
- **JDBC to ODBC Bridge**: Application uses JDBC, but database supports ODBC.
- **OLEDB to ODBC**.
- **ADO.NET to ODBC**.

These bridge drivers allow flexible connections.

### 33.6. Embedded SQL

**Concept**: Write SQL statements directly inside a host language program using a special tag.

**Syntax examples:**
- `EXEC SQL` (C, C++)
- `#sql` (Java SQLJ)

**Basic structure:**
```c
EXEC SQL CONNECT TO server USER username USING password;

EXEC SQL
    SELECT sid, sname, sage
    INTO :sid, :sname, :sage
    FROM sailors
    WHERE sid = :sid;
```

**Host language variables in SQL:**
- Declared in the host language.
- Referenced in SQL with a colon prefix (`:variable_name`).

### 33.7. Cursors in Embedded SQL

**Purpose**: To handle query results that return multiple rows.

**Steps:**
1. **DECLARE** the cursor:
   ```sql
   EXEC SQL DECLARE c CURSOR FOR
       SELECT ID, name FROM instructor WHERE dept_name = :dept;
   ```
2. **OPEN** the cursor:
   ```sql
   EXEC SQL OPEN c;
   ```
3. **FETCH** rows:
   ```sql
   EXEC SQL FETCH c INTO :id, :name;
   ```
   Each FETCH retrieves the next tuple.
4. **CLOSE** the cursor:
   ```sql
   EXEC SQL CLOSE c;
   ```

### 33.8. Embedded SQL for Updates

You can use cursors to update data:

```sql
EXEC SQL DECLARE c CURSOR FOR
    SELECT * FROM instructor
    FOR UPDATE;
...
EXEC SQL UPDATE instructor
    SET salary = salary * 1.05
    WHERE CURRENT OF c;
```

This updates the current row being pointed to by cursor c.

### 33.9. Full Embedded SQL Example (C)

```c
#include <stdio.h>
#include "sqlca.h"

EXEC SQL INCLUDE SQLCA;

int main(int argc, char *argv[]) {
    EXEC SQL BEGIN DECLARE SECTION;
        char sid[10];
        char sname[50];
        int sage;
    EXEC SQL END DECLARE SECTION;

    if (argc > 1) {
        strcpy(sid, argv[1]);
    }

    EXEC SQL CONNECT TO 'mydb' USER 'user';

    EXEC SQL
        SELECT sname, sage
        INTO :sname, :sage
        FROM sailors
        WHERE sid = :sid;

    printf("Sailor %s: %s is %d years old\n", sid, sname, sage);

    EXEC SQL DISCONNECT;
    return 0;
}
```

This example retrieves a sailor's name and age by ID.

### 34.4. Complete Python Examples

#### 34.4.1. Connect Function

```python
def connect_db(database, user, password, host, port):
    try:
        conn = psycopg2.connect(
            database=database,
            user=user,
            password=password,
            host=host,
            port=port
        )
        print("Connected")
        return conn
    except Exception as e:
        print("Error:", e)
        return None
```

#### 34.4.2. Create Table

```python
def create_table(conn):
    cursor = conn.cursor()
    try:
        cursor.execute("""
            CREATE TABLE products (
                id INT PRIMARY KEY,
                name VARCHAR(50)
            )
        """)
        conn.commit()
        print("Table created")
    except Exception as e:
        print("Error:", e)
    finally:
        cursor.close()
```

#### 34.4.3. Insert Data

```python
def insert_data(conn, id, name):
    cursor = conn.cursor()
    try:
        cursor.execute(
            "INSERT INTO products (id, name) VALUES (%s, %s)",
            (id, name)
        )
        conn.commit()
        print("1 row inserted")
    except Exception as e:
        print("Error:", e)
    finally:
        cursor.close()
```

#### 34.4.4. Delete Data

```python
def delete_data(conn, id):
    cursor = conn.cursor()
    try:
        cursor.execute("DELETE FROM products WHERE id = %s", (id,))
        conn.commit()
        print(f"{cursor.rowcount} rows deleted")
    except Exception as e:
        print("Error:", e)
    finally:
        cursor.close()
```

#### 34.4.5. Update Data

```python
def update_data(conn, id, new_name):
    cursor = conn.cursor()
    try:
        cursor.execute(
            "UPDATE products SET name = %s WHERE id = %s",
            (new_name, id)
        )
        conn.commit()
        print(f"{cursor.rowcount} rows updated")
    except Exception as e:
        print("Error:", e)
    finally:
        cursor.close()
```

#### 34.4.6. Select Data

```python
def select_all(conn):
    cursor = conn.cursor()
    try:
        cursor.execute("SELECT * FROM products")
        rows = cursor.fetchall()
        for row in rows:
            print(row[0], row[1])
    except Exception as e:
        print("Error:", e)
    finally:
        cursor.close()
```

```html
<html>
<body>
<h2>User List</h2>
<table border="1">
    <tr>
        <th>Email</th>
        <th>Name</th>
        <th>Phone</th>
    </tr>
    {% for row in rows %}
    <tr>
        <td>{{ row[0] }}</td>
        <td>{{ row[1] }}</td>
        <td>{{ row[2] }}</td>
    </tr>
    {% endfor %}
</table>
<a href="/">Go Home</a>
</body>
</html>
```

Flask uses **Jinja2** templating engine, which allows `{% %}` for control statements and `{{ }}` for variable substitution.

### 35.1. Rapid Application Development (RAD)

**RAD** is an **agile software development** model that emphasizes quick iterations and customer involvement.

**Contrast with Waterfall:**
- **Waterfall**: Requirement → Design → Implementation → Testing → Delivery (sequential)
- **RAD**: Small chunks of each stage, rapid prototyping, customer feedback loop

**Phases of RAD:**
1. **Business Modeling**: Understand what the business needs (e.g., selling products, giving offers, bundling items).
2. **Data Modeling**: Define data entities, attributes, constraints (the database design we learned).
3. **Process Modeling**: Define workflows (e.g., item selection → cart → checkout → payment → order tracking).
4. **Testing and Turnover**: Prototype, get feedback, refine.

**Key principles:**
- Customer-in-the-loop
- Quick turnaround iterations
- Integration from the beginning

### 35.3. Performance Optimization

#### 35.3.1. Caching

**Caching** stores copies of computation results to avoid recomputation.

**Server-side caching:**
1. **JDBC Connection Pooling**: Reuse database connections instead of creating new ones for each request.
2. **Cache query results**: Store results of frequent queries (e.g., "Neeraj Chopra" search during Olympics).
3. **Cache generated HTML**: Store entire HTML pages for identical requests.

**Client-side caching:**
- **Web proxy caching**: The web proxy stores copies of responses and serves them to multiple clients without forwarding to the server.

**Challenge**: Cached data can become **stale** (outdated). Strategies needed to refresh or invalidate cache when underlying data changes.

### 35.4. Security

#### 35.4.1. SQL Injection

**SQL Injection** is a security vulnerability where malicious input is inserted into SQL queries.

**Example vulnerability:**
```python
query = "SELECT * FROM users WHERE name = '" + user_input + "'"
cursor.execute(query)
```
If user_input is `' OR '1'='1`, the query becomes:
```sql
SELECT * FROM users WHERE name = '' OR '1'='1'
```
This returns all users!

**Prevention**: Use **parameterized queries**:
```python
cursor.execute("SELECT * FROM users WHERE name = %s", (user_input,))
```

#### 35.4.2. Password Leakage

- Never store passwords in **clear text** in scripts or configuration files.
- Encrypt passwords.
- Be careful with backup files (e.g., `.bak` files) that may contain secrets.
- Restrict database access to specific IP ranges.

#### 35.4.3. Authentication

- Password-only authentication is weak.
- **Two-factor authentication** is essential: Password + SMS/OTP/biometric.
- Common second factors:
  - SMS one-time password
  - OTP device (hardware token)
  - Fingerprint, face recognition, iris scan

#### 35.4.4. Fine-Grained Authorization

**Problem**: SQL does not support **row-level authorization**. It only provides table and column-level grants.

**Example**: "Students can see their own grades but not others' grades" requires row-level authorization.

**Workarounds:**
- Enforce at the **application layer**: Include user ID in the query and filter rows.
- Oracle Virtual Private Database (VPD): Automatically adds predicates to enforce fine-grained authorization (not widely available in open-source systems).

#### 35.4.5. Audit Trails

**Audit trails** record events that have occurred (who did what, when).

**Purpose:**
- Detect security breaches.
- Repair damages.
- Investigate incidents.
- Meet regulatory requirements.

**Levels:**
- Database-level audit trails (transaction logs)
- Application-level audit trails (user actions)

### 35.5. Mobile Applications

#### 35.5.1. Unique Characteristics

Mobile devices have:
- Smaller form factors
- Touch screen interfaces
- Limited memory
- Limited computing power
- Limited battery
- Limited bandwidth
- Additional sensors (accelerometer, GPS, gyroscope)
- Gesture-based navigation

#### 35.5.2. Mobile Website vs. Native App

**Mobile Website:**
- Runs in mobile browser.
- Browser-based HTML.
- Designed for touch screen and small form factor.
- Easier to update (no app store).

**Native App:**
- Downloaded and installed.
- Runs directly on device.
- Platform-specific (iOS or Android).
- Can access device sensors.
- Better performance.
- Can work offline (partially).

#### 35.5.3. Three-Tier Architecture for Mobile

Similar to web, but with **local data** split:

1. **Presentation Layer**: User interface on mobile device.
2. **Business Logic Layer (Application Facade)**: Workflow and entities.
3. **Data Access Layer**:
   - **Local database** (small, cached data for offline/quick access)
   - **Remote database** (bulk data, accessed via network)

#### 35.5.4. Development Platforms

- **iOS**: Objective-C or Swift, Xcode IDE.
- **Android**: Java, Kotlin, Android Studio.
- **Cross-platform**: React Native, Flutter, Xamarin.
- **Hybrid apps**: Native shell + WebView for some content.

#### 35.5.5. Top-Level Design Issues for Mobile

1. Determine the device(s).
2. Note the resources (memory, battery, network).
3. Consider bandwidth carefully (minimize data transfer).
4. Decide architectural layers.
5. Select technology.
6. Define UI and navigation.
7. Maintain flow.
