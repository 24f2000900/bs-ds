## Module 36: Algorithms and Complexity Analysis

### 36.1. Introduction

The module revisits core concepts from algorithms and data structures courses, contextualizing them for database system design. The primary goal is to understand **what efficiency means**—how to compare different ways of organizing and accessing data in terms of time and space.

**Key questions addressed:**
- **Why analyze?** Resources are scarce; we want to do more with less.
- **What analyze?** Time complexity and space complexity given input size.
- **How analyze?** Using counting models, asymptotic analysis, recurrence relations, and Master's Theorem.
- **Where analyze?** Worst case, average case, best case, amortized case.
- **When to stop analyzing?** Related to lower bounds and NP-completeness (not covered in depth here).

### 36.2. Algorithms and Programs

**Algorithm:** A **finite sequence of well-defined steps** to solve a class of problems or perform a computation.

**Key properties:**
- **Finite**: Must terminate in finite time.
- **Well-defined**: Each step is unambiguous.
- **Computer implementable**: Can be translated to program code.

**Program:** A collection of instructions written in a programming language that implements an algorithm.

**Key difference:**
- An **algorithm must terminate** (provably).
- A **program may or may not terminate**.

**Examples of non-terminating programs:**
- Operating systems (run until the system is shut down)
- Database management systems (must keep running)
- Web servers (continuously listening for requests)

These programs **embody** algorithms (which terminate) but themselves do not terminate.

### 36.3. Why Analyze Algorithms?

**Primary reasons:**

1. **Predict Performance**: Determine how long an algorithm will take to execute and how much space it will require.

2. **Compare Algorithms**: Choose the best among multiple algorithms for the same task.
   - Example: For sorting, compare Quicksort, Mergesort, Insertion Sort, Selection Sort, Heap Sort, etc.

3. **Provide Guarantees**: Prove that an algorithm will always perform within certain bounds.
   - Example: A Red-Black tree guarantees O(log n) insertion and deletion.

4. **Understand Theoretical Basis**: Lower bounds, NP-completeness, and other theoretical foundations help us know when we've reached optimal efficiency.

### 36.4. What to Analyze?

**Parameters:**
- Input size (n): Number of elements, size of input, value of input, or multiple inputs.

**Resources to measure:**
- **Time**: The most common analysis factor.
- **Space**: Critical for handheld devices and large databases.
- **Power**: Important for mobile devices.
- **Bandwidth**: Important for network connections.
- **Processor cycles**: Important for multi-threaded applications.

For database applications, we focus primarily on **time** and **space**.

**Example: Sum of n Numbers**

```c
int sum(int a[], int n) {
    int s = 0;                    // 1 assignment
    for (int i = 0; i < n; i++)   // n increments, n comparisons
        s += a[i];               // n additions
    return s;                    // 1 return
}
```

- **Time**: O(n) additions (dominant operation)
- **Space**: O(1) (just variables s and i)

**Key insight**: We identify the **dominant operation** (the one that scales with input size) and count its frequency.

### 36.5. Counting Model

The **counting model** is a straightforward approach to algorithm analysis:

1. **Identify operations**: Determine which operations are fundamental (comparisons, assignments, arithmetic operations, etc.).
2. **Count frequencies**: Count how many times each operation is executed.
3. **Estimate costs**: Assign a cost to each operation.
4. **Total running time**: Sum of (cost × frequency) for all operations.

**Example: Character Search**

```c
int find(char *str, char c) {
    for (int i = 0; i < strlen(str); i++) {
        if (str[i] == c) return 1;
    }
    return 0;
}
```

- Input size: n (length of string)
- `strlen(str)` is called n times, each taking O(n) → O(n²) total for strlen
- n comparisons (checking each character)
- Total: O(n + n²) = **O(n²)**

**Optimization:** Precompute `strlen(str)` once:
```c
int find(char *str, char c) {
    int len = strlen(str);    // O(n)
    for (int i = 0; i < len; i++) {
        if (str[i] == c) return 1;
    }
    return 0;
}
```
- Total: O(n + n) = **O(n)**

This simple optimization dramatically improves performance.

### 36.6. Asymptotic Analysis

Comparing actual execution times is impossible because they depend on:
- Processor clock speed
- Architecture
- Bus speed
- Compiler optimizations
- Operating system

Instead, we compare **growth rates**. If input size n grows, how does the running time grow?

**Key idea**: Ignore constant factors and lower-order terms. Focus on the dominant term.

**Big-Oh Notation (O)**: Upper bound
**Big-Omega (Ω)**: Lower bound
**Big-Theta (Θ)**: Tight bound

**Tilde (~) Approximation:**
```
f(n) ~ g(n) if lim(n→∞) f(n)/g(n) = 1
```

**Example:**
If T(n) = αn + βn² for some constants α, β:
- For large n, βn² dominates αn.
- So T(n) ~ βn², which we write as T(n) = O(n²).

### 36.7. Common Complexity Classes

| Notation | Name | Typical Example |
|---|---|---|
| O(1) | Constant | Array access by index |
| O(log n) | Logarithmic | Binary search |
| O(n) | Linear | Linear search, sum of n numbers |
| O(n log n) | Linear-Logarithmic | Merge Sort, Quick Sort |
| O(n²) | Quadratic | Nested loops, Insertion Sort |
| O(n³) | Cubic | Matrix multiplication |
| O(2ⁿ) | Exponential | Brute-force subset generation |

**Growth rates (from slowest to fastest):**
```
O(1) < O(log n) < O(n) < O(n log n) < O(n²) < O(n³) < O(2ⁿ)
```

**Practical guidance:**
- O(1) is excellent.
- O(log n) is very good.
- O(n) is acceptable.
- O(n log n) is typical for good sorting.
- O(n²) and above should be avoided for large n.
- Exponential is infeasible for large n.

### 36.8. Worst-Case and Average-Case Analysis

**Worst-Case**: What is the maximum time for any input of size n?
- Provides a **guarantee**.
- Most commonly used for database operations because we need predictable performance.

**Average-Case**: What is the expected time for a random input, assuming a specific probability distribution?
- Provides an **expectation**.
- More realistic for some applications but harder to compute.

**Other cases:**
- **Best-Case**: Minimum time (not very useful).
- **Amortized**: Average over a sequence of operations (important for dynamic structures like hash tables).

### 36.9. Complexity Chart for Common Data Structures

**Arrays:**
- Access: O(1) (index-based)
- Search: O(n) (linear search) or O(log n) (binary search on sorted array)
- Insertion: O(n) (worst case, at beginning)
- Deletion: O(n) (worst case, at beginning)

**Linked Lists:**
- Access: O(n) (must traverse from header)
- Search: O(n) (linear traversal)
- Insertion: O(1) if position known, O(n) if position unknown
- Deletion: O(1) if position known, O(n) if position unknown

**Stack/Queue (array or linked list implementation):**
- Push/Pop/Peek: O(1)
- Enqueue/Dequeue: O(1)

### 36.10. Complexity Chart for Sorting Algorithms

| Algorithm | Best Case | Average Case | Worst Case | Space |
|---|---|---|---|---|
| Selection Sort | O(n²) | O(n²) | O(n²) | O(1) |
| Insertion Sort | O(n) | O(n²) | O(n²) | O(1) |
| Bubble Sort | O(n) | O(n²) | O(n²) | O(1) |
| Merge Sort | O(n log n) | O(n log n) | O(n log n) | O(n) |
| Quick Sort | O(n log n) | O(n log n) | O(n²) | O(log n) |
| Heap Sort | O(n log n) | O(n log n) | O(n log n) | O(1) |

## Module 37: Data Structures – Linear Structures

### 37.1. What is a Data Structure?

A **data structure** specifies a way of organizing and storing **in-memory data** that enables efficient access and modification.

**Key distinction from databases:**
- A data structure is **in-memory**—it exists only as long as the program is running.
- A database is **persistent**—data survives program termination.

**Classification:**
1. **Linear Data Structures**: Elements are arranged in a linear sequence (arrays, linked lists, stacks, queues).
2. **Non-Linear Data Structures**: Elements are arranged in a hierarchical or networked structure (trees, graphs, hash tables).

**Designer's perspective:**
- **Container**: How data is stored (contiguous memory for arrays, nodes with pointers for linked lists).
- **Operations**: What can be done efficiently (create, insert, delete, find, access, close).

### 37.2. Computer Organization Basis

Linear data structures are built on two fundamental mechanisms:

1. **Indexed/Sequential Mechanism (Arrays):**
   - Data elements are placed in consecutive memory locations.
   - The next element is at the address immediately after the current one.
   - **Random Access**: Can go directly to any element using its index.
   - O(1) access time.

2. **Referential Mechanism (Linked Lists):**
   - Each memory location stores data plus an address (pointer/reference) to the next element.
   - Must follow pointers to reach a specific element.
   - **Sequential Access**: Must traverse from the header to find any element.
   - O(n) access time.

### 37.3. Arrays

An **array** is a collection of elements stored in contiguous memory locations.

**Key Properties:**
- **Fixed size**: Size is determined at creation.
- **Contiguous storage**: Elements are adjacent in memory.
- **Random access**: O(1) time to access any element using its index.
- **Packed from one end**: Typically elements are stored from index 0 to size-1.

**Representation:**
```
Indices:   [0] [1] [2] [3] [4] [5] [6] [7]
Memory:    +---+---+---+---+---+---+---+---+
           | A | B | C | D | E |   |   |   |   (unused space at right)
           +---+---+---+---+---+---+---+---+
```

**Operations:**

1. **Access**: O(1) — compute address as base + index × element_size.

2. **Insertion**:
   - **At the end**: O(1) if there is space.
   - **At arbitrary position**: O(n) — must shift all elements to the right.

3. **Deletion**:
   - **From the end**: O(1).
   - **From arbitrary position**: O(n) — must shift all elements to the left.

**Summary:**
- ✓ Very efficient for access (O(1))
- ✗ Inefficient for insert/delete (O(n) at arbitrary positions)
- ✗ Fixed size (not flexible)

### 37.4. Linked Lists

A **linked list** is a collection of **nodes**, where each node contains data and a pointer to the next node.

**Key Properties:**
- **Flexible size**: Can grow and shrink dynamically.
- **Non-contiguous**: Nodes are scattered in memory.
- **Sequential access**: Must traverse from the header.

**Representation:**
```
Header → [A|→] → [B|→] → [C|→] → [D|NULL]
```

**Operations:**

1. **Access**: O(n) — must start from header and follow pointers.

2. **Insertion**:
   - **At the beginning**: O(1) — just adjust pointers.
   - **At the end**: O(1) if we have a reference to the last node; O(n) if we need to find it.
   - **At arbitrary position**: O(1) if we know the position; O(n) if we need to find it.

3. **Deletion**:
   - **From the beginning**: O(1) — just adjust header pointer.
   - **From the end**: O(1) if we know the previous node; O(n) otherwise.
   - **From arbitrary position**: O(1) if we know the previous node; O(n) otherwise.

**Summary:**
- ✗ Inefficient for access (O(n))
- ✓ Very efficient for insert/delete once position is known (O(1))
- ✓ Flexible size

### 37.5. Stacks and Queues

**Stack**: Last In, First Out (LIFO)
- **Operations**: Push (add to top), Pop (remove from top), Peek (view top).
- All operations are O(1).
- Can be implemented using array or linked list.

**Queue**: First In, First Out (FIFO)
- **Operations**: Enqueue (add to rear), Dequeue (remove from front), Peek (view front).
- All operations are O(1).
- Can be implemented using array (circular) or linked list.

**Key insight**: Stack and queue are defined by their **operations**, not their container. The underlying data structure (array or linked list) can vary.

### 37.6. Search Algorithms

#### 37.6.1. Linear Search

Sequentially check each element until the key is found or the end is reached.

**Complexity:**
- Worst case: O(n) (key is not present or is at the end)
- Average case: O(n)
- Best case: O(1) (key is the first element)

**Works on any data structure** (array, linked list), ordered or unordered.

#### 37.6.2. Binary Search

Requires a **sorted array**. Repeatedly compare with the middle element and eliminate half the remaining elements.

**Complexity:**
- Worst case: O(log n)
- Average case: O(log n)

**How it works:**
1. Find the middle element.
2. If it matches the key, done.
3. If key < middle, search left half.
4. If key > middle, search right half.
5. Repeat until found or array size becomes 1.

**Why O(log n)?**
- With n elements, after each comparison, half is eliminated.
- After k comparisons, we have n/2^k elements remaining.
- We stop when n/2^k = 1, so k = log₂(n).

**Limitation**: Binary search requires a sorted **array** (random access). It does not work on linked lists because accessing the middle element is O(n).

### 37.7. Trade-off Analysis

| Operation | Array (Unordered) | Array (Ordered) | Linked List |
|---|---|---|---|
| Access | O(1) | O(1) | O(n) |
| Search | O(n) | O(log n) | O(n) |
| Insert | O(1) | O(n) | O(1) |
| Delete | O(n) | O(n) | O(1) |

**Fundamental problem**: Linear data structures have **contradictory performance**:
- Arrays: good for access and search (if ordered), bad for insert/delete.
- Linked lists: good for insert/delete, bad for access and search.

**Conclusion**: Linear data structures cannot provide efficient performance for **all** operations simultaneously. This motivates **non-linear data structures**.

## Module 38: Data Structures – Non-Linear Structures

### 38.1. Motivation for Non-Linear Data Structures

From the previous module, we saw:
- Arrays: O(1) access, O(log n) search (if ordered), but O(n) insert/delete.
- Linked lists: O(1) insert/delete, but O(n) access/search.

We need a data structure that offers good performance for **all** operations: access, search, insert, delete.

**Non-linear data structures** trade off these extremes to provide "moderate good" performance for every operation.

### 38.2. Types of Non-Linear Data Structures

1. **Graph**: Collection of nodes and edges connecting them.
2. **Tree**: Connected, acyclic graph (hierarchical structure).
3. **Hash Table**: Uses a hash function to map keys to array indices.
4. **Skip List**: Multiple layered linked lists for efficient search (not covered in detail here).

### 38.3. Graphs

A **graph** G = (V, E) consists of:
- V: A set of vertices (nodes).
- E: A set of edges (pairs of vertices).

**Types:**
- **Undirected**: Edge (u, v) means u is connected to v and vice versa.
- **Directed**: Edge (u, v) means connection from u to v (order matters).
- **Weighted**: Edges have associated weights.
- **Unweighted**: All edges have equal weight (typically 1).
- **Cyclic**: Contains cycles (paths that return to the starting vertex).
- **Acyclic**: No cycles.
- **Connected**: Every vertex is reachable from every other vertex.

**Applications:**
- ER diagrams (nodes = entities, edges = relationships)
- Facebook friendships (social network)
- Knowledge graphs
- Electrical circuits
- Transportation networks
- Water and sanitation networks

### 38.4. Trees

A **tree** is a special type of graph that is:
1. **Connected**: Every node is reachable from every other node.
2. **Acyclic**: There are no cycles.

**Additional properties:**
- **Rooted**: One node is designated as the root; all others have a unique parent.
- **Binary**: Maximum of 2 children per node.
- **N-ary**: Maximum of n children per node.
- **Balanced**: Subtrees at every node have approximately equal height.
- **Unbalanced**: Subtrees can have very different heights.

**Forest**: A collection of disconnected trees.

**Examples:**
- Family trees (genealogy)
- Composite attribute representation (name → first, middle, last)
- Date representation (date → day, month, year)
- File system directory structure
- HTML DOM

### 38.5. Tree Terminology

| Term | Definition |
|---|---|
| **Root** | Node without a parent |
| **Parent** | Predecessor of a node |
| **Child** | Node following a given node (successor) |
| **Leaf** | Node with no children |
| **Internal Node** | Node with at least one child |
| **Subtree** | Tree rooted at a non-root node |
| **Path** | Sequence of nodes following edges |
| **Sibling** | Nodes with the same parent |
| **Arity** | Number of children of a node |
| **Arity of Tree** | Maximum arity among all nodes |
| **Level** | Distance (number of edges) from the root |
| **Height** | Maximum level in the tree |

### 38.6. Key Facts About Binary Trees

1. **Number of edges**: A tree with n nodes has exactly **n-1** edges.

2. **Maximum nodes at level l**: 2^l
   - Level 0: 1 node
   - Level 1: up to 2 nodes
   - Level 2: up to 4 nodes
   - Level l: up to 2^l nodes

3. **Maximum nodes in a tree of height h**: 2^(h+1) - 1
   - Total = 1 + 2 + 4 + ... + 2^h = 2^(h+1) - 1

4. **Minimum height for n nodes**: ⌈log₂(n+1)⌉ - 1
   - Derived from n ≤ 2^(h+1) - 1

5. **Maximum height for n nodes**: n - 1
   - Occurs when every node has exactly one child (a "spine" tree).

**Asymptotic bounds on height:**
- Ω(log n) ≤ h ≤ O(n)

**Key insight**: The height of a binary tree determines the efficiency of operations. We want the height to be as small as possible (close to log n) to achieve O(log n) operations.

### 38.7. Hash Tables (Brief Overview)

A **hash table** stores key-value pairs where the key is transformed into an array index using a **hash function**.

**Key characteristics:**
- Uses an array as the underlying container.
- The hash function maps keys to indices in the array.
- Supports O(1) average-case search, insert, and delete.
- Handles collisions (two keys mapping to the same index) using chaining or open addressing.

**Detailed discussion deferred to a later module** (hashing for database indexing).

### 38.8. Binary Search Trees (BST)

#### 38.8.1. Motivation

We want a data structure that:
- Has O(log n) search (like binary search on sorted arrays)
- Has O(1) insert/delete (like linked lists)
- Supports all operations efficiently

The BST combines the benefits of both linear structures.

#### 38.8.2. Concept

A **Binary Search Tree** is a binary tree where for every node with value X:
- All values in the **left subtree** are **less than X**.
- All values in the **right subtree** are **greater than X**.
- This property holds at **every node** (not just the root).

#### 38.8.3. Construction

**Example: Build a BST from a sorted array of 7 elements:**
```
Elements: [A, B, C, D, E, F, G]
```
- Root: D (middle element)
- Left subtree: [A, B, C] with root B (middle of left half)
- Right subtree: [E, F, G] with root F (middle of right half)
- Left of B: A; Right of B: C
- Left of F: E; Right of F: G

```
        D
      /   \
     B     F
    / \   / \
   A   C E   G
```

This tree is **balanced**—both subtrees have equal size.

#### 38.8.4. Insertion Algorithm

To insert a value X:
1. Start at the root.
2. Compare X with the current node.
3. If X < current: go left.
4. If X > current: go right.
5. Repeat until we reach a null pointer.
6. Insert X at that position.

**Example: Inserting into an empty tree (order: E, L, P, H, A, N, T)**

```
E           L           P           H           A           N               T
 \         / \         / \         / \         / \         / \             / \
  L       E   P       E   P       E   P       E   P       E   P           E   P
                     /           /   /       /   /       /   / \         /   / \
                    L           L   H       L   H       L   H   T       L   H   T
                                            /           /               /   /   /
                                           A           A               A   A   N
```

Final BST for [E, L, P, H, A, N, T]:
```
        E
      /   \
     A     L
          / \
         H   P
            / \
           N   T
```

**Note**: This tree is **unbalanced**—the left subtree has 1 node, the right subtree has 5 nodes.

#### 38.8.5. Search Algorithm

To search for a key X:
1. Start at the root.
2. Compare X with the current node.
3. If equal: found.
4. If X < current: go left.
5. If X > current: go right.
6. If null: not found.

**Example: Search for H in the above tree**
- E: H > E → go right
- L: H < L → go left
- H: equal → found (3 comparisons)

**Example: Search for Q (not present)**
- E: Q > E → right
- L: Q > L → right
- P: Q > P → right
- T: Q < T → left (null) → not found (4 comparisons)

#### 38.8.6. Complexity Analysis

**Worst case**: O(h) where h is the height of the tree.

**Best case (balanced)**: h = O(log n), so search is O(log n).

**Worst case (unbalanced)**: h = O(n) (spine tree), so search is O(n).

**Insert**: O(h) = O(search + pointer manipulation) = O(h).

**Delete**: O(h) = O(search + pointer adjustment) = O(h).

#### 38.8.7. Balancing

The key to making BST efficient is **balancing**—keeping subtrees at approximately equal size at every node.

When balanced:
- Height = O(log n)
- All operations (search, insert, delete) = O(log n)

Balanced BSTs include:
- **AVL Trees**
- **Red-Black Trees**
- **B-Trees** (used in databases)

These guarantee O(log n) even in the worst case.

### 38.9. Linear vs Non-Linear Data Structures Comparison

| Aspect | Linear | Non-Linear |
|---|---|---|
| Arrangement | Direct, sequential | Hierarchical, networked |
| Levels | Single level | Multiple levels |
| Implementation | Simpler | More complex |
| Traversal | One way (forward/backward) | Multiple ways (DFS, BFS, in-order, pre-order, post-order) |
| Space | O(n) | O(n) |
| Average-case operations | Some O(1), some O(n) | O(log n) for BST |
| Worst-case operations | Same as average | BST can degrade to O(n) |

**Balanced BSTs (like Red-Black, B-Tree)** achieve O(log n) for all operations in both average and worst cases, with O(n) space—the optimal combination.

## Module 39: Storage and File Structure – Part 1: Physical Storage

### 39.1. Overview of Storage Hierarchy

Database systems require persistent storage of large volumes of data. The storage hierarchy includes:

| Tier | Media | Speed | Cost | Capacity | Volatility |
|---|---|---|---|---|---|
| Primary | Cache | Fastest (ns) | Highest | Very small (MB) | Volatile |
| Primary | Main Memory | Fast (10-100 ns) | High | Small (GB) | Volatile |
| Primary | Flash Memory | Moderate | Moderate | Medium (GB-TB) | Non-volatile |
| Secondary | Magnetic Disk | Moderate (ms) | Moderate | Large (TB) | Non-volatile |
| Tertiary | Optical Disk | Slow | Low | Large (GB) | Non-volatile |
| Tertiary | Magnetic Tape | Slowest | Lowest | Very large (PB) | Non-volatile |

### 39.2. Volatile vs Non-Volatile Storage

**Volatile Storage:**
- Loses data when power is removed.
- Examples: Cache, Main Memory.
- Used for computation, not for persistent database storage.
- Too expensive and too small to store entire databases.

**Non-Volatile Storage:**
- Survives power failure.
- Examples: Flash, Magnetic Disk, Tape, Optical.
- Used for persistent database storage.

### 39.3. Flash Memory

**Characteristics:**
- Like main memory but survives power failure.
- Data can be written to a location only once; must be **erased** before rewriting.
- Erasure is done in **blocks** (not individual bytes).
- Limited number of erase cycles (10,000 to 1,000,000).

**Types:**
- **NOR Flash**: Allows random byte access; used for code storage.
- **NAND Flash**: Faster for sequential access; used for data storage (SD cards, SSDs).

**Applications:**
- USB drives (pen drives)
- SD cards (mobile devices)
- Solid State Drives (SSDs)

### 39.4. Magnetic Disks

The **primary persistent storage** for databases.

#### 39.4.1. Physical Structure

A magnetic disk is a set of spinning platters with read/write heads.

**Components:**
- **Platters**: Circular disks coated with magnetic material. Both sides can be used.
- **Spindle**: Rotates the platters at high speed (typically 5,400 to 15,000 RPM).
- **Read/Write Heads**: Positioned over each platter surface. They move in tandem (together) along a radius.
- **Arms**: Hold the heads; all heads move together.

**Tracks:**
- Concentric circles on each platter surface.
- Data is stored on tracks.
- 50,000 to 100,000 tracks per platter.

**Sectors:**
- Radial lines divide tracks into sectors.
- Each sector holds a fixed amount of data (typically 512 bytes).
- Outer tracks have more sectors than inner tracks (because they are longer).

**Blocks/Clusters:**
- Groups of consecutive sectors on the same track.
- The unit of data transfer between disk and memory.

**Cylinder:**
- The collection of all tracks at the same radial position across all platter surfaces.
- Since all heads move together, they all point to tracks at the same radius.
- Viewing these tracks together forms a virtual cylinder.

#### 39.4.2. Disk Controller

The **disk controller** manages the disk:

1. Receives commands to read/write sectors.
2. Moves the read/write head to the correct track.
3. Waits for the correct sector to rotate under the head.
4. Reads or writes the sector.
5. Uses **checksums** to verify data integrity.
6. **Remaps bad sectors** to spare sectors.

**Disk interfaces:**
- ATA/SATA (older desktop)
- SCSI (enterprise)
- NVMe (modern SSDs)

**Storage networks:**
- SAN (Storage Area Network): Dedicated high-speed network for storage.
- NAS (Network Attached Storage): Storage device connected to the network.

#### 39.4.3. Disk Performance Metrics

**Access Time:** Time from issuing a read/write request to the start of data transfer.

- **Seek Time**: Time to position the head on the correct track.
  - Average seek time: 4-11 milliseconds (about half of the worst case).
- **Rotational Latency**: Time for the correct sector to rotate under the head.
  - Average latency: 4-11 milliseconds (half a rotation).

**Total Access Time** = Seek Time + Rotational Latency

**Data-Transfer Rate:**
- Rate at which data can be transferred from disk to memory.
- Typically 25-100 MB/s (inner tracks have lower rates than outer tracks).
- Depends on the disk and the interface.

**Mean Time To Failure (MTTF):**
- Average time between disk failures.
- Typically 3-5 years (1-1.2 million hours for new disks).
- If you have 1000 disks with MTTF of 1200 hours, one fails every 1.2 hours.

### 39.5. Magnetic Tapes

**Characteristics:**
- Sequential access only—must rewind/forward to read a specific position.
- Very high capacity (multiple petabytes per tape).
- Very low cost per gigabyte.
- High transfer rates for sequential reads.

**Use:** Tertiary storage for archival and backup.

**Formats:**
- DAT (Digital Audio Tape)
- LTO (Linear Tape-Open)
- Various enterprise formats

### 39.6. Optical Storage

**Examples:** CD-ROM, DVD, Blu-ray.

**Characteristics:**
- Write-once, read-many (WORM) devices.
- Good for archival distribution.
- Not suitable for live database storage (slow write).

### 39.7. Cloud Storage

**Definition:** Storage services provided over the Internet.

**Examples:** Google Drive, Amazon Drive, OneDrive, Dropbox, Evernote.

**Advantages:**
- Lower upfront cost (pay-as-you-go model)
- Reliability (provider uses high-end storage like RAID)
- File sharing and accessibility from anywhere
- Backup and recovery handled by provider
- Safe from site disasters

**Disadvantages:**
- Requires network connectivity
- Potential latency
- Security and privacy concerns

### 39.8. Solid State Drives (SSDs)

**Characteristics:**
- Flash-based memory, no moving parts.
- Very fast (10-100 times faster than HDDs).
- Larger than main memory.
- More expensive per gigabyte than HDDs.

**Comparison HDD vs SSD:**

| Parameter | HDD | SSD |
|---|---|---|
| Access time | 5-10 ms | 0.1 ms |
| Transfer rate | 50-200 MB/s | 500-3000 MB/s |
| Cost per GB | Lower | Higher |
| Durability | Mechanical failure | Limited write cycles |
| Power consumption | Higher | Lower |
| Noise | Audible | Silent |

### 39.9. Future Storage Technologies

#### 39.9.1. DNA Digital Storage

- Uses synthesized DNA strands for data storage.
- DNA has four nucleotides: A, T, G, C.
- Data is encoded using these nucleotides.
- **Very high density**: 1 cubic nanometer can store an exabyte of data.
- Currently expensive and slow, but promising.

#### 39.9.2. Quantum Storage

- Quantum mechanical versions of memory use **qubits**.
- Expected to provide higher capacity and efficiency.
- Still in early research and development.

## Module 40: Storage and File Structure – Part 2: File Structure

### 40.1. Database File Organization

**Hierarchical containment:**
```
Database
  └── Files (one per relation)
        └── Records (tuples)
              └── Fields (attributes)
```

**Simplified assumptions (for starting):**
- Records are of **fixed size**.
- Each file contains records of **one type** only.
- Different relations are stored in different files.

These assumptions are relaxed later (variable-length records, multiple record types).

### 40.2. Fixed-Length Records

Records of the same size are stored contiguously.

**Accessing record i:**
```
Address of record i = Base_Address + i × Record_Size
```

This is O(1) access, just like an array.

**Key constraint:** Records should not cross block boundaries.

#### 40.2.1. Deletion Strategies

When deleting a record, we have three options:

1. **Compaction (Move up):** Shift all subsequent records to fill the gap.
   - Expensive: O(n) moves.

2. **Move Last Record:** Move the last record into the deleted position.
   - Cheaper: O(1) moves.

3. **Free List:** Mark the deleted record as free and link it to a free list.
   - Store the address of the first deleted record in the file header.
   - Use the deleted record's space to store the address of the next deleted record.
   - No extra space overhead.
   - Insertion checks the free list first.
   - Most common strategy.

### 40.3. Variable-Length Records

**Reasons for variable-length records:**
- Multiple record types in a file.
- Fields that are variable length (varchar).
- Repeating fields (e.g., arrays).

**Representation:**
- Fixed-length attributes come first.
- Variable-length attributes are stored after all fixed-length attributes.
- Each variable-length attribute is stored as `(offset, length)` pairs.
- Null values are represented by a **null bitmap** (a sequence of bits indicating which fields are null).

**Example:**
```
instructor(ID: 10101, name: "Srinivasan", dept_name: "Comp. Sci.", salary: 65000)

Fixed part:   [ID: 10101] [salary: 65000] [name_offset: 21] [name_length: 10] [dept_offset: 31] [dept_length: 10]
Variable part: "Srinivasan" "Comp. Sci."
```

The fixed part allows direct access to any attribute (including the offsets for variable-length attributes).

### 40.4. Slotted Page Structure

A **slotted page** organizes records within a block for efficient space management.

**Structure:**
```
+---------------------------+
| Header                    |
| - Number of records       |
| - End of free space       |
| - Location and size of    |
|   each record             |
+---------------------------+
| Record entries (points to |
|   actual data)            |
+---------------------------+
| Free space                |
+---------------------------+
| Records (data)            |
+---------------------------+
```

**Advantages:**
- Records can be moved around within the block (e.g., for compaction) without affecting external references.
- Only the header needs to be updated when records are moved.
- Efficient for insertion and deletion.

### 40.5. File Organizations

**Ways to organize records in a file:**

1. **Heap**: Records are placed anywhere there is space. No ordering.

2. **Sequential**: Records are stored in order of a **search key** value.

3. **Hashing**: A hash function determines the storage location based on some attribute(s).

4. **Multi-Table Clustering**: Records from multiple related relations are stored together.

### 40.6. Sequential Organization

Records are arranged in a logical order based on the value of a **search key** (typically the primary key).

**Advantages:**
- Good for sequential processing (reading records in key order).
- Supports efficient range queries.

**Deletion:**
- Just adjust pointers (like linked list deletion).
- Deleted space goes to the free list.

**Insertion:**
- Check the free list.
- If no free space, allocate an **overflow block**.
- Insert the record and update pointers.
- May require periodic reorganization to maintain sequential order.

### 40.7. Multi-Table Clustering

Store related records from multiple relations in the same block to reduce I/O during joins.

**Example:**
```
department(Comp. Sci., Taylor, 100000)
instructor(10101, Srinivasan, Comp. Sci., 65000)
instructor(10200, Katz, Comp. Sci., 75000)
instructor(10300, Brandt, Comp. Sci., 80000)
department(Biology, Watson, 90000)
instructor(10400, Crick, Biology, 72000)
```

Here, `department` records are immediately followed by all `instructor` records from that department.

**Advantages:**
- Efficient for queries that join department and instructor.
- Good for queries involving a single department and its instructors.

**Disadvantages:**
- Inefficient for queries involving only departments (must skip instructor records).
- Variable-size records due to mixed types.

**Key insight**: Works well because of the **many-to-one** relationship (many instructors to one department).

### 40.8. Data Dictionary (System Catalog)

The **data dictionary** stores **metadata**—data about data.

**Contents:**
- Names of all relations and attributes
- Data types and lengths of attributes
- Views
- Integrity constraints
- User information
- Index information
- Physical storage information

**Key tables in the data dictionary:**

**Relation Metadata:**
| Relation Name | Number of Attributes | Storage Organization | Location |
|---|---|---|---|
| instructor | 4 | Heap | /data/instructor.dat |
| department | 3 | Sequential | /data/department.dat |

**Attribute Metadata:**
| Relation Name | Attribute Name | Domain Type | Position | Length |
|---|---|---|---|---|
| instructor | ID | CHAR(5) | 1 | 5 |
| instructor | name | VARCHAR(20) | 2 | 20 |

**Index Metadata:**
| Index Name | Relation Name | Index Type | Indexing Attributes |
|---|---|---|---|
| idx_instructor_id | instructor | B-Tree | ID |

**User Metadata:**
| User Name | Encrypted Password | Group |
|---|---|---|
| amit | ****** | admin |

**Importance:**
- Every query processing consults the data dictionary.
- The data dictionary is itself stored in the database (recursive).
- It should be optimized for fast access (may be kept in main memory).

### 40.9. Buffer Management

**Motivation:**
- Data is on disk (slow, large).
- Computation is in main memory (fast, small).
- We need to bring data from disk to memory, operate, and write back.
- Minimizing disk I/O is critical for performance.

**Solution: Buffer Pool**

A **buffer pool** is a region of main memory allocated to store copies of disk blocks.

**Buffer Manager:** The subsystem responsible for allocating and managing buffer space.

#### 40.9.1. Buffer Manager Operations

When a program requests a block:

1. **If block is already in buffer:**
   - Return the buffer address directly.

2. **If block is not in buffer:**
   - If there is free space in the buffer: allocate it.
   - If the buffer is full: choose a block to replace.
   - If the replaced block was modified: write it back to disk.
   - Read the requested block from disk into the allocated space.
   - Return the buffer address.

#### 40.9.2. Replacement Policies

**LRU (Least Recently Used):**
- Replace the block that has not been used for the longest time.
- Based on the assumption that past reference patterns predict future references.
- Standard strategy for operating systems.

**Limitations of LRU for databases:**
- Certain query patterns (like nested loop joins) repeatedly access the same blocks.
- LRU may evict blocks that will be needed again soon.

**Specific strategies:**
- **Pinned Block**: A block that cannot be written back to disk (marked as "pinned").
- **Toss-Immediate**: Free the space as soon as the block is processed.
- **Most Recently Used (MRU)**: Replace the block used most recently.
- **Data dictionary blocks kept in memory**: Since the data dictionary is accessed frequently, its blocks are given dedicated buffer space.

#### 40.9.3. Additional Considerations

- **Forced Output**: Write back modified blocks for recovery purposes.
- **Statistical Information**: Use probabilities to inform replacement decisions.
- **Query Optimizer Hints**: The optimizer may specify replacement strategies for specific queries.
