## 1. What is a Database and Why Should You Care?

A **database** is an organized collection of data. Data could be anything: names, account balances, train reservations, emails, Facebook posts, etc. A **Database Management System (DBMS)** is the software that helps you store, retrieve, and manage that data efficiently.

Databases are everywhere:
- **Banking** – UPI, net banking, Paytm
- **Reservations** – IRCTC, airlines
- **Universities** – registrations, grades
- **E‑commerce** – Amazon, Flipkart
- **Social media** – Facebook, Instagram, WhatsApp
- **Email** – Gmail, Yahoo Mail

Without databases, none of these would work at scale.

---

## 2. The Old Way: Storing Data in Files (and Why It Fails)

Before DBMS, people stored data in plain **files** (like CSV or Excel sheets). While this is simple, it has major problems:

| Problem                | What it means |
|------------------------|---------------|
| **Data redundancy**    | The same piece of data is copied in many places. |
| **Inconsistency**      | When you update one copy, you might forget to update another. |
| **Difficulty accessing** | To find something, you have to manually read through files. |
| **Isolation**          | Multiple files with different formats are hard to integrate. |
| **Integrity issues**   | Enforcing rules (e.g., minimum balance in an account) is buried in the code and can be broken. |
| **Atomicity problems** | If a transaction (like transferring money) is interrupted halfway, data becomes inconsistent (money vanishes). |
| **Concurrency**        | Multiple users accessing the same file at the same time leads to conflicts (e.g., double booking). |
| **Security**           | File‑system passwords are weak; you can’t give different access levels to different people. |

Databases solve all these problems through built‑in mechanisms.

## 3. A Quick Look Back: How We Managed Records Over Time

### Physical Records (Bookkeeping)
- Ledgers, journals, file cabinets.
- Fragile: termites, fire, humidity.
- Not scalable; searching is slow.
- No real security or consistency.

### Spreadsheets (Excel, Google Sheets)
- Better than paper, but still limited.
- Max rows, no true multi‑user access, hard to enforce business rules, limited permissions.

### Electronic Databases – History
- **1950s‑60s**: Magnetic tapes, punch cards.
- **1970s**: Hard disks, CODASYL, relational model proposed by Ted Codd.
- **1980s**: SQL became standard; relational databases (RDBMS) took over.
- **1990s**: Internet made databases global.
- **2000s**: E‑commerce boom, unstructured data (images, videos) → NoSQL databases born.
- **2010s**: Data science, big data, cloud databases.

## 4. Example: File‑based vs. Database for a Bank Transfer

The professor showed two ways to implement a simple money transfer (debit account A, credit account B):

### Using Python with CSV files
- You must open the file handles manually.
- Code to read, check balance, update both accounts, write back to the file.
- If the system crashes **after debiting but before crediting**, the file is left in an inconsistent state and you can’t undo it easily.
- You have to close all files yourself.

### Using SQL (a DBMS)
- You just write `UPDATE`, `INSERT`, and `COMMIT` statements.
- The DBMS automatically handles file I/O, caching, and crash recovery.
- If a crash happens before `COMMIT`, the DBMS **rolls back** the partial changes — the debited money reappears. This is **atomicity**.
- No need to manage file handles; everything is cleaner and more reliable.

**Key takeaway**: DBMS provides **atomic transactions**, **crash recovery**, and **less code**, which are extremely difficult to implement manually.

## 5. Levels of Abstraction: How a DBMS Hides Complexity

A DBMS organizes data at three levels:

1. **Physical Level**  
   - The actual bits and bytes on disk, file structures, indexes.  
   - “How is the data stored physically?”

2. **Logical Level**  
   - What data is stored and how it’s related.  
   - Example: an `instructor` has an `ID`, `name`, `dept_name`, `salary`.  
   - Programmers work mostly at this level.

3. **View Level**  
   - A customized window into the data.  
   - You can **hide** certain fields (e.g., salary) and even show **computed** values (e.g., age from date of birth).  
   - Different users see different views of the same underlying data.

**Data Independence** means changes at one level don’t drastically affect the others. For example, moving the database to a new disk doesn’t break the logical tables.

## 6. Schema vs. Instance

- **Schema** = the **blueprint** or structure.  
  - What tables exist, their columns, data types.  
  - Like the design of a building.  
- **Instance** = the **actual data** at a moment in time.  
  - The rows currently in the table.  
  - Like the people living in the building.

- **Physical schema** – how data is stored in files.  
- **Logical schema** – the overall logical structure of all tables.

## 7. Data Models: How We Describe Data

A **data model** is a set of tools to describe data, relationships, rules, and constraints.

Important models:
- **Relational Model** – uses **tables** (relations). This is the main focus of the course.
- **Entity‑Relationship (ER) Model** – a graphical way to design databases before creating tables.
- **Object‑based models** – extend relational model with complex data types (like objects in OOP).
- **Semi‑structured models** (XML, JSON) – used for data interchange and NoSQL.
- **Graph models** – for highly connected data (social networks).
- **NoSQL models** – for huge, unstructured datasets.

### The Relational Model (in a nutshell)
- Data is stored in **tables** (relations).
- Columns = **attributes** (e.g., `name`, `salary`).
- Rows = **records** (or tuples).
- All values are **atomic** (simple, like a number or text).

## 8. DDL, DML, and SQL

### Data Definition Language (DDL)
- Used to **define / modify** the schema.
- Example: `CREATE TABLE instructor (...);`
- The schema definitions themselves are stored in the database inside a **data dictionary**.

### Data Manipulation Language (DML)
- Used to **query** and **update** the data.
- Example: `SELECT`, `INSERT`, `UPDATE`, `DELETE`.
- Often called a **query language**.

### SQL (Structured Query Language)
- Combines DDL, DML, and more.
- It is the standard language for relational databases.
- It is **not** Turing‑complete – you can’t write every possible program in SQL alone. So for complex logic, SQL is **embedded** in a host language like Python, Java, C, etc.

## 9. Database Design – The Key Challenge

**Goal**: Create a set of “good” tables that avoid redundancy and anomalies.

### Example of a Bad Design
A single table with `faculty` info + `department` info repeated for every faculty member:
- Same building and budget repeated for each professor in the same department → **redundancy**.
- If the department budget changes, you must update many rows → possible **update anomaly**.
- If you delete the last professor of a department, you lose all department info → **deletion anomaly**.

### Solution: Split into two tables
- `instructor (ID, name, dept_name, salary)`
- `department (dept_name, building, budget)`

Now each piece of information appears only once. This process is called **normalization**.

Two design approaches are taught:
1. **Entity‑Relationship (ER) Model** – captures the business requirements in a diagram.
2. **Normalization Theory** – a set of formal rules to refine tables and eliminate redundancies.

The course will cover both.

## 10. The Database Engine – What’s Under the Hood

A DBMS has three core internal components:

### A. Storage Manager
- Translates logical data into **files** on disk.
- Manages **indexes** and **hashing** for fast retrieval.
- Handles **buffering** (moving data between disk and memory).
- Ensures data is stored efficiently and can be accessed quickly.

### B. Query Processor
- Takes an SQL query and processes it in steps:
  1. **Parse** – check syntax.
  2. **Translate** – convert SQL to an internal form (like relational algebra).
  3. **Optimize** – find the most efficient way to execute the query.
  4. **Execute** – actually fetch or modify data.
- An **evaluation engine** collects statistics to help future queries.

### C. Transaction Manager
- Ensures that **transactions** (logical units of work, e.g., a money transfer) are:
  - **Atomic** – all or nothing.
  - **Consistent** – the database stays correct before and after.
  - **Isolated** – concurrent transactions don’t interfere.
  - **Durable** – once committed, changes survive crashes.
- Handles **concurrency** so multiple users don’t corrupt data.
- Manages **backup and recovery**.

## 1. The Relational Model: Fundamentals (Module 6)

The relational model organizes data into **tables** (formally called **relations**). Each table represents an entity or a relationship between entities. This model is the backbone of almost all modern database systems.

### 1.1 Attributes, Domains, and Tuples

- **Attribute**: A column of a table, like `name`, `age`, or `salary`. It has a name and a **domain** (the set of allowed values).
- **Domain**: The type and range of values an attribute can take. For example, `salary` might be a positive number; `ID` might be a five-character alphanumeric string. Domains are atomic—they cannot be split into sub-parts in the basic relational model. A date like `27-03-1997` is treated as an atomic value, even though it contains day, month, and year; the database provides a special `DATE` data type for it.
- **Tuple**: A row in the table. It is an ordered list of values, one for each attribute. A tuple must be distinct from all other tuples (no duplicate rows) because a relation is a **set** of tuples.
- **Null**: A special marker meaning “unknown” or “not applicable”. An attribute can be nullable if we allow it to be left empty (e.g., a student may not have a passport number). A primary key attribute cannot be null.

#### Example Relation: `student`

| roll_number | first_name | last_name | dob        | passport_no | aadhaar   | dept_name |
|-------------|------------|-----------|------------|-------------|-----------|-----------|
| 14CS92P01   | Jatin      | Chopra    | 1997-03-27 | NULL        | 123456789012 | CSE       |
| 14EC45R02   | Anjali     | Sharma    | 1996-08-15 | A1234567    | 234567890123 | ECE       |

### 1.2 Schema and Instance

- **Schema**: The logical design of the table—its name, its attributes, and their domains (types). For the `student` table, the schema is `student(roll_number: alphanumeric, first_name: alpha, ...)`. Formally: `R(A1:D1, A2:D2, ..., An:Dn)`.
- **Instance**: The actual data stored in the table at a given moment. It changes over time (tuples are inserted, updated, deleted), but the schema usually stays fixed.

A relation is a subset of the Cartesian product of the domains:  
\( R \subseteq D_1 \times D_2 \times \dots \times D_n \).  
Each tuple `t` satisfies `t[Ai] ∈ Di`.

### 1.3 Keys

Keys are attributes (or sets of attributes) used to identify rows uniquely.

- **Superkey**: Any set of attributes whose values together are unique for each tuple. For `student`, `{roll_number, first_name}` is a superkey, and `{roll_number, first_name, last_name}` is also one.
- **Candidate Key**: A minimal superkey—removing any attribute destroys the uniqueness property. For `student`, `{roll_number}` is a candidate key, and `{aadhaar}` is also a candidate key if every student has one. `{first_name, last_name}` might be a candidate key if the combination is guaranteed unique, but in practice it often isn’t.
- **Primary Key**: The candidate key chosen by the designer to be the main identifier. It cannot be null. From the candidate keys, we pick one as primary key; for `student`, `roll_number` is a natural choice because it also encodes useful information (year, department, etc.) in a composite way. Aadhaar would also work but lacks that encoding benefit.
- **Foreign Key**: An attribute in one table that refers to the primary key of another table. It enforces referential integrity. For example, `dept_name` in `student` could be a foreign key referencing `department.dept_name`. A foreign key can be null (if the relationship is optional) or may have cascading actions.

**Alternate / Secondary Keys**: Candidate keys not chosen as primary key.

**Surrogate Key**: A synthetic key introduced when no natural attribute or combination can serve as a unique identifier, or when it’s simpler. For an `order` table, an `order_id` is generated by the system because multiple orders may have identical natural attributes (same item, same customer, same time). Surrogate keys are often auto-increment integers.

**Composite Key**: A key consisting of more than one attribute, e.g., `{first_name, last_name}`.

## 2. Relational Algebra: The Formal Query Language (Module 7)

Relational algebra is a procedural query language: you specify **how** to get the answer by applying a sequence of operations. Each operation takes one or two relations as input and produces a new relation as output. This closure property allows composition.

### 2.1 Fundamental Operations

#### 2.1.1 Select (σ)
Chooses **rows** that satisfy a predicate.

`σ<condition>(R)`

- Condition is a Boolean expression using comparisons (`=`, `<`, `>`, `≤`, `≥`, `≠`) and logical connectives (`∧`, `∨`, `¬`).
- Example: `σ_{dept_name='Comp. Sci.' ∧ salary > 80000}(instructor)` returns instructors in the CS department with salary > 80000.

#### 2.1.2 Project (π)
Chooses **columns** (attributes) and removes duplicate rows (since output is a set).

`π<attribute list>(R)`

- Example: `π_{name, dept_name}(instructor)` lists name and department of all instructors. If multiple instructors have the same name and dept_name, only one row appears.

#### 2.1.3 Union (∪)
Combines two relations with identical attribute sets (union-compatible). Duplicates are eliminated.

`R ∪ S`

#### 2.1.4 Set Difference (−)
Returns tuples in R that are not in S. Union-compatible.

`R − S`

#### 2.1.5 Cartesian Product (×)
Produces all combinations of a tuple from R and a tuple from S. The resulting schema is the concatenation of attributes of R and S. If attributes have the same name, they are disambiguated by prefixing the relation name (e.g., `R.A`, `S.A`).

`R × S`

- If R has `m` tuples, S has `n` tuples, product has `m × n` tuples.
- Usually combined with a Select to perform a **join** (see below).

#### 2.1.6 Rename (ρ)
Changes the name of a relation or its attributes.

`ρ_{S}(R)` — rename relation R to S.  
`ρ_{S(B1, B2, ...)}(R)` — rename and also change attribute names.

Useful for self-joins, where you need to distinguish two copies of the same relation.

### 2.2 Natural Join (⋈)

The natural join is a binary operation that combines tuples from two relations based on **equal values of their common attributes**. It is a very common and powerful operation.

`R ⋈ S`

- Let `A` be the set of common attribute names (R ∩ S). For each pair of tuples `t_r ∈ R` and `t_s ∈ S`, if `t_r[A] = t_s[A]` for all attributes in the intersection, then the concatenated tuple is included.
- In the result, the duplicate common attributes are removed (only one copy is kept).
- Example: `instructor ⋈ teaches` will join on `ID` (assuming `ID` is the only common attribute). The result combines instructor details with the courses they teach.

Natural join is equivalent to: Cartesian product → Select on equality of common attributes → Project to remove duplicate columns.

### 2.3 Aggregate Functions and Grouping (Preview)

While not part of pure relational algebra (which only produces relations), aggregate functions like `SUM`, `AVG`, `COUNT`, `MAX`, `MIN` are introduced to compute single values from a multiset of values. Grouping (`GROUP BY`) allows partitioning a relation into groups and applying aggregates per group, returning a relation of group values and aggregates.

This will be covered in more detail under SQL.

## 3. SQL: Structured Query Language (Modules 8–10)

SQL is the industry standard for relational databases. It is a **declarative** language: you specify **what** you want, not how to get it. It encompasses DDL (Data Definition Language) to define and modify schemas, and DML (Data Manipulation Language) to query and update data.

### 3.1 History and Standardization

- Originally “Structured English Query Language” (SEQUEL) at IBM.
- Renamed SQL (Structured Query Language).
- Standardized: SQL-86, SQL-89 (integrity constraints), SQL-92 (major international standard), SQL:1999 (regular expressions, recursive queries, triggers, embedding in Java), SQL:2003 (XML), SQL:2006 (full XML), up to SQL:2019.
- Most systems are SQL-92 compliant, many support newer features.

### 3.2 Data Definition (DDL)

#### 3.2.1 Basic Data Types
- `CHAR(n)`: Fixed-length string of n characters.
- `VARCHAR(n)`: Variable-length string up to n characters.
- `INT` / `INTEGER`: Integer (possibly with `SMALLINT`).
- `NUMERIC(p, s)`: Fixed-point number with total p digits, s after decimal.
- `FLOAT(n)`, `REAL`, `DOUBLE PRECISION`: Floating-point.
- `DATE`, `TIME`, `TIMESTAMP`: Date/time types.

#### 3.2.2 Creating Tables
```sql
CREATE TABLE instructor (
    ID         CHAR(5),
    name       VARCHAR(20) NOT NULL,
    dept_name  VARCHAR(20),
    salary     NUMERIC(8,2),
    PRIMARY KEY (ID),
    FOREIGN KEY (dept_name) REFERENCES department(dept_name)
);
```
- `NOT NULL`: constraint disallows nulls.
- `PRIMARY KEY (ID)`: implies `ID` is unique and not null.
- `FOREIGN KEY (dept_name) REFERENCES department(dept_name)`: ensures every instructor's department exists in the `department` table.

#### 3.2.3 Altering and Dropping Tables
- `DROP TABLE instructor;` — remove table and its definition.
- `DELETE FROM instructor;` — delete all rows but keep the table.
- `ALTER TABLE instructor ADD COLUMN start_date DATE;` — adds a new column, initially null for existing rows.
- `ALTER TABLE instructor DROP COLUMN start_date;` — removes a column and its data.

### 3.3 Basic Query Structure: SELECT-FROM-WHERE

```sql
SELECT [DISTINCT] A1, A2, ..., An
FROM R1, R2, ..., Rm
WHERE condition;
```

- **FROM**: Lists the tables used. Implicitly produces the Cartesian product of all tables.
- **WHERE**: Filters rows by a predicate (selection).
- **SELECT**: Specifies the columns to keep (projection). Without `DISTINCT`, duplicates are retained (multiset semantics). `*` means all attributes from all tables.

#### Example: Instructors in the Computer Science department with salary > 80,000.
```sql
SELECT name
FROM instructor
WHERE dept_name = 'Comp. Sci.' AND salary > 80000;
```

### 3.4 Joins in SQL

By listing multiple tables in FROM and using WHERE to specify join conditions, we can perform various joins.

**Equi-join / Natural Join in SQL:**
```sql
SELECT name, course_id
FROM instructor, teaches
WHERE instructor.ID = teaches.ID;
```
Here we match instructor with the courses they teach. The condition `instructor.ID = teaches.ID` is the join predicate.

**Explicit NATURAL JOIN syntax** (if supported):
```sql
SELECT name, course_id
FROM instructor NATURAL JOIN teaches;
```
Automatically joins on common attribute names (here `ID`). It omits the duplicate column.

**Self-join:** Joining a table with itself. Requires renaming with aliases.
```sql
SELECT T.name, S.name
FROM instructor AS T, instructor AS S
WHERE T.salary > S.salary AND S.dept_name = 'Comp. Sci.';
```
List instructors who earn more than **any** instructor in Computer Science. `T` and `S` are two independent “copies” of the instructor table.

**Using AS for renaming:**
```sql
SELECT T.salary / 12 AS monthly_salary
FROM instructor T;
```
Renames a column in the output.

### 3.5 String Operations and Pattern Matching

- `LIKE` operator for pattern matching. Uses:
  - `%` : matches any string (including empty).
  - `_` : matches exactly one character.
- Strings inside quotes are case-sensitive.
- Example: `WHERE name LIKE '%dar%'` matches names containing "dar" (e.g., Darwin, Majumdar).  
  `WHERE name LIKE '___'` matches names exactly three characters long.

### 3.6 Ordering and Limiting Results

- `ORDER BY attribute [ASC|DESC]` — sort output. Default is ascending.
  ```sql
  SELECT DISTINCT name
  FROM instructor
  ORDER BY name DESC;
  ```
- Limiting the number of rows:
  - SQL Server/MS Access: `SELECT TOP 10 name FROM ...`
  - MySQL: `SELECT name FROM ... LIMIT 10`
  - Oracle/Standard: `FETCH FIRST 10 ROWS ONLY`

### 3.7 Additional Predicates

- **BETWEEN**: `WHERE salary BETWEEN 90000 AND 100000` is equivalent to `salary >= 90000 AND salary <= 100000`.
- **Tuple comparison**: `WHERE (instructor.ID, dept_name) = (teaches.ID, 'Biology')`. All corresponding elements are compared.
- **IN**: tests set membership.  
  `WHERE dept_name IN ('Comp. Sci.', 'Biology')` is equivalent to `dept_name = 'Comp. Sci.' OR dept_name = 'Biology'`.

### 3.8 Duplicates and Multiset Semantics

SQL tables and query results are **multisets** (bags) by default: they allow duplicate rows. This differs from relational algebra’s set semantics.

- **Operations on multisets**: When projecting a column, if the same value appears multiple times, each copy is kept. The number of copies in a Cartesian product multiplies, etc.
- To force set semantics (no duplicates), use `SELECT DISTINCT`.
- The `ALL` keyword in set operations (see 3.9) preserves duplicates as per multiset rules.

#### Duplicate counts in multiset operations:
Suppose relation R has tuple t appearing m times, S has t appearing n times.
- `R UNION ALL S`: m + n copies of t.
- `R INTERSECT ALL S`: min(m, n) copies.
- `R EXCEPT ALL S`: max(0, m − n) copies.

If you don’t use `ALL`, duplicates are eliminated.

### 3.9 Set Operations

SQL supports union, intersection, and difference using `UNION`, `INTERSECT`, `EXCEPT` (Oracle uses `MINUS`). The relations must be union-compatible (same number and types of attributes).

- **UNION**: All rows that appear in either. Duplicates removed by default.
- **INTERSECT**: Rows that appear in both.
- **EXCEPT**: Rows in the first but not in the second.

Adding `ALL` retains duplicates:
```sql
(SELECT course_id FROM section WHERE semester='Fall' AND year=2009)
UNION
(SELECT course_id FROM section WHERE semester='Spring' AND year=2010);
```
This gives courses that ran in Fall 2009 **or** Spring 2010.

A creative example using set difference to find the maximum salary (without aggregates):
```sql
(SELECT DISTINCT salary FROM instructor)
EXCEPT
(SELECT DISTINCT T.salary
 FROM instructor AS T, instructor AS S
 WHERE T.salary < S.salary);
```
The inner query finds all salaries that are less than some other salary (i.e., not the maximum). The outer subtracts them from all salaries, leaving only the max.

### 3.10 Null Values and Three-Valued Logic

`NULL` represents unknown or missing information. It is **not** equal to any value, not even another `NULL`. Comparisons involving `NULL` yield `UNKNOWN`, a third truth value.

#### Three-valued logic (3VL):

| p       | q       | p OR q | p AND q |
|---------|---------|--------|---------|
| TRUE    | TRUE    | TRUE   | TRUE    |
| TRUE    | FALSE   | TRUE   | FALSE   |
| TRUE    | UNKNOWN | TRUE   | UNKNOWN |
| FALSE   | TRUE    | TRUE   | FALSE   |
| FALSE   | FALSE   | FALSE  | FALSE   |
| FALSE   | UNKNOWN | UNKNOWN| FALSE   |
| UNKNOWN | TRUE    | TRUE   | UNKNOWN |
| UNKNOWN | FALSE   | UNKNOWN| FALSE   |
| UNKNOWN | UNKNOWN | UNKNOWN| UNKNOWN |

- `NOT UNKNOWN` → `UNKNOWN`.

In a `WHERE` clause, if the condition evaluates to `UNKNOWN`, the row is **not** included (treated as false). To test for null, use `IS NULL` or `IS NOT NULL`.  
`WHERE salary IS NULL` finds rows where salary is unknown.

### 3.11 Aggregate Functions

- `AVG( [DISTINCT] col )`
- `MIN( col )`
- `MAX( col )`
- `SUM( [DISTINCT] col )`
- `COUNT( [DISTINCT] col )` or `COUNT(*)`

Aggregates operate on a multiset of values, ignoring nulls (except `COUNT(*)`, which counts rows). `COUNT(DISTINCT col)` counts distinct non-null values.

#### `GROUP BY` and `HAVING`

- `GROUP BY` partitions the table into groups based on the values of one or more attributes. Aggregates are then applied per group.
  ```sql
  SELECT dept_name, AVG(salary) AS avg_salary
  FROM instructor
  GROUP BY dept_name;
  ```
  This computes the average salary for each department.

- The `SELECT` clause in a grouped query may only contain the grouping attributes and aggregate expressions.
- `HAVING` filters groups after aggregation, much like `WHERE` filters rows before grouping.
  ```sql
  SELECT dept_name, AVG(salary)
  FROM instructor
  GROUP BY dept_name
  HAVING AVG(salary) > 50000;
  ```
  Shows only departments with average salary above 50,000.

- **Nulls in aggregation**: All aggregates except `COUNT(*)` ignore nulls. If all values in a group are null, `COUNT` returns 0; other aggregates return `NULL`.

#### Example using multiple tables with grouping:
```sql
SELECT student.ID, COUNT(course_id) AS num_courses
FROM student LEFT OUTER JOIN takes ON student.ID = takes.ID
GROUP BY student.ID;
```
(Outer join not covered yet, but this would count courses per student.)

## 4. Putting It All Together: A Sample University Database Schema

Recall the schema from the lectures:

- `instructor(ID, name, dept_name, salary)`
- `department(dept_name, building, budget)`
- `student(ID, name, dept_name, tot_cred)`
- `course(course_id, title, dept_name, credits)`
- `teaches(ID, course_id, sec_id, semester, year)`
- `takes(ID, course_id, sec_id, semester, year, grade)`

Typical queries:
- Find all students who have taken a course taught by an instructor in the Comp. Sci. department.
  ```sql
  SELECT DISTINCT S.name
  FROM student S, takes T, teaches TE, instructor I
  WHERE S.ID = T.ID
    AND T.course_id = TE.course_id
    AND T.sec_id = TE.sec_id
    AND T.semester = TE.semester
    AND T.year = TE.year
    AND TE.ID = I.ID
    AND I.dept_name = 'Comp. Sci.';
  ```

This query involves multiple joins. In practice, we would use explicit `JOIN` syntax for readability, but the principle is the same.

## 5. Summary of Key Concepts

- **Relational Model**: Tables (relations) with attributes, tuples, domains; schemas and instances; keys (superkey, candidate key, primary key, foreign key).
- **Relational Algebra**: Procedural query language with operations select, project, union, set difference, Cartesian product, rename, natural join. Closure property allows composition. Not Turing-complete.
- **SQL**: Declarative language for defining schemas (DDL) and querying/manipulating data (DML).
  - Basic query: `SELECT ... FROM ... WHERE ...`
  - Joins via Cartesian product + selection, natural join syntax, self-joins with aliases.
  - String matching with `LIKE`, ordering with `ORDER BY`, limiting rows with `TOP`/`LIMIT`/`FETCH`.
  - Set operations `UNION`, `INTERSECT`, `EXCEPT`, and multiset variants with `ALL`.
  - Nulls introduce three-valued logic; tested with `IS NULL`.
  - Aggregates: `COUNT`, `SUM`, `AVG`, `MIN`, `MAX`; grouping with `GROUP BY`; filter groups with `HAVING`.

Everything covered here lays the foundation for advanced topics such as subqueries, views, transactions, and database design (normalization). The key is to practice writing queries and thinking in sets and multisets.

## 1. SQL Examples – A Thorough Recap with Practice (Module 11)

The purpose of this module is to solidify your understanding of the basic `SELECT-FROM-WHERE` structure, string matching, ordering, set operations, and aggregation by walking through concrete examples using the university database. The schema used throughout is:

- **classroom**(*building*, *room_number*, *capacity*)
- **department**(*dept_name*, *building*, *budget*)
- **course**(*course_id*, *title*, *dept_name*, *credits*)
- **instructor**(*ID*, *name*, *dept_name*, *salary*)
- **section**(*course_id*, *sec_id*, *semester*, *year*, *building*, *room_number*, *time_slot_id*)
- **teaches**(*ID*, *course_id*, *sec_id*, *semester*, *year*)
- **student**(*ID*, *name*, *dept_name*, *tot_cred*)
- **takes**(*ID*, *course_id*, *sec_id*, *semester*, *year*, *grade*)

### 1.1 Simple Selection and Projection

**Problem**: Find the names of all buildings that contain at least one classroom with a capacity of less than 100. Remove duplicates.

```sql
SELECT DISTINCT building
FROM classroom
WHERE capacity < 100;
```

- The `FROM` clause indicates we work on the `classroom` table.
- The `WHERE` clause filters rows: only those where the capacity is strictly less than 100.
- The `SELECT` clause picks the `building` column.
- `DISTINCT` eliminates duplicate building names (because the same building may have several small classrooms).
- Without `DISTINCT`, the result would be a multiset; for instance, “Watson” would appear twice if it has two such classrooms.

### 1.2 Cartesian Product with Condition (Implicit Join)

**Problem**: List all students and the budget of their department, but only for departments whose budget is less than $100,000.

```sql
SELECT name, budget
FROM student, department
WHERE student.dept_name = department.dept_name
  AND budget < 100000;
```

- Two tables in `FROM` produce a Cartesian product: every student combined with every department.
- The `WHERE` condition keeps only pairs where the student's department name matches the department’s name (i.e., a valid affiliation).
- The extra condition `budget < 100000` further filters.
- The output lists the student’s name and the department’s budget.

### 1.3 Using Aliases (Renaming)

```sql
SELECT S.name AS student_name, D.budget AS dept_budget
FROM student AS S, department AS D
WHERE S.dept_name = D.dept_name;
```

- `student AS S` gives the table an alias `S`, making it easier to reference.
- Column aliases (`student_name`, `dept_budget`) rename the output headers.
- The `AS` keyword is optional; `student S` works too.

### 1.4 Complex WHERE Predicates with AND/OR and `IN`

**Problem**: Find the names of all instructors who work in the Finance department or whose department is located in the Watson or Taylor buildings.

```sql
SELECT DISTINCT I.name
FROM instructor AS I, department AS D
WHERE I.dept_name = D.dept_name
  AND (I.dept_name = 'Finance' OR D.building IN ('Watson', 'Taylor'));
```

- The join condition (`I.dept_name = D.dept_name`) is mandatory for linking.
- Parentheses ensure correct evaluation: the join condition must hold, **and** either the instructor's department is Finance **or** the building is Watson/Taylor.
- `IN` provides a concise way to check set membership.

### 1.5 String Pattern Matching with `LIKE`

**Problem**: Find the titles of courses whose `course_id` contains exactly three letters before a hyphen (like “BIO-301”).

```sql
SELECT title
FROM course
WHERE course_id LIKE '___-%';
```

- `_` matches a single character; three underscores mean exactly three characters.
- `-` matches the hyphen literally.
- `%` matches any sequence of characters (including empty), so the remainder of the string is ignored.
- The pattern `'%dar%'` would match any course_id containing “dar” anywhere.

### 1.6 Ordering Results

**Problem**: List all students, sorted by department name ascending, and within each department by total credits descending.

```sql
SELECT name, dept_name, tot_cred
FROM student
ORDER BY dept_name ASC, tot_cred DESC;
```

- `ORDER BY` can specify multiple columns.
- Default order is ascending (`ASC`); `DESC` must be stated explicitly.

### 1.7 Set Operations: UNION, INTERSECT, EXCEPT

**UNION**: Courses offered in Fall 2018 **or** Spring 2019 (removing duplicates):

```sql
(SELECT course_id FROM section WHERE semester='Fall' AND year=2018)
UNION
(SELECT course_id FROM section WHERE semester='Spring' AND year=2019);
```

- Both queries must have the same number and types of columns.
- `UNION` eliminates duplicates; use `UNION ALL` to keep them.

**INTERSECT**: Instructors in Comp. Sci. **and** salary < 80000 (using a set membership approach):

```sql
(SELECT ID FROM instructor WHERE dept_name IN ('Comp. Sci.', 'Finance'))
INTERSECT
(SELECT ID FROM instructor WHERE salary < 80000);
```

Alternatively, with a subquery:
```sql
SELECT ID FROM instructor
WHERE dept_name IN ('Comp. Sci.', 'Finance')
  AND salary < 80000;
```

**EXCEPT**: Instructors in Comp. Sci. or Finance **but not** with salary between 70000 and 90000.

```sql
(SELECT ID FROM instructor WHERE dept_name IN ('Comp. Sci.', 'Finance'))
EXCEPT
(SELECT ID FROM instructor WHERE salary BETWEEN 70000 AND 90000);
```

### 1.8 Aggregation and Grouping

**Average capacity per building, only for buildings with average capacity > 25**:

```sql
SELECT building, AVG(capacity) AS avg_capacity
FROM classroom
GROUP BY building
HAVING AVG(capacity) > 25;
```

- `GROUP BY building` partitions rows by building.
- `AVG(capacity)` computes the average within each group.
- `HAVING` filters the groups after aggregation. (Note: `WHERE` would filter before grouping.)

**Total credits offered by each department**:

```sql
SELECT dept_name, SUM(credits) AS total_credits
FROM course
GROUP BY dept_name;
```

**Count courses per building**:

```sql
SELECT building, COUNT(DISTINCT course_id) AS num_courses
FROM section
GROUP BY building;
```

`COUNT(DISTINCT ...)` avoids counting duplicate course_ids in the same building.

## 2. Intermediate SQL – Subqueries and Data Modification (Module 12)

### 2.1 Nested Subqueries

A subquery is a `SELECT` statement embedded inside another query. It can appear in:
- `WHERE` clause (most common),
- `FROM` clause (as a derived table),
- `SELECT` clause (must return a single value – a scalar subquery).

#### 2.1.1 Subqueries in `WHERE` with Set Membership

**Find courses offered in Fall 2009 and also in Spring 2010** (intersection via `IN`):

```sql
SELECT DISTINCT course_id
FROM section
WHERE semester = 'Fall' AND year = 2009
  AND course_id IN (
      SELECT course_id
      FROM section
      WHERE semester = 'Spring' AND year = 2010
  );
```

- The inner query returns all courses from Spring 2010.
- The outer query checks for each Fall 2009 course whether its `course_id` is in that set.

**“Not in”** for set difference: courses in Fall 2009 but not in Spring 2010.

```sql
SELECT DISTINCT course_id
FROM section
WHERE semester = 'Fall' AND year = 2009
  AND course_id NOT IN (
      SELECT course_id
      FROM section
      WHERE semester = 'Spring' AND year = 2010
  );
```

#### 2.1.2 Some (Existential) and All (Universal) Quantifiers

**Instructors whose salary is greater than that of at least one instructor in Biology**:

```sql
SELECT DISTINCT T.name
FROM instructor AS T
WHERE T.salary > SOME (
      SELECT S.salary
      FROM instructor AS S
      WHERE S.dept_name = 'Biology'
  );
```

- `> SOME (subquery)` is true if `T.salary` is greater than at least one value in the subquery result.
- Equivalently: `T.salary > ANY (...)`.

**Instructors whose salary is greater than that of all instructors in Biology**:

```sql
SELECT name
FROM instructor
WHERE salary > ALL (
      SELECT salary
      FROM instructor
      WHERE dept_name = 'Biology'
  );
```

- `> ALL` requires that the condition holds for every value in the subquery result.
- This is useful for finding the highest salary relative to a group.

Formal logic:
- `value > SOME (subquery)` ⇔ ∃ t ∈ subquery such that value > t.
- `value > ALL (subquery)` ⇔ ∀ t ∈ subquery, value > t.

#### 2.1.3 Existence Tests: `EXISTS` and `NOT EXISTS`

**Courses taught in both Fall 2009 and Spring 2010 using EXISTS**:

```sql
SELECT S.course_id
FROM section AS S
WHERE S.semester = 'Fall' AND S.year = 2009
  AND EXISTS (
      SELECT *
      FROM section AS T
      WHERE T.semester = 'Spring' AND T.year = 2010
        AND T.course_id = S.course_id
  );
```

- The inner correlated subquery references the outer table `S`.
- `EXISTS` returns true if the subquery returns at least one row.
- This essentially performs a semi-join.

**`NOT EXISTS`** is used to check emptiness, often to express “for all” conditions. For example, “students who have taken all courses” is classically formulated with a double `NOT EXISTS`.

#### 2.1.4 `UNIQUE` Test

`UNIQUE (subquery)` evaluates to true if the subquery contains no duplicate tuples. This is rarely used but available.

#### 2.1.5 Subqueries in the `FROM` Clause (Derived Tables)

**Average salary of departments where the average salary exceeds 42,000**:

```sql
SELECT dept_name, avg_salary
FROM (
      SELECT dept_name, AVG(salary) AS avg_salary
      FROM instructor
      GROUP BY dept_name
) AS dept_avg
WHERE avg_salary > 42000;
```

- The inner query produces a table with `dept_name` and `avg_salary`.
- The outer query treats this result as a temporary table `dept_avg` and filters it.
- This avoids `HAVING`, though `HAVING` would also work.

#### 2.1.6 `WITH` Clause (Common Table Expressions)

The `WITH` clause creates a temporary named relation that can be used in the main query. It improves readability, especially for complex queries.

**Departments with maximum budget**:

```sql
WITH max_budget(value) AS (
      SELECT MAX(budget)
      FROM department
)
SELECT dept_name
FROM department, max_budget
WHERE department.budget = max_budget.value;
```

- `max_budget` is defined as a one-attribute, one-row table containing the maximum budget.
- The main query then compares each department’s budget to that value.
- This is equivalent to a subquery but more structured.

#### 2.1.7 Scalar Subqueries in `SELECT`

Subqueries in the `SELECT` clause must return a single value (one row, one column). They can be used to add computed columns.

**List all departments with the number of instructors in each**:

```sql
SELECT dept_name,
       (SELECT COUNT(*)
        FROM instructor
        WHERE instructor.dept_name = department.dept_name) AS num_instructors
FROM department;
```

- The subquery is correlated: for each department, it counts matching instructors.
- The result is a scalar per department.

### 2.2 Data Modification Statements

#### 2.2.1 DELETE

- Delete all instructors of the Finance department:
  ```sql
  DELETE FROM instructor
  WHERE dept_name = 'Finance';
  ```

- Delete instructors whose department is in the Watson building (using subquery):
  ```sql
  DELETE FROM instructor
  WHERE dept_name IN (
        SELECT dept_name
        FROM department
        WHERE building = 'Watson'
  );
  ```

**Important caveat**: When deleting rows based on a condition that depends on the same table being modified (e.g., “delete instructors whose salary is less than the average salary of instructors”), you must be careful. The average changes as rows are deleted, leading to unpredictable behavior. The solution is to compute the threshold first (e.g., using a temporary table or a `WITH` clause), then delete using that static value.

#### 2.2.2 INSERT

- Insert a single course:
  ```sql
  INSERT INTO course VALUES ('CS-437', 'Database Systems', 'Comp. Sci.', 4);
  ```
  (Values must follow the order of columns as defined in the schema.)

- Alternatively, specify columns:
  ```sql
  INSERT INTO course (course_id, title, dept_name, credits)
  VALUES ('CS-437', 'Database Systems', 'Comp. Sci.', 4);
  ```

- Insert with `NULL`:
  ```sql
  INSERT INTO student VALUES ('12345', 'John', 'Physics', NULL);
  ```

- Insert the result of a query (e.g., insert all instructors into a `student` table with initial total credits 0):
  ```sql
  INSERT INTO student (ID, name, dept_name, tot_cred)
  SELECT ID, name, dept_name, 0
  FROM instructor;
  ```
  The subquery is fully evaluated before any insertion, so it is safe.

#### 2.2.3 UPDATE

- Give a 3% raise to instructors with salary > 100,000:
  ```sql
  UPDATE instructor
  SET salary = salary * 1.03
  WHERE salary > 100000;
  ```

- Give a 5% raise to those with salary ≤ 100,000:
  ```sql
  UPDATE instructor
  SET salary = salary * 1.05
  WHERE salary <= 100000;
  ```

**Problem**: If you execute two separate `UPDATE` statements, the first may change salaries, causing the second to apply incorrectly. To handle both cases atomically and correctly, use the `CASE` expression:

```sql
UPDATE instructor
SET salary = CASE
                WHEN salary > 100000 THEN salary * 1.03
                ELSE salary * 1.05
             END;
```

This ensures each salary is updated exactly once, based on the original value.

#### 2.2.4 Update with Scalar Subquery

You can set a column to the result of a scalar subquery. Example: set the budget of each department to the sum of salaries of its instructors (hypothetical). The subquery must be correlated and return a single value.

## 3. Intermediate SQL – Joins and Views (Module 13)

### 3.1 Join Expressions

Joins combine rows from two or more tables based on a related condition. While the implicit join (Cartesian product with WHERE) works, explicit `JOIN` syntax is often clearer and offers additional types.

Given two example tables:
- **course**(*course_id*, *title*, *dept_name*, *credits*)
- **prereq**(*course_id*, *prereq_id*)

We’ll explore joins where some courses have no prerequisites and some prerequisites refer to non-existing courses.

#### 3.1.1 Cross Join

Cartesian product of two tables:

```sql
SELECT *
FROM course CROSS JOIN prereq;
```

Equivalent to `FROM course, prereq`. Yields `|course| * |prereq|` rows.

#### 3.1.2 Inner Join

Only rows that have matching values in both tables are retained. The most common is the equi-join on `course_id`:

```sql
SELECT *
FROM course INNER JOIN prereq ON course.course_id = prereq.course_id;
```

- Rows from `course` without a matching `prereq` (like CS-315) and rows from `prereq` without a matching `course` (like CS-347) are **excluded**.
- The result contains only the intersection of the two tables on the join attributes.

If we add the `NATURAL` keyword (i.e., `NATURAL INNER JOIN`), the join is performed automatically on all common-named columns, and duplicate columns are eliminated from the output.

#### 3.1.3 Outer Joins

Outer joins preserve tuples that do not have a match, filling missing values with `NULL`.

- **LEFT OUTER JOIN**: preserves every row from the **left** table.
  ```sql
  SELECT *
  FROM course NATURAL LEFT OUTER JOIN prereq;
  ```
  Rows like CS-315 appear with `prereq_id = NULL`.

- **RIGHT OUTER JOIN**: preserves every row from the **right** table.
  ```sql
  SELECT *
  FROM course NATURAL RIGHT OUTER JOIN prereq;
  ```
  CS-347 appears, but its course details are `NULL`.

- **FULL OUTER JOIN**: preserves rows from **both** tables.
  ```sql
  SELECT *
  FROM course NATURAL FULL OUTER JOIN prereq;
  ```
  All courses and all prerequisites are shown, with `NULL` where information is missing.

#### 3.1.4 Join Conditions: ON, USING, NATURAL

- `ON` specifies an arbitrary join predicate (e.g., `ON course.course_id = prereq.course_id AND ...`). This is the most flexible.
- `USING (column_list)` requires that the listed columns exist in both tables and only those columns are used for equality. It also removes duplicate columns from the result (like `NATURAL`).
- `NATURAL` implicitly uses all common-named columns.

Example with `USING`:
```sql
SELECT *
FROM course INNER JOIN prereq USING (course_id);
```
Only one `course_id` column appears in the result.

### 3.2 Views

A view is a **virtual relation** defined by a query. It does not store data physically; instead, its definition is saved in the data dictionary, and every time the view is queried, its defining query is executed.

#### 3.2.1 Creating and Using Views

**Create a view that hides the salary column**:
```sql
CREATE VIEW faculty AS
SELECT ID, name, dept_name
FROM instructor;
```

Now users can query `faculty` as if it were a table:
```sql
SELECT name
FROM faculty
WHERE dept_name = 'Biology';
```
The DBMS internally replaces the reference to `faculty` with its defining query, combining it with the outer query (view expansion).

Views can contain computed columns:
```sql
CREATE VIEW department_total_salary AS
SELECT dept_name, SUM(salary) AS total_salary
FROM instructor
GROUP BY dept_name;
```

#### 3.2.2 View Expansion and Dependencies

When a view is used, it is expanded recursively until only base tables remain. For example, `physics_fall_2009` might be defined on `course` and `section`. If we create `physics_fall_2009_Watson` that references the first view, the final expansion embeds both queries.

A view **directly depends** on another view if the second is used in its definition. Recursive views (a view that depends on itself) are not allowed in standard SQL unless explicitly supported (e.g., `WITH RECURSIVE`), which is an advanced topic not covered here.

#### 3.2.3 Updating Views

Updates on views (INSERT, UPDATE, DELETE) must be translated into updates on the underlying base tables. This is only possible under restricted conditions, typically when the view is defined from a single table and does not involve aggregation or joins.

- Insert into `faculty` (simple view) translates to an insert into `instructor`, with `NULL` for the missing `salary` column (if nullable).
- Insert into a view that joins two tables (like `instructor_info` that shows instructor name and building) is generally not allowed because it’s ambiguous: which building corresponds to which department?

Also, inserting a tuple that violates the view’s defining predicate (e.g., inserting a Biology instructor into a view that only shows History instructors) would make the tuple invisible in the view, which is usually prevented.

**Materialized views**: These are views that actually store the result set physically. They improve performance but need to be refreshed when base tables change.

## 4. Intermediate SQL – Transactions, Integrity, Data Types, Authorization (Module 14)

### 4.1 Transactions

A transaction is a sequence of database operations that form a single logical unit of work. It must be:

- **Atomic**: all or nothing.
- **Consistent**: leave the database in a valid state.
- **Isolated**: concurrent transactions do not interfere.
- **Durable**: committed changes survive failures.

SQL defines the boundaries of a transaction:
- Start is implicit (first SQL statement after a connection or after a commit/rollback).
- End with `COMMIT` (make changes permanent) or `ROLLBACK` (undo all changes since start).

Most databases have auto-commit mode by default (each statement is its own transaction). You can disable it to explicitly group statements.

Example:
```sql
BEGIN;
UPDATE account SET balance = balance - 100 WHERE account_id = 'A';
UPDATE account SET balance = balance + 100 WHERE account_id = 'B';
COMMIT;
```
If a failure occurs before `COMMIT`, the `ROLLBACK` can be executed to revert.

Transactions will be studied in great depth later in the course.

### 4.2 Integrity Constraints

Constraints ensure the data obeys business rules and remains consistent.

#### 4.2.1 NOT NULL
An attribute declared `NOT NULL` cannot be left empty.
```sql
name VARCHAR(20) NOT NULL
```

#### 4.2.2 UNIQUE
The combination of attributes must be distinct across all tuples. Unlike the primary key, a unique attribute may contain `NULL` (except when also `NOT NULL`).

```sql
UNIQUE (passport_no)
```

#### 4.2.3 CHECK
Ensures a column value satisfies a predicate. Example:
```sql
semester VARCHAR(6) CHECK (semester IN ('Fall', 'Winter', 'Spring', 'Summer'))
```
This forbids any other value.

#### 4.2.4 Referential Integrity (Foreign Key)

A foreign key column references the primary key of another table, ensuring related rows exist.

```sql
FOREIGN KEY (dept_name) REFERENCES department(dept_name)
```

When the referenced row is deleted or updated, actions can be specified:
- `ON DELETE CASCADE`: automatically delete the dependent rows.
- `ON UPDATE CASCADE`: propagate the updated value.
- `ON DELETE SET NULL`: set the foreign key to `NULL` (if allowed).
- `ON DELETE SET DEFAULT`: set to default value.
- `ON DELETE RESTRICT` / `NO ACTION`: prevent the operation if dependents exist.

Example:
```sql
FOREIGN KEY (dept_name) REFERENCES department(dept_name)
    ON DELETE CASCADE
    ON UPDATE CASCADE
```

#### 4.2.5 Deferred Constraints

Sometimes a constraint cannot be satisfied at the moment of a single statement but will be after the transaction. For example, inserting a person and their parents (who are also persons) with mutual foreign keys. The constraint can be deferred until commit time, but this is an advanced feature.

### 4.3 Additional Data Types

#### 4.3.1 Date/Time Types

- `DATE`: stores year, month, day.
- `TIME`: stores hour, minute, second.
- `TIMESTAMP`: both date and time.
- `INTERVAL`: a duration, e.g., `INTERVAL '1' DAY`, can be added to dates.

These support many functions (extracting year, formatting, etc.).

#### 4.3.2 Indexes

An index is a separate data structure (e.g., B-tree) that speeds up data retrieval. Creating an index on a primary key or a frequently searched column can drastically improve performance.

```sql
CREATE INDEX ID_index ON student(ID);
```

Indices are transparent to SQL queries; the optimizer decides when to use them. We'll study them in detail later.

#### 4.3.3 User-Defined Types (UDTs) and Domains

- **UDT**: a simple alias for an existing type, e.g.:
  ```sql
  CREATE TYPE Dollars AS NUMERIC(12,2);
  ```
  Then use `Dollars` in column definitions.

- **Domain**: an extended UDT that can carry constraints.
  ```sql
  CREATE DOMAIN PersonName AS VARCHAR(20) NOT NULL;
  CREATE DOMAIN DegreeLevel AS VARCHAR(10)
      CHECK (VALUE IN ('Bachelors', 'Masters', 'PhD'));
  ```
  Domains promote consistency and reusability across tables.

#### 4.3.4 Large Object Types

- `BLOB` (Binary Large Object): for images, videos, etc.
- `CLOB` (Character Large Object): for large text.

These are typically stored outside the main database files, with a pointer in the row.

### 4.4 Authorization (Security)

Authorization controls what users can do with the database. Privileges are granted and revoked.

#### 4.4.1 Common Privileges

- `SELECT` (read)
- `INSERT`
- `UPDATE`
- `DELETE`
- `REFERENCES` (permission to create a foreign key referencing the table)
- `ALL PRIVILEGES`

#### 4.4.2 Granting Privileges

```sql
GRANT SELECT ON instructor TO Amit;
```

To allow the grantee to further grant the privilege to others:
```sql
GRANT SELECT ON department TO Amit WITH GRANT OPTION;
```

Grant to a role:
```sql
GRANT SELECT ON takes TO instructor_role;
```

#### 4.4.3 Revoking Privileges

```sql
REVOKE SELECT ON branch FROM user1, user2;
```

If a user received the privilege through `WITH GRANT OPTION`, revoking may cascade.

#### 4.4.4 Roles

A role is a named collection of privileges. Users are assigned roles, inheriting their privileges. This simplifies management.

```sql
CREATE ROLE dean;
GRANT instructor TO dean;
GRANT dean TO Satoshi;
```

Now Satoshi has all privileges of `instructor` plus any granted directly to `dean`.

#### 4.4.5 Views and Authorization

Views can be used to grant access to only specific rows or columns. For example, create a view `geo_instructor` that shows only instructors from the Geography department, and grant `SELECT` on that view to a user who has no direct access to the `instructor` table. This provides content-based access control.

## 5. Advanced SQL – Functions, Procedures, and Triggers (Module 15)

SQL, primarily declarative, has been extended with procedural constructs (since SQL:1999) to write functions, procedures, and triggers. This allows encapsulation of logic inside the database.

### 5.1 Functions and Procedures

#### 5.1.1 SQL Functions

A function returns a single value or a table.

**Example**: Function that counts the number of instructors in a given department.

```sql
CREATE FUNCTION dept_count(dept_name VARCHAR(20))
RETURNS INTEGER
BEGIN
    DECLARE d_count INTEGER;
    SELECT COUNT(*) INTO d_count
    FROM instructor
    WHERE instructor.dept_name = dept_name;
    RETURN d_count;
END;
```

Usage:
```sql
SELECT dept_name, budget
FROM department
WHERE dept_count(dept_name) > 12;
```

- `DECLARE` defines local variables.
- `INTO` stores the result of a scalar query into a variable.
- `RETURN` returns the value.

#### 5.1.2 Table Functions

Functions can return entire tables. They are called with `TABLE(...)`.

```sql
CREATE FUNCTION instructor_of(dept_name VARCHAR(20))
RETURNS TABLE (
    ID VARCHAR(5),
    name VARCHAR(20),
    salary NUMERIC(8,2)
)
BEGIN
    RETURN TABLE (
        SELECT ID, name, salary
        FROM instructor
        WHERE instructor.dept_name = dept_name
    );
END;
```

Usage:
```sql
SELECT *
FROM TABLE(instructor_of('Music'));
```

#### 5.1.3 SQL Procedures

Procedures do not return a value; they can have `OUT` parameters.

```sql
CREATE PROCEDURE count_instructors(
    IN  dept_name VARCHAR(20),
    OUT d_count  INTEGER
)
BEGIN
    SELECT COUNT(*) INTO d_count
    FROM instructor
    WHERE instructor.dept_name = dept_name;
END;
```

Call:
```sql
DECLARE cnt INTEGER;
CALL count_instructors('Physics', cnt);
```

Procedures are invoked with `CALL`; they are not part of an expression.

#### 5.1.4 Procedural Language Constructs

SQL provides typical control-flow statements inside `BEGIN ... END` blocks:

- **Compound statement**: `BEGIN ... END`
- **While loop**:
  ```sql
  WHILE condition DO
      statement;
  END WHILE;
  ```
- **Repeat loop**:
  ```sql
  REPEAT
      statement;
  UNTIL condition
  END REPEAT;
  ```
- **For loop** over query result:
  ```sql
  FOR r AS (SELECT budget FROM department) DO
      SET n = n + r.budget;
  END FOR;
  ```
- **If-then-else**:
  ```sql
  IF condition THEN statement;
  ELSEIF condition THEN statement;
  ELSE statement;
  END IF;
  ```
- **Case statement** (two forms):
  - Simple case:
    ```sql
    CASE n
        WHEN 1 THEN ...;
        WHEN 2 THEN ...;
    END CASE;
    ```
  - Searched case:
    ```sql
    CASE
        WHEN n > 0 THEN ...;
        ELSE ...;
    END CASE;
    ```
- **Exception handling**: `SIGNAL` and `HANDLER` for catching errors.

These constructs make SQL/PSM a full procedural environment, though with database-oriented features.

#### 5.1.5 External Language Routines

Functions and procedures can be written in external languages like C, Java, Python. This is useful for complex logic, specialized data types, or performance.

```sql
CREATE PROCEDURE my_proc(IN arg INT)
LANGUAGE C
EXTERNAL NAME '/path/to/library!my_func';
```

- The external routine must be compiled into a shared library.
- Security risks: malicious code can corrupt data. So, safe languages (Java, C#) are preferred, or the DBMS runs them in a sandbox.

### 5.2 Triggers

A trigger is a piece of code that automatically executes in response to certain events on a table (INSERT, UPDATE, DELETE). Triggers enforce complex integrity constraints, logging, or cascading actions.

#### 5.2.1 Components of a Trigger

- **Event**: `INSERT`, `UPDATE`, or `DELETE` on a table.
- **Timing**: `BEFORE` or `AFTER` the event.
- **Granularity**: `FOR EACH ROW` or `FOR EACH STATEMENT`.
- **Referencing**: access old/new values (for row-level triggers) or old/new tables (for statement-level).
- **Action**: procedural SQL block.

#### 5.2.2 Row-Level Trigger Example

**Before updating a grade: if the new grade is blank, set it to NULL.**

```sql
CREATE TRIGGER setnull_grade
BEFORE UPDATE OF grade ON takes
REFERENCING NEW ROW AS nrow
FOR EACH ROW
BEGIN
    IF nrow.grade = ' ' THEN
        SET nrow.grade = NULL;
    END IF;
END;
```

The trigger fires before each row update, checks the new grade, and modifies it if necessary. The action is atomic.

**After insert on `takes`, recompute the student’s total credits** (based on passed courses). This would be a more complex trigger with queries on the `course` table.

#### 5.2.3 Statement-Level Triggers and Transition Tables

A statement trigger fires once per SQL statement, even if multiple rows are affected. It can reference `OLD TABLE` and `NEW TABLE` to see all affected rows.

#### 5.2.4 Best Practices for Triggers

**When to use:**
- Logging changes to a history table.
- Auditing user actions.
- Maintaining derived values (e.g., total credits).
- Enforcing complex cross-table constraints that cannot be expressed with standard CHECK or foreign keys.

**When NOT to use:**
- Too many triggers can create maintenance nightmares and performance issues.
- Avoid complex logic inside triggers; keep them simple.
- Do not reference cross-server databases.
- Avoid recursive triggers (a trigger firing another trigger that fires the original again) – most DBMSs disable them by default.
- Avoid putting expensive operations (aggregations, joins on large tables) inside row-level triggers; they can severely slow down DML operations.
- Do not call functions/procedures that themselves issue DML on the same table (risk of infinite loop).

The golden rule: triggers should be lightweight, well-documented, and used sparingly. They are powerful tools for ensuring data integrity, but misuse can lead to unmaintainable and unpredictable databases.

## 1. Formal Relational Query Languages – Relational Algebra (Module 16)

Relational algebra is a **procedural** query language: you specify *how* to obtain the answer by applying a sequence of operators to relations. Each operator takes one or two relations and produces a new relation. This closure property permits nesting of arbitrary depth.

### 1.1 Basic Operators (Recap with Formal Set‑Theoretic Definitions)

Six fundamental operators form the core of relational algebra.

#### 1.1.1 Select (`σ`)

Chooses tuples that satisfy a predicate `p`.

Formally:  
`σ_p(r) = { t | t ∈ r and p(t) }`  

`p` is a Boolean formula built from comparisons (`=`, `<`, `>`, `≤`, `≥`, `≠`), constants, and logical connectives (`∧`, `∨`, `¬`).

**Example:** `σ_{salary > 80000 ∧ dept_name='Physics'}(instructor)`

#### 1.1.2 Project (`π`)

Returns a relation with only the specified attributes, eliminating duplicate tuples (set semantics).

`π_{A1, A2, …, Ak}(r)`

Formally: `π_{A1,…,Ak}(r) = { t[A1,…,Ak] | t ∈ r }`  
The result is a set of distinct subtuples.

#### 1.1.3 Union (`∪`)

`r ∪ s = { t | t ∈ r or t ∈ s }`  

Requirements: `r` and `s` must have the same **arity** (number of attributes) and the corresponding attributes must have compatible domains.

#### 1.1.4 Set Difference (`−`)

`r − s = { t | t ∈ r and t ∉ s }`  

Same compatibility requirements as union.

#### 1.1.5 Cartesian Product (`×`)

`r × s = { (t_r , t_s) | t_r ∈ r and t_s ∈ s }`  

Attributes of `r` and `s` must be disjoint (if not, use renaming). The resulting schema is the concatenation of the two schemas.

#### 1.1.6 Rename (`ρ`)

Allows us to change the name of a relation or its attributes, ensuring unambiguous names.

`ρ_{x(A1, A2, …, An)}(E)`  

Renames the result of expression `E` to relation `x` with attributes `A1,…,An`. This is essential for self‑joins and for disambiguating attributes before Cartesian product.

### 1.2 Derived Operators: Join and Division

Additional operators can be expressed in terms of the basic six. The most important ones are the various **join** operations and the **division** operator.

#### 1.2.1 Theta Join (`⋈_θ`)

`r ⋈_θ s = σ_θ (r × s)`  

The condition `θ` is a predicate. If `θ` is equality on common attribute names, and the result automatically retains only one copy of each common attribute, it becomes the **natural join** (`r ⋈ s`).

- **Equi‑join**: a theta join where `θ` uses only equality comparisons.
- **Natural join**: implicit equality on all common attributes, duplicate columns removed.

#### 1.2.2 Division Operator (`÷`)

Division is a binary operation that answers queries involving “for all”. Given relations `r(Z)` and `s(X)` where `X ⊂ Z`, the result `r ÷ s` contains the values of the remaining attributes `Y = Z − X` that are paired with **every** tuple of `s` in `r`.

Formal definition:  
`r ÷ s = { t | t ∈ π_Y(r) and for every tuple t_s ∈ s, there exists a tuple t_r ∈ r such that t_r[Y] = t[Y] and t_r[X] = t_s[X] }`

An equivalent expression using basic operators:  
`r ÷ s = π_Y(r) − π_Y( (π_Y(r) × s) − r )`  

This expression first builds all possible combinations of `Y`‑values with every `s` tuple, subtracts those that actually exist in `r`, and then the remaining `Y`‑values (those that lack at least one `s` tuple) are removed from all `Y` values.

**Intuitive example:**  
Relation `takes(course_id, student_id)` and relation `all_courses(course_id)`.  
`takes ÷ all_courses` yields the students who have taken **all** courses.

**Detailed example with a small table:**  

Let `r(A,B)` = {(α,1), (α,2), (β,1), (γ,1), (γ,2)} and `s(B)` = {(1),(2)}.  
We want `r ÷ s`, i.e., `A`‑values that are associated with both 1 and 2 in `r`.

- All possible `A` values: {α, β, γ}
- For α: pairs (α,1) and (α,2) both exist → α qualifies.
- For β: only (β,1) exists, missing (β,2) → β fails.
- For γ: both (γ,1) and (γ,2) exist → γ qualifies.

Result: {α, γ}.

If `s` had only (1), then `r ÷ s` would be {α, β, γ}. Division truly captures “for all” semantics.

**Application example:** “Find customers who have an account in every branch of the bank.”  
`account(cust_name, branch_name) ÷ branch(branch_name)` gives exactly those customers.

### 1.3 Formal Summary

Every relational algebra expression is built from relation names using the operators. The result is always a relation – no duplicates, no ordering. The algebra is not Turing‑complete, but it is the formal underpinning of SQL (which adds multiset semantics, aggregation, and more).

## 2. Tuple Relational Calculus & Domain Relational Calculus (Module 17)

While relational algebra is procedural, the relational calculi are **declarative**: they describe *what* the result should contain, not how to compute it. They are based on **predicate calculus** (first‑order logic).

### 2.1 Crash Course in Predicate Logic

- **Proposition**: A statement that is either true or false (e.g., `5 > 3`).
- **Predicate**: A statement containing variables; it becomes a proposition when the variables are assigned specific values. Written as `P(x)`. Example: `P(x) ≡ "x > 3"`. `P(5)` is true, `P(2)` is false.
- **Quantifiers**: Extend predicates to make statements about sets.
  - **Universal quantifier** `∀x P(x)`: “For all x in the domain, `P(x)` is true.”
  - **Existential quantifier** `∃x P(x)`: “There exists at least one x in the domain such that `P(x)` is true.”
- The domain of discourse (universe) must be specified or understood.

### 2.2 Tuple Relational Calculus (TRC)

A TRC query has the form:  
`{ t | P(t) }`  
where `t` is a **tuple variable** and `P` is a formula built from:

- **Atom**: `R(t)` – `t` is a tuple of relation `R`.
- **Atom**: `t.A op s.B` or `t.A op constant` – comparison between attributes of tuples or with constants.
- **Logical connectives**: `∧`, `∨`, `¬`
- **Quantification**: `∃ t (P(t))`, `∀ t (P(t))`

#### 2.2.1 Examples with University Schema

**1. Find the first names of students older than 21.**  
`{ t.Fname | ∃ s ∈ student ( s.age > 21 ∧ t.Fname = s.Fname ) }`  
or simply:  
`{ s.Fname | s ∈ student ∧ s.age > 21 }`

**2. Names of students who have taken ‘DBMS’.**  
`{ s.name | s ∈ student ∧ ∃ t ∈ takes ( t.ID = s.ID ∧ ∃ c ∈ course ( c.course_id = t.course_id ∧ c.title = 'DBMS' ) ) }`

**3. Pilots (employee IDs) certified for Boeing aircraft.**  
Relations: `certified(eid, aid)`, `aircraft(aid, aname, …)`, `employee(eid, …)`.  
`{ c.eid | c ∈ certified ∧ ∃ a ∈ aircraft ( a.aid = c.aid ∧ a.aname = 'Boeing' ) }`

**4. Find students who have taken all courses taught by instructor ‘Einstein’.**  
`{ s.name | s ∈ student ∧ ∀ c ∈ course ( ∃ t ∈ teaches ( t.ID = 'Einstein' ∧ t.course_id = c.course_id ) → ∃ tk ∈ takes ( tk.ID = s.ID ∧ tk.course_id = c.course_id ) ) }`  
The implication (`→`) is used to express “for all courses that Einstein teaches, the student has taken them”.

### 2.3 Domain Relational Calculus (DRC)

Instead of tuple variables, DRC uses **domain variables** that range over individual attribute values. An expression looks like:  
`{ <x1, x2, …, xn> | P(x1, x2, …, xn) }`  
where each `xi` is a domain variable.

**Example:** The “instructor name for CS department” query:  
`{ <n> | ∃ i, s, d ( instructor(i, n, d, s) ∧ d = 'Comp. Sci.' ) }`  
Here `i`, `n`, `d`, `s` correspond to ID, name, dept, salary.

DRC and TRC are equivalent in expressive power and equally equivalent to relational algebra.

### 2.4 Safety of Expressions

A calculus expression is **safe** if its result is guaranteed to be finite and depends only on the actual contents of the database. For example,  
`{ t | ¬ (t ∈ instructor) }`  
is unsafe because it would yield an infinite set (all possible tuples not in instructor). SQL and practical query systems restrict expressions to safe ones.

### 2.5 Equivalence of Algebra and Calculus

Every relational algebra expression can be translated into a tuple/domain calculus expression, and vice versa. For instance:

- Selection `σ_{A=17}(r)` in TRC: `{ t | t ∈ r ∧ t.A = 17 }`
- Projection `π_{A,B}(r)` in TRC: `{ t | ∃ u ∈ r ( t.A = u.A ∧ t.B = u.B ) }`
- Union `r ∪ s`: `{ t | t ∈ r ∨ t ∈ s }`
- Difference `r − s`: `{ t | t ∈ r ∧ ¬(t ∈ s) }`
- Cartesian product `r × s`: `{ t | ∃ u ∈ r, ∃ v ∈ s ( t.A1 = u.A1 ∧ … ) }` (with appropriate renaming)

The formal proofs construct the translation scheme. This equivalence assures that any query expressible in one formalism can be expressed in the others; they form the theoretical foundation of relational databases.

## 3. Entity‑Relationship Model (Modules 18–20)

The E‑R model is the dominant conceptual design tool for databases. It captures the real‑world enterprise in terms of **entities**, **attributes**, and **relationships**, independent of implementation details.

### 3.1 Design Process Overview

1. **Requirement Analysis**: Understand functional specifications, stakeholders, business rules.
2. **Conceptual Design**: Produce a high‑level description using the E‑R model.
3. **Logical Design**: Transform the E‑R diagram into a relational schema (tables).
4. **Physical Design**: Decide on storage structures, indexes, etc.

We focus on the conceptual → logical steps.

### 3.2 Core Concepts

#### 3.2.1 Entities and Entity Sets

An **entity** is a distinguishable object in the real world (e.g., the instructor “Einstein”). An **entity set** is a collection of similar entities sharing the same attributes. In E‑R diagrams:
- **Rectangle** represents an entity set.
- **Attributes** are listed inside the rectangle (or connected via lines). The **primary key** is underlined.

#### 3.2.2 Attributes

- **Simple**: atomic value (integer, string, date).
- **Composite**: composed of sub‑parts (e.g., name: first, middle, last). Shown by indenting sub‑attributes.
- **Single‑valued**: only one value per entity (default).
- **Multi‑valued**: may have multiple values (e.g., phone numbers). Shown by `{ }` around the attribute.
- **Derived**: value computed from other attributes (e.g., age from dateOfBirth). Shown with parentheses `( )`.
- **Domain**: set of permitted values.

#### 3.2.3 Relationships and Relationship Sets

A **relationship** is an association among two or more entities. A **relationship set** is a set of such associations. Represented by a **diamond** connected to the participating entity sets.

- **Degree**: number of entity sets involved (binary, ternary, etc.). Binary relationships are most common.
- **Roles**: when the same entity set participates more than once, role names distinguish them (e.g., prerequisite `course_id` vs `prereq_id`).
- **Descriptive attributes**: attributes that belong to the relationship itself (e.g., `date` of advising). Shown with a **dashed line** from the diamond.

#### 3.2.4 Cardinality Constraints

They specify how many entities from one set can be associated with entities of another.

- **One‑to‑one (1:1)**: An entity in A is associated with at most one entity in B, and vice versa.
- **One‑to‑many (1:N)**: A related to many B, but B related to at most one A.
- **Many‑to‑one (N:1)**: reverse of 1:N.
- **Many‑to‑many (M:N)**: both sides can have multiple associations.

In E‑R diagrams, an **arrow** pointing to an entity set indicates a **“one”** side; no arrow means “many”. A **double line** indicates **total participation** (every entity in the set must participate in the relationship), while a single line indicates **partial participation**.

Alternative notation: explicit cardinality limits like `1..1` (one and only one), `0..*` (zero or more).

### 3.3 Weak Entity Sets

A **weak entity set** does not have a primary key by itself; it depends on a **strong (owner) entity set** for identification.

- Discriminator (partial key): set of attributes that can distinguish entities within the context of the owner.
- The primary key of the weak entity = primary key of the owner + discriminator.
- The relationship connecting the weak entity to its owner is called the **identifying relationship**.
- Weak entity set: **double‑lined rectangle**. Discriminator: **dashed underline**.
- Identifying relationship: **double‑lined diamond**.
- Total participation of the weak entity in the identifying relationship is mandatory (double line connecting them).

**Example:** `section` is a weak entity set depending on `course`. `section(sec_id, semester, year)` does not uniquely identify a section across all courses. But `course_id + sec_id + semester + year` is unique. The identifying relationship `sec_course` links them.

### 3.4 Extended E‑R Features

#### 3.4.1 Specialization / Generalization (ISA)

Represents an inheritance hierarchy among entity sets.

- **Specialization**: top‑down, a higher‑level entity set is split into lower‑level entity sets with distinct additional attributes or relationships.
- **Generalization**: bottom‑up, common attributes of several entity sets are factored into a higher‑level entity set.
- Denoted by a **triangle** labeled “ISA”. The higher‑level set is the superclass, lower‑level sets are subclasses.
- **Inheritance**: subclasses inherit all attributes and relationships of the superclass.
- **Constraints**:
  - **Overlapping / Disjoint**: can an entity belong to more than one subclass? (e.g., an employee can be a student → overlapping; instructor vs secretary → disjoint)
  - **Total / Partial**: must every entity in the superclass belong to at least one subclass? (total: every student must be UG or PG; partial: a person may be neither employee nor student)

**Example:** `person` superclass, `employee` and `student` as subclasses. `employee` further specializes into `instructor` and `secretary`.

#### 3.4.2 Aggregation

Aggregation treats a **relationship set as an abstract entity set**, enabling it to participate in further relationships. This eliminates redundancy when a ternary (or higher) relationship would otherwise duplicate an existing relationship.

**Example:** `project_guide(project, instructor, student)` is a ternary relationship. A separate `eval_for` evaluation relationship should be linked to the exact same (project, instructor, student) combination, not just any project. By aggregating `project_guide` into an entity, we can directly connect `eval_for` to it.

Diagram: draw a dashed rectangle around the relationship and participating entity sets, then link the new relationship diamond to this rectangle.

### 3.5 E‑R Diagram Symbols (Summary)

- **Entity set**: rectangle (strong: single line; weak: double line)
- **Attribute**: ellipse or plain text; primary key underlined, discriminator dashed underline, derived `(…)`, multivalued `{…}`
- **Relationship**: diamond (identifying: double line)
- **Lines**: connect entity to relationship; double line = total participation; arrow = “one” side
- **ISA**: triangle
- **Aggregation**: dashed rectangle around relationship + entities

### 3.6 Reducing E‑R Diagrams to Relational Schemas

The transformation from conceptual to logical (tables) follows a set of standard rules.

#### 3.6.1 Strong Entity Set

Create a relation with the same name. Attributes become columns. The primary key of the entity set becomes the primary key of the relation.

#### 3.6.2 Weak Entity Set

Create a relation with the discriminator attributes plus the primary key of the owner entity set. The primary key of this relation is (owner’s primary key) ∪ (discriminator). Also include any other attributes. The foreign key referencing the owner is part of the primary key.

**Example:** `section(course_id, sec_id, semester, year, …)` where `course_id` references `course`.

#### 3.6.3 Relationships

- For a binary **many‑to‑many** relationship: create a separate relation containing the primary keys of both participating entity sets and any descriptive attributes. The primary key is the union of the foreign keys.
- For a binary **one‑to‑many** or **many‑to‑one** relationship, especially if it is **total** on the many side, we can **merge** the relationship into the “many” side relation by adding the primary key of the “one” side as a foreign key. This avoids an extra join.
- For **one‑to‑one** relationships, either side can take the foreign key; choose based on participation constraints or natural semantics.
- For **n‑ary relationships** (degree > 2), create a separate relation with foreign keys to all participating entity sets, plus descriptive attributes. The primary key is usually the combination of all foreign keys unless constraints allow a smaller subset.

#### 3.6.4 Composite Attributes

Flatten into simple attributes (e.g., `first_name`, `middle_name`, `last_name` instead of a single `name`). Alternatively, create a separate entity set for the composite attribute, but flattening is simpler and usually preferred.

#### 3.6.5 Multivalued Attributes

Create a separate relation with the entity’s primary key and the multivalued attribute. The primary key of this new relation is the combination. This normalizes the schema.

**Example:** `inst_phone(ID, phone_number)`, where `ID` references `instructor`.

#### 3.6.6 Specialization/Generalization (ISA)

Two common approaches:

1. **Separate tables for superclass and subclasses**:  
   - Superclass table holds common attributes, primary key.  
   - Each subclass table holds its specific attributes, with the same primary key (foreign key referencing superclass).  
   - Queries that need combined data require joins.
2. **Flattened subclass tables**:  
   - For each subclass, create a table containing **all** inherited attributes plus its own.  
   - Eliminates joins but can duplicate common data, causing redundancy and potential update anomalies if overlapping subclasses.

Choice depends on performance requirements and the degree of overlap.

#### 3.6.7 Aggregation

Create a relation for the aggregated relationship (its primary key is the combination of the foreign keys of the involved entity sets). Then create a relation for the higher‑level relationship, which includes the primary key of the aggregated relation as a foreign key and any additional attributes.

### 3.7 Design Issues and Trade‑offs

#### 3.7.1 Attribute vs. Entity Set

If a property is multi‑valued, it **must** be modeled as a separate entity set (or weak entity). Even for single‑valued properties, if it has a complex structure and participates in many relationships on its own, promoting it to an entity can be beneficial.

#### 3.7.2 Entity vs. Relationship

Not always clear‑cut. A relationship set can sometimes be modeled as an entity set, especially if it has its own attributes and participates in further relationships. The “project‑guide” example shows when aggregation is needed; alternatively, one could introduce a weak entity set directly.

#### 3.7.3 Binary vs. Non‑binary Relationships

Most relationships are binary; ternary (or higher) relationships can often be decomposed into binary ones by introducing an artificial entity set (e.g., a weak entity with a surrogate key). However, constraints that naturally involve all three entities may be lost or harder to enforce. Decomposition is valid only if it is **lossless** and preserves all functional dependencies and constraints.

#### 3.7.4 Strong vs. Weak Entity Sets

If an entity has a natural global identifier, make it strong. Weak entities are convenient for dependent objects (e.g., sections of a course, rooms in a building). They enforce existence dependency and composite keys.

#### 3.7.5 Use of Specialization/Generalization

Improves modularity and clarity. Clearly separate properties of subclasses. However, mapping to relational tables adds complexity (joins vs. duplication). The designer must decide based on query patterns and update frequency.

#### 3.7.6 Use of Aggregation

Use aggregation when a relationship itself is the subject of another relationship. It eliminates redundancy and keeps the model faithful to the real‑world semantics.

### 3.8 Summary of the E‑R Design Workflow

1. Identify entity sets and their attributes.
2. Determine primary keys for strong entities; discriminators for weak entities.
3. Identify relationship sets and their degrees.
4. Specify cardinality constraints and participation (total/partial).
5. Draw the E‑R diagram using standard notation.
6. Review for design issues; refine (introduce specialization, aggregation if needed).
7. Translate to relational schemas following the mapping rules.
8. Apply normalization (next topic) to eliminate remaining anomalies.

The E‑R model thus bridges the gap between the informal real‑world requirements and the rigorous relational tables that SQL can implement.

By mastering relational algebra, calculi, and the E‑R model, you gain both the theoretical underpinnings and the practical design skills needed to build robust, efficient database applications. The subsequent modules will build on this foundation, moving into normalization, physical storage, and transaction management.

# Entity-Relationship (ER) Diagrams Explained in Perfect Clarity

An Entity-Relationship diagram (ER diagram or ERD) is a visual blueprint for the data that an application needs to store. It shows what **entities** (things or objects) exist, what **attributes** (properties) they have, and how they are **related** to each other. It is not tied to any specific database software; it is a conceptual tool used to design a database before writing a single line of code.

Think of an ER diagram as a map of your data. Just as an architect draws a floor plan before building a house, a database designer draws an ER diagram before creating tables.

---

## The Core Components of an ER Diagram

### 1. Entities
An entity is a thing, person, place, or concept about which you want to store data. In a diagram, an entity is usually drawn as a **rectangle** with the entity name inside.

*Examples from the lecture:*  
- **Customer**  
- **Order**  
- **Shipment**  
- **Student** (from the gradebook example)  
- **Course**

Each entity will eventually become a **table** in the database. A single occurrence of an entity (e.g., one particular customer named "Alice") will become a **row** in that table.

### 2. Attributes
Attributes are the pieces of information that describe an entity. They are typically listed inside the entity rectangle, below the entity name, separated by a line.

*Example for the **Customer** entity:*  
```
Customer
---------
customer_id   PK
customer_name
```
Here, `customer_id` and `customer_name` are attributes. Attributes become **columns** in the database table.

**Primary Key (PK):** One special attribute (or combination of attributes) that uniquely identifies each entity instance. It is often marked with the label **PK** or by underlining. In the example, `customer_id` is the primary key because no two customers share the same ID, even if they happen to have the same name.

Other constraints on attributes (like "cannot be null" or "maximum 50 characters") may be noted as well, but the key idea is that the ER diagram captures what data is stored for each entity.

### 3. Relationships
A relationship is an association between two entities. It is drawn as a **line** connecting the two entity rectangles. The line often has a diamond-shaped label that names the relationship, though in many modern notations (like Crow's Foot) the label is simply written next to the line.

*Examples from the lecture:*  
- A **Customer** *places* an **Order**.
- An **Order** *has* **Shipment(s)**.
- A **Student** *enrolls in* a **Course** (many-to-many, requiring a junction table).

Relationships tell us how to link tables together using **foreign keys**.

### 4. Cardinality (Multiplicity)
This is the most crucial part of the diagram. Cardinality specifies **how many** instances of one entity can be associated with instances of the other entity. It is shown by special symbols at the ends of the relationship line.

The lecture focuses on the **Crow's Foot notation**, one of the most widely used.

---

## Crow's Foot Notation Symbols Explained

At each end of a relationship line, you will see two parts placed next to each other: an **inner symbol** (closer to the entity) and an **outer symbol** (farther from the entity).

### The inner symbol – "one or many"
- A single vertical line `|` means **one**.
- A three-pronged "crow's foot" (like a little fork with three lines) means **many**.

### The outer symbol – "mandatory or optional"
- A second vertical line next to the inner line `||` means **exactly one** (mandatory one).
- A circle `O` next to the symbol means **zero** (optional).
- If only the inner line appears, it often implies "one" (the exact mandatory/optional may be clarified by the presence or absence of the circle).

By combining these, we get the four most common cardinalities:

| Symbol at end of line | Meaning |
|----------------------|---------|
| `||` (two vertical lines) | Exactly one (mandatory one) |
| `O|` (circle then vertical line) | Zero or one (optional one) |
| `|<` or `O<`? Actually, the crow's foot is often drawn with a circle before it for optional many, or just the foot for "many" (which could mean "one or many"). In the lecture, they used: `O<` (circle then crow's foot) meaning "zero or many", and just `<` meaning "one or many" (mandatory many). | Zero or many (optional many) |
| `|<` (vertical line then crow's foot) | One or many (mandatory many) |

Let's clarify the exact notation as presented in the lecture:

- The lecture describes the line crossing the main line (single line) indicating "one". That is the inner `|`.
- The three-pronged "crow's foot" indicates "many".
- The addition of a circle `O` before the symbol indicates optionality (zero allowed).

So:
- `|` at the end of a line (just a vertical tick) = "one".
- `<` (crow's foot) = "many".
- `O|` = "zero or one".
- `O<` = "zero or many".
- `||` = "exactly one" (often drawn as two parallel vertical lines).
- `|<` = "one or many".

In practice, the symbols are drawn directly on the relationship line, right where it touches the entity.

## Step-by-Step Example: E-Commerce System

Let's reconstruct the ER diagram from the lecture that models a simplified e-commerce system with Customers, Orders, and Shipments.

### Entity: Customer
Attributes:
- `customer_id` (Primary Key, integer, not null)
- `customer_name` (string, length up to 50)

### Entity: Order
Attributes:
- `order_id` (Primary Key)
- `order_date` (date)
- `customer_id` (Foreign Key referencing Customer)

### Entity: Shipment
Attributes:
- `shipment_id` (Primary Key)
- `order_id` (Foreign Key referencing Order)

### Relationships and Cardinalities

**Relationship between Customer and Order:**
- One **Customer** can place **zero or many** Orders.  
  Symbol at the Customer end: `O<` (circle + crow's foot), meaning "zero or many".  
  Why? A customer may register but never buy anything (0 orders), or buy several times (many orders).

- One **Order** must be associated with **exactly one** Customer.  
  Symbol at the Order end: `||` (two vertical lines), meaning "exactly one".  
  Why? An order cannot exist without a customer. It is mandatory.

So the line connecting Customer and Order will have `O<` near Customer and `||` near Order.

**Relationship between Order and Shipment:**
- One **Order** can have **zero or many** Shipments.  
  Symbol at the Order end: `O<`.  
  Why? An order might not yet be shipped (0), or it might be split into multiple shipments (many).

- One **Shipment** must belong to **exactly one** Order.  
  Symbol at the Shipment end: `||`.  
  Why? Every shipment is for one specific order.

## How to Read an ER Diagram

Reading from left to right is common, but the diagram is bidirectional. For the Customer–Order relationship:

- **Customer → Order:** "A Customer places zero, one, or many Orders."
- **Order → Customer:** "An Order is placed by exactly one Customer."

The symbol closest to an entity tells you about that entity's participation in the relationship.

## How ER Diagrams Translate to Database Tables

The purpose of an ER diagram is to create a database schema.

1. **Each entity becomes a table.**  
   - `Customer` table with columns: `customer_id` (PK), `customer_name`  
   - `Order` table with columns: `order_id` (PK), `order_date`, `customer_id` (FK)  
   - `Shipment` table with columns: `shipment_id` (PK), `order_id` (FK)

2. **Primary keys** are placed in the entity's table.

3. **One-to-many relationships** are implemented by putting a **foreign key** in the table on the "many" side that refers to the primary key of the "one" side.  
   - Because one Customer has many Orders, the `Order` table gets a `customer_id` column that points to `Customer.customer_id`.  
   - Because one Order has many Shipments, the `Shipment` table gets an `order_id` column that points to `Order.order_id`.

4. **Mandatory vs optional** are enforced by constraints:  
   - The `customer_id` in the `Order` table is set to `NOT NULL` (mandatory).  
   - The `order_id` in the `Shipment` table is set to `NOT NULL` (mandatory).  
   - The optionality on the "one" side (Customer may have 0 orders) does not require a special constraint; it simply means the foreign key column in Order is not required to reference every possible Customer.

5. **Many-to-many relationships** (not in this example, but in the gradebook's Student–Course relationship) require a separate **junction table**.  
   - For Student and Course, you create a table `student_courses` with columns: `student_id` (FK to Student), `course_id` (FK to Course), and optionally `marks`. The primary key of this table is usually the combination `(student_id, course_id)`.

## Why Are ER Diagrams So Important?

- **Communication:** They provide a clear, unambiguous picture that both technical and non-technical stakeholders can understand.
- **Design before implementation:** Mistakes in the data model are extremely costly to fix later. An ER diagram lets you think through the structure upfront.
- **Blueprint for the database:** Once the diagram is correct, writing the SQL `CREATE TABLE` statements is a mechanical task.
- **Documentation:** They serve as living documentation of your database schema.

The Crow's Foot notation, with its visual combination of lines, circles, and forks, allows you to grasp the logic of the data instantly: a circle says "might not exist," a straight tick says "exactly one," and the crow's foot says "multiple." Once you internalize these symbols, reading and creating ER diagrams becomes a powerful skill for designing any data-driven application.
